## 1️⃣ ¿Qué es una cola?

Una **cola** (*Queue*) es una estructura de datos lineal que sigue el principio **FIFO** (*First In, First Out*): el primer elemento que entra es el primero en salir.

Es exactamente la fila de un banco o de un supermercado: la persona que llegó primero es la primera en ser atendida, sin importar cuánta gente llegue detrás.

> 🔎 **Frase para quedártela:** *"El que llegó primero, se va primero. Todos los demás esperan su turno en orden."*

### ⚙️ Operaciones principales

| Operación | Qué hace |
|---|---|
| `enqueue(elemento)` | Agrega un elemento al **final** de la cola. |
| `dequeue()` | Elimina y devuelve el elemento del **frente**. Si está vacía, lanza una excepción. |
| `peek()` | Consulta el elemento del frente **sin** eliminarlo. Si está vacía, lanza una excepción. |
| `is_empty()` / `len()` | Indica si la cola está vacía / cuántos elementos tiene. |

**Dos extremos, dos roles fijos:** a diferencia de la pila (un solo extremo activo), la cola usa **frente** (por donde sale) y **final** (por donde entra) — y nunca se cambian los roles.

### 🆚 Cola vs. pila — la comparación que más te va a servir este semestre

| Característica | Pila (LIFO) | Cola (FIFO) |
|---|---|---|
| ¿Qué sale primero? | El último que entró | El primero que entró |
| Inserción | `push`, en la cima | `enqueue`, al final |
| Eliminación | `pop`, de la cima | `dequeue`, del frente |
| Ejemplo cotidiano | Torre de platos | Fila del banco |
| Se usa cuando... | El orden a procesar es el **inverso** al de llegada | El orden a procesar es el **mismo** que el de llegada |

### 🏗 Casos de uso reales

1. **Sistemas de impresión** — los documentos se imprimen en el orden en que se enviaron.
2. **Gestión de procesos** — un sistema operativo atiende procesos en la CPU siguiendo el orden en que llegaron a la cola de ejecución.
3. **Atención al cliente / líneas de espera** — call centers, bancos, cualquier sistema de "turnos".
4. **BFS (Breadth-First Search)** — recorrer un grafo o árbol **nivel por nivel**, visitando primero todo lo que está a distancia 1, luego todo lo que está a distancia 2, etc.
5. **Buffers de datos** — manejo de paquetes en una red, streaming de datos: se procesa en el mismo orden en que llegó.

| Señal en el enunciado | Por qué apunta a una cola |

## 3️⃣ ✏️ El hábito: traza el estado antes de escribir código

Mismo hábito que instalamos con pilas, aplicado a colas: antes de tocar el teclado, simula la cola a mano para una entrada pequeña.

### Ejemplo trabajado: fila de un restaurante de comida rápida

**Escenario:** un cajero atiende clientes en el orden en que llegaron a la fila. Vamos a trazar la secuencia de operaciones: llegan **Laura**, **Marcos** y **Sofía** (en ese orden); el cajero **atiende** a un cliente (`dequeue`); llega **Andrés**; el cajero consulta quién sigue sin atenderlo todavía (`peek`); el cajero **atiende** dos veces más.

| Operación | Frente → Final tras la operación | Valor devuelto |
|---|---|---|
| `enqueue("Laura")` | `Laura` | — |
| `enqueue("Marcos")` | `Laura, Marcos` | — |
| `enqueue("Sofía")` | `Laura, Marcos, Sofía` | — |
| `dequeue()` | `Marcos, Sofía` | `"Laura"` |
| `enqueue("Andrés")` | `Marcos, Sofía, Andrés` | — |
| `peek()` | `Marcos, Sofía, Andrés` (sin cambios) | `"Marcos"` |
| `dequeue()` | `Sofía, Andrés` | `"Marcos"` |
| `dequeue()` | `Andrés` | `"Sofía"` |

Fíjate en algo clave de la traza: `peek()` **consulta pero no modifica** el estado — por eso la fila queda igual en esa fila de la tabla. Es un error común confundir `peek()` con `dequeue()` al programar; la traza a mano te obliga a ser explícito sobre cuál de los dos estás usando en cada paso.

👉 Practica con esta misma tabla, en papel, antes de programar cualquier ejercicio de esta unidad: qué operación ocurre → cómo queda la cola → qué devuelve (si devuelve algo).

```python
from base import Queue

fila = Queue()
fila.enqueue("Laura")
fila.enqueue("Marcos")
fila.enqueue("Sofía")
print(fila)              # ['Laura', 'Marcos', 'Sofía']

print(fila.dequeue())    # 'Laura'
fila.enqueue("Andrés")
print(fila)              # ['Marcos', 'Sofía', 'Andrés']

print(fila.peek())       # 'Marcos'  -- consulta, no modifica
print(fila)              # ['Marcos', 'Sofía', 'Andrés']  -- sigue igual

print(fila.dequeue())    # 'Marcos'
print(fila.dequeue())    # 'Sofía'
print(fila)              # ['Andrés']
```

## 4️⃣ 🏗️ Implementando el ADT `Queue`

Igual que con la pila, definimos el contrato (`enqueue`, `dequeue`, `peek`) y elegimos cómo representarla por dentro. Acá la decisión importa **más** que en la pila, porque las dos representaciones típicas no tienen el mismo costo:

| Representación | `enqueue()` | `dequeue()` | ¿Por qué? |
|---|---|---|---|
| Lista de Python, agregando al final y sacando del frente (`list.append` + `list.pop(0)`) | O(1) amortizado | **O(n)** | `pop(0)` tiene que **recorrer y correr una posición** cada uno de los elementos restantes para tapar el hueco que deja el primero |
| Lista enlazada, con referencia a `head` (frente) y `tail` (final) | O(1) | O(1) | Ni agregar al final ni quitar del frente requieren recorrer nada — son solo actualizaciones de punteros |

> ⚠️ **Ojo con esto — es una pregunta típica de parcial:** la `Queue` de `base.py` de este curso está implementada sobre una lista de Python (por simplicidad), así que su `dequeue()` es **O(n)**, no O(1). Si te preguntan por la complejidad de `dequeue()` en esa implementación, la respuesta correcta es **O(n)**. La versión con lista enlazada de abajo sí logra O(1) en ambas operaciones — es la representación que usarías si la eficiencia de `dequeue()` importa de verdad.

### Representación 1: arreglo dinámico

```python
from typing import Any


class EmptyQueue(Exception):
    ...


class QueueArray:
    def __init__(self) -> None:
        self.__items: list[Any] = []

    def enqueue(self, element: Any) -> None:
        self.__items.append(element)

    def dequeue(self) -> Any:  # O(n): pop(0) recorre y recorre la lista
        if len(self.__items) == 0:
            raise EmptyQueue("Cola vacía... no podés hacer dequeue")
        return self.__items.pop(0)

    def peek(self) -> Any:
        if len(self.__items) == 0:
            raise EmptyQueue("Cola vacía... no podés hacer peek")
        return self.__items[0]

    def __len__(self) -> int:
        return len(self.__items)

    def __repr__(self) -> str:
        return str(self.__items)


q = QueueArray()
q.enqueue(5)
q.enqueue(4)
q.enqueue(1)
print(q)            # [5, 4, 1]
print(q.dequeue())  # 5
print(q)            # [4, 1]
```

### Representación 2: lista enlazada

Aquí el `frente` de la cola coincide con la `head` de la lista enlazada, y el `final` con la `tail`. `enqueue` agrega en la `tail` (O(1)), `dequeue` quita de la `head` (O(1)) — sin desplazar nada.

```python
from typing import Any


class Node:
    def __init__(self, value: Any, next: "Node | None" = None) -> None:
        self.value = value
        self.next = next

    def __repr__(self) -> str:
        return f"{self.value} -> {self.next}"


class LinkedList:
    def __init__(self) -> None:
        self.head: "Node | None" = None
        self.tail: "Node | None" = None
        self.size: int = 0

    def append(self, value: Any) -> None:  # agrega en la tail -- O(1)
        if self.head is None:
            self.head = Node(value)
            self.tail = self.head
        else:
            self.tail.next = Node(value)
            self.tail = self.tail.next
        self.size += 1

    def delete_and_return_first(self) -> Any:  # quita de la head -- O(1)
        if self.head is None:
            return None
        old_head = self.head
        self.head = self.head.next
        if self.head is None:  # la lista quedó vacía
            self.tail = None
        old_head.next = None
        self.size -= 1
        return old_head.value

    def __len__(self) -> int:
        return self.size

    def __repr__(self) -> str:
        return f"{self.head}"


class QueueLinkedList:
    def __init__(self) -> None:
        self.__items = LinkedList()

    def enqueue(self, element: Any) -> None:
        self.__items.append(element)

    def dequeue(self) -> Any:
        if len(self.__items) == 0:
            raise EmptyQueue("Cola vacía... no podés hacer dequeue")
        return self.__items.delete_and_return_first()

    def peek(self) -> Any:
        if len(self.__items) == 0:
            raise EmptyQueue("Cola vacía... no podés hacer peek")
        return self.__items.head.value

    def __len__(self) -> int:
        return len(self.__items)

    def __repr__(self) -> str:
        return str(self.__items)


q = QueueLinkedList()
q.enqueue(5)
q.enqueue(4)
q.enqueue(1)
print(q)            # 5 -> 4 -> 1 -> None
print(q.dequeue())  # 5
print(q)            # 4 -> 1 -> None
```

### Traza rápida: cola de prioridad tipo `"min"` (el menor número = mayor urgencia)

| Operación | Contenido (ordenado, el primero es el próximo en salir) |
|---|---|
| `enqueue(10)` | `[10]` |
| `enqueue(3)` | `[3, 10]` |
| `enqueue(7)` | `[3, 7, 10]` |
| `dequeue()` → `3` | `[7, 10]` |
| `enqueue(1)` | `[1, 7, 10]` |
| `dequeue()` → `1` | `[7, 10]` |

Fíjate que el orden en que salen los números (`3, 1, ...`) **no tiene nada que ver** con el orden en que entraron (`10, 3, 7, 1`) — esa es justamente la diferencia con una cola normal.

### Implementación

```python
from typing import Any


class PriorityQueueList:
    """Cola de prioridad sobre una lista ordenada de Python."""

    def __init__(self, priority: str) -> None:
        self.__items: list[Any] = []
        self.__priority: str = priority  # "min" o "max"

    def enqueue(self, element: Any) -> None:
        if self.__priority == "min":
            self.__items.append(element)
            self.__items.sort()
        elif self.__priority == "max":
            self.__items.append(element)
            self.__items.sort(reverse=True)
        else:
            raise ValueError("Grave... este tipo de cola no existe...")

    def dequeue(self) -> Any:
        if len(self.__items) == 0:
            raise EmptyQueue("Cola vacía... no podés hacer dequeue")
        return self.__items.pop(0)

    def peek(self) -> Any:
        if len(self.__items) == 0:
            raise EmptyQueue("Cola vacía... no podés hacer peek")
        return self.__items[0]

    def __repr__(self) -> str:
        return str(self.__items)


pq = PriorityQueueList("min")
pq.enqueue(10)
pq.enqueue(3)
pq.enqueue(7)
print(pq)            # [3, 7, 10]
print(pq.dequeue())  # 3
pq.enqueue(1)
print(pq)            # [1, 7, 10]
print(pq.dequeue())  # 1
```

## 5️⃣ 🚑 Cola de prioridad (Priority Queue)

### La señal que la distingue de una cola simple

> 🔎 **Frase para quedártela:** *"Siempre necesito el mínimo (o el máximo) disponible en este momento, sin importar en qué orden fue llegando."*

Una cola de prioridad es una variación de la cola donde **cada elemento tiene una prioridad**, y **la prioridad decide el orden de salida** — no el orden de llegada. Piensa en una sala de urgencias: un paciente con un infarto se atiende antes que uno con un dolor de cabeza, sin importar quién llegó primero al hospital.

| | Cola simple | Cola de prioridad |
|---|---|---|
| ¿Qué determina quién sale? | Orden de llegada | El valor de prioridad |
| ¿Cuándo la usas? | "en el orden en que llegaron" | "siempre el más urgente/mayor/menor primero" |

### ¿Cómo se representa por dentro?

| Representación | `enqueue()` | `dequeue()` | ¿Cuándo conviene? |
|---|---|---|---|
| Lista desordenada | O(1) | O(n) — hay que **buscar** el de mayor prioridad | Pocas eliminaciones, muchas inserciones |
| Lista ordenada (se inserta ya en su lugar) | O(n) — hay que **encontrar dónde insertar** | O(1) — ya está al frente | Pocas inserciones, muchas eliminaciones |
| Montículo (*heap*, estructura de árbol) | O(log n) | O(log n) | El mejor balance — lo vas a ver formalmente en `07_arboles` |

La implementación que vas a usar en este curso (`base.py`) usa una **lista ordenada**: cada `enqueue` inserta y vuelve a ordenar, así que `dequeue`/`peek` son inmediatos. Es una decisión de simplicidad, no la más eficiente posible — y es importante que sepas justificar ese costo si te lo preguntan.
