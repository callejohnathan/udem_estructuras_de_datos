## 1️⃣ ¿Qué es una pila?

Una **pila** (*Stack*) es una estructura de datos lineal que sigue el principio **LIFO** (*Last In, First Out*): el último elemento que entra es el primero en salir.

Imagínala como una torre de platos: solo puedes agregar un plato encima de la torre, y solo puedes quitar el que está más arriba. No puedes sacar el plato de la mitad sin antes quitar todos los que están encima.

> 🔎 **Frase para quedártela:** *"Solo puedo tocar la cima. Todo lo que está debajo tiene que esperar su turno."*

### ⚙️ Operaciones principales

| Operación | Qué hace |
|---|---|
| `push(elemento)` | Agrega un elemento en la cima de la pila. |
| `pop()` | Elimina y devuelve el elemento en la cima. Si está vacía, lanza una excepción. |
| `peek()` | Consulta el elemento en la cima **sin** eliminarlo. Si está vacía, lanza una excepción. |
| `is_empty()` / `len()` | Indica si la pila está vacía / cuántos elementos tiene. |

**Acceso restringido:** a diferencia de una lista, no puedes leer ni modificar el elemento en la posición 3 directamente — solo existe "la cima". Esa restricción es justamente lo que hace útil a la pila: te obliga (y le permite al algoritmo) a procesar las cosas en un orden muy específico.

### 🏗 Casos de uso reales

1. **Pila de llamadas de funciones** — cuando una función invoca a otra, la que fue llamada más recientemente es la primera en terminar y devolver el control.
2. **Deshacer / rehacer (Undo/Redo)** — editores de texto y navegadores apilan tus acciones; deshacer siempre revierte la más reciente primero.
3. **Verificación de anidamiento** — paréntesis, corchetes, etiquetas HTML: lo último que se abrió es lo primero que se debe cerrar.
4. **Historial de navegación** — el botón "atrás" de un navegador te lleva a la página que visitaste más recientemente.
5. **DFS (Depth-First Search)** — recorrer un grafo o árbol "yendo tan profundo como se pueda" antes de retroceder, ya sea con la pila de llamadas recursiva o con una pila explícita.

### 🆚 Pila vs. lista

| Característica | Lista | Pila |
|---|---|---|
| Acceso | A cualquier posición (`lista[i]`) | Solo a la cima |
| Inserción | En cualquier posición | Solo en la cima (`push`) |
| Eliminación | En cualquier posición | Solo en la cima (`pop`) |
| ¿Por qué usarla? | Cuando necesitas acceso flexible | Cuando el **orden LIFO** es parte del problema, y quieres que la estructura misma te lo garantice |

## 2️⃣ 🧠 Reconociendo la señal LIFO en el enunciado

Esta es la parte más importante de la clase — **más que la implementación**.

La falla más común no es "no saber programar una pila". Es no darse cuenta de que un problema *necesita* una pila hasta después de haber perdido 20 minutos intentando resolverlo con una lista y variables sueltas. Antes de escribir código, entrena el ojo para detectar ciertas palabras y patrones en el enunciado:

| Señal en el enunciado | Por qué apunta a una pila |
|---|---|
| "deshacer la última acción", "el cambio más reciente" | Necesitas revertir en el orden **inverso** al que ocurrieron las cosas → LIFO |
| "verificar que esté bien anidado/balanceado" (paréntesis, etiquetas) | Lo último que se abrió debe ser lo primero que se cierre → LIFO |
| "el más reciente primero", "el último en llegar" | Es la definición misma de LIFO |
| "simular una función recursiva sin recursión" | La recursión ya usa una pila internamente (la pila de llamadas); si la simulas a mano, necesitas tu propia pila |
| "explorar tan profundo como se pueda antes de retroceder" (DFS) | Tienes que "recordar" el camino para poder devolverte por el mismo, en orden inverso |
| "revertir", "invertir el orden de entrada" | Sacar todo de una pila siempre te da el orden invertido de cómo entró |

⚠️ **Contraejemplo importante:** que un problema mencione "orden" no significa automáticamente pila. Si el enunciado dice *"procesar en el orden en que llegaron"* o *"por turnos, el primero que llegó es el primero atendido"*, **eso NO es LIFO** — es el patrón contrario (lo vas a ver en la próxima clase). La señal que buscas específicamente es: *¿lo que se resuelve o revisa primero es siempre lo último que entró?*

## 3️⃣ ✏️ El hábito: traza el estado antes de escribir código

Ya reconociste que un problema necesita una pila. **Todavía no abras el editor.** El siguiente paso es simular la pila a mano (en papel, en una tabla, como sea) para una entrada pequeña, y ver qué pasa en cada paso. Si puedes trazarla a mano, programarla es solo traducir esa traza a código — línea por línea. Si no puedes trazarla a mano, tampoco vas a poder programarla, y es mucho más barato descubrir eso *antes* de escribir código que después.

### Ejemplo trabajado: ¿está bien balanceada la expresión?

**Enunciado:** dada una cadena con los símbolos `( ) [ ] { }`, determina si está correctamente balanceada (cada símbolo de apertura tiene su cierre correspondiente, y en el orden correcto).

**Paso 1 — reconocer la señal:** "correctamente balanceada" + símbolos de apertura/cierre anidados → es la señal de pila que ya vimos.

**Paso 2 — decidir la regla, en español, antes de código:**
* Si veo un símbolo de **apertura** (`(`, `[`, `{`), lo apilo (`push`).
* Si veo un símbolo de **cierre** (`)`, `]`, `}`), reviso la cima con `peek`/`pop`: si no coincide con el cierre correspondiente (o la pila ya está vacía), la expresión **no** está balanceada.
* Al final, si la pila quedó vacía, todo cerró correctamente. Si quedó algo sin cerrar, **no** está balanceada.

**Paso 3 — trazar a mano** con la entrada `"( [ { } ] )"` (sin espacios: `"([{}])"`), carácter por carácter:

| Carácter leído | Acción | Estado de la pila (cima a la derecha) | ¿Balance roto hasta ahora? |
|---|---|---|---|
| `(` | apertura → `push` | `(` | No |
| `[` | apertura → `push` | `( [` | No |
| `{` | apertura → `push` | `( [ {` | No |
| `}` | cierre → `pop`, compara con `{` | `( [` | No (coincide) |
| `]` | cierre → `pop`, compara con `[` | `(` | No (coincide) |
| `)` | cierre → `pop`, compara con `(` | *(vacía)* | No (coincide) |
| — fin de la cadena — | pila vacía → **balanceada** ✅ | *(vacía)* | — |

Ahora prueba trazar tú mismo, en una tabla igual a esta, la entrada `"([)]"` (que **no** está balanceada) y encuentra en qué fila se rompe. Solo cuando tengas esa traza clara, sigue a la celda de código de abajo.

Esta misma tabla — "qué leo → qué hago → cómo queda la pila" — es la herramienta que quiero que uses en cada ejercicio de esta unidad, sin excepción, antes de tocar el teclado.

```python
from base import Stack, EmptyStack

CIERRES = {")": "(", "]": "[", "}": "{"}
APERTURAS = {"(", "[", "{"}


def esta_balanceada(expresion: str) -> bool:
    pila = Stack()
    for caracter in expresion:
        if caracter in APERTURAS:
            pila.push(caracter)
        elif caracter in CIERRES:
            if len(pila) == 0 or pila.pop() != CIERRES[caracter]:
                return False
    return len(pila) == 0


print(esta_balanceada("([{}])"))  # True -- la traza de la tabla de arriba
print(esta_balanceada("([)]"))    # False -- traza tú esta y confirma en qué paso falla
print(esta_balanceada("((("))     # False -- quedan símbolos sin cerrar
```

## 4️⃣ 🏗️ Implementando el ADT `Stack`

Ya sabemos **qué** debe hacer una pila y **cómo pensarla** antes de programar. Ahora construyámosla como *Tipo de Dato Abstracto*: definimos el contrato (`push`, `pop`, `peek`) y decidimos **cómo representarla internamente** — esa decisión es independiente de cómo la vamos a usar después.

Dos representaciones internas comunes:

| Representación | `push()` | `pop()` | Ventaja |
|---|---|---|---|
| Lista/arreglo dinámico de Python (`list.append` / `list.pop()`) | O(1) amortizado | O(1) | Simple, poco código |
| Lista enlazada (agregando/quitando en la cola de la lista) | O(1) | O(1) | O(1) *real* en cada operación, sin amortización ni redimensionamiento |

Ambas cumplen el mismo contrato — quien use la pila (`push`, `pop`, `peek`) no tiene por qué saber cuál usaste por dentro. Eso es exactamente el punto de un ADT.

### Representación 1: arreglo dinámico

```python
from typing import Any


class EmptyStack(Exception):
    ...


class StackArray:
    def __init__(self) -> None:
        self.__items: list[Any] = []

    def push(self, element: Any) -> None:
        self.__items.append(element)

    def pop(self) -> Any:
        if len(self.__items) == 0:
            raise EmptyStack("No puedes hacer pop... la pila está vacía")
        return self.__items.pop()

    def peek(self) -> Any:
        if len(self.__items) == 0:
            raise EmptyStack("No puedes hacer peek... la pila está vacía")
        return self.__items[-1]

    def __len__(self) -> int:
        return len(self.__items)

    def __repr__(self) -> str:
        return str(self.__items)


s = StackArray()
s.push("Plato A")
s.push("Plato B")
s.push("Plato C")
print(s)          # ['Plato A', 'Plato B', 'Plato C']  -- 'Plato C' es la cima
print(s.peek())    # 'Plato C' -- consulto sin sacar
print(s.pop())      # 'Plato C' -- lo saco
print(s)          # ['Plato A', 'Plato B']
```

### Representación 2: lista enlazada

La misma pila, pero por dentro usamos una lista enlazada en vez de un arreglo. La cima de la pila la hacemos coincidir con la **cola** (`tail`) de la lista enlazada, porque agregar/quitar ahí es O(1) sin tener que recorrer nada. (Recuerda `04_listas_enlazadas`: agregar/quitar en la **cabeza** también es O(1) — funcionaría igual de bien si la cima fuera la `head`.)

```python
from typing import Any


class Node:
    def __init__(self, value: Any, next: "Node | None" = None) -> None:
        self.value = value
        self.next = next

    def __repr__(self) -> str:
        return f"{self.value} -> {self.next}"


class LinkedList:
    def __init__(self) -> None:
        self.head: "Node | None" = None
        self.tail: "Node | None" = None
        self.size: int = 0

    def append(self, value: Any) -> None:
        if self.head is None:
            self.head = Node(value)
            self.tail = self.head
        else:
            self.tail.next = Node(value)
            self.tail = self.tail.next
        self.size += 1

    def delete_and_return_last(self) -> Any:
        if self.size == 1:
            old_tail = self.head
            self.head = None
            self.tail = None
            self.size -= 1
            return old_tail.value

        current = self.head
        while current.next != self.tail:
            current = current.next

        old_tail = self.tail
        self.tail = current
        self.tail.next = None
        self.size -= 1
        return old_tail.value

    def __len__(self) -> int:
        return self.size

    def __repr__(self) -> str:
        return f"{self.head}"


class StackLinkedList:
    def __init__(self) -> None:
        self.__items = LinkedList()  # estructura de datos implementadora

    def push(self, element: Any) -> None:  # agrega después de la cola de la lista
        self.__items.append(element)

    def pop(self) -> Any:
        if len(self.__items) == 0:
            raise EmptyStack("No puedes hacer pop... la pila está vacía")
        return self.__items.delete_and_return_last()

    def peek(self) -> Any:
        if len(self.__items) == 0:
            raise EmptyStack("No puedes hacer peek... la pila está vacía")
        return self.__items.tail.value

    def __len__(self) -> int:
        return len(self.__items)

    def __repr__(self) -> str:
        return str(self.__items)


s = StackLinkedList()
s.push(1)
s.push(2)
s.push(3)
print(s)         # 1 -> 2 -> 3 -> None
print(s.pop())   # 3
print(s.peek())  # 2
print(s)         # 1 -> 2 -> None
```
