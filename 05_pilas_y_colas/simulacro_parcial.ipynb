## 1️⃣ 🏷️ Verificador de anidamiento de etiquetas

**Contexto.** Un editor de texto simplificado quiere validar si las etiquetas de un documento (al estilo HTML, pero muy simplificadas) están correctamente anidadas.

**Formato de entrada.** Un string con etiquetas separadas por espacios. Una etiqueta de apertura se ve como `<nombre>` y su cierre como `</nombre>` — por ejemplo: `"<div> <span> </span> </div>"`.

**Objetivo.** Implementa una función que reciba el string y determine si las etiquetas están correctamente anidadas: cada etiqueta de apertura debe tener su cierre correspondiente **con el mismo nombre**, y deben cerrarse en el orden correcto (no pueden cruzarse).

**Restricciones.**
* Usa solo las operaciones de `Stack`.
* No puedes usar expresiones regulares ni librerías de parseo HTML — el `.split()` sobre espacios para separar tokens sí está permitido, no es la parte que se está evaluando.

**Ejemplos.**
```
"<div> <span> </span> </div>"   → True
"<div> <span> </div> </span>"   → False   (se cruzan: span abre después de div, pero div cierra antes que span)
"<div> </span>"                 → False   (el cierre no coincide con ninguna apertura pendiente)
"<div> <span>"                  → False   (quedan etiquetas sin cerrar)
```

## 2️⃣ 🖨️ Cola de impresión con cancelaciones

**Contexto.** Una impresora compartida en la oficina procesa los trabajos en el orden en que llegaron. Antes de imprimir, el sistema revisa si alguno de los trabajos en espera fue cancelado por su dueño.

**Objetivo.** Implementa una función que reciba una **cola** de identificadores de trabajos (enteros) y una **cola** de identificadores cancelados, y devuelva una **nueva cola** con los trabajos que sí se van a imprimir, respetando el orden original de llegada. Los identificadores cancelados no aparecen en el resultado.

**Restricciones.**
* Usa solo operaciones de `Queue` (`enqueue`, `dequeue`, `peek`, `len`) sobre ambas colas de entrada y sobre cualquier cola auxiliar que necesites.
* La cola de trabajos original y la cola de cancelados deben quedar **vacías** al terminar (asume que se “consumen” en el proceso — no necesitas restaurarlas en este problema particular).

**Ejemplo.**
```
trabajos    = [101, 102, 103, 104, 105]   (frente → final)
cancelados  = [102, 104]

Resultado: [101, 103, 105]
```

## 3️⃣ ✈️ Embarque prioritario

**Contexto.** Una aerolínea embarca a sus pasajeros por grupos de prioridad:

1. **Elite** (mayor prioridad)
2. **Business**
3. **Economy** (menor prioridad)

Dentro de un mismo grupo, se respeta el orden en que hicieron *check-in* (el que hizo check-in primero, embarca primero dentro de su grupo).

**Objetivo.** Implementa una cola de prioridad para pasajeros: cada pasajero se agrega con su nombre y su grupo (`1`, `2` o `3`). Al pedir el siguiente pasajero, debe devolverse siempre el de menor número de grupo disponible y, en caso de empate, el que hizo *check-in* primero.

**Restricciones.**
* Debes implementar tu propia cola de prioridad (con una lista de Python por dentro, como viste en `teoria_colas.ipynb`) — no existe un `PriorityQueue` con desempate por orden de llegada en `base.py`.

Orden de embarque:
Diego (1) → Andrés (1) → Sofía (2) → Marta (3) → Laura (3)

Fíjate en el desempate: Diego y Andrés son ambos grupo 1, y Diego embarca primero porque hizo check-in antes (paso 2 vs. paso 4).
