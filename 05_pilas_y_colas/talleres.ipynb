### 2️⃣ Triaje de pacientes por nivel de urgencia

Estás construyendo el módulo de **triaje** para una IPS. Debes diseñar la lógica que, dada la descripción del motivo de consulta de cada paciente y una tabla de puntajes por problemática, permita **atender siempre al paciente de mayor urgencia** y actualizar el flujo de atención en tiempo real.

#### Tu tarea

Implementa una **función en Python** que reciba:

1. `pacientes`: un **listado** de pacientes, donde cada paciente tiene:

   * `nombre` (str)
   * `edad` (int)
   * `diagnostico` (str) — descripción libre del motivo de consulta (p. ej., “dolor en el pecho con mareo”).

* **Calcular un puntaje de urgencia** para cada paciente:

  * Suma los puntajes de **cada problemática** cuya **palabra o frase clave** aparezca al menos una vez en el `diagnostico` (sin distinguir mayúsculas/minúsculas).
  * Si ninguna problemática coincide, el puntaje es 0.
* **Permitir operar en un ciclo interactivo (infinito)** con las siguientes acciones:

  * `atender`: extrae y muestra el **siguiente paciente a atender** (el de mayor puntaje).

  1. Mayor `puntaje` primero.
  2. Mayor `edad` primero.
  3. Orden de llegada (el que entró antes, primero).

### 3️⃣ Pila de colas: buscar un elemento y devolver su posición

Se tiene una **pila de colas**, es decir, una colección organizada como una pila (LIFO) en la que cada elemento es, a su vez, una **cola** con varios datos almacenados.

Dado un elemento `e`, se requiere **determinar si este se encuentra en alguna de las colas** dentro de la pila. En caso de estar presente, el programa deberá retornar la **posición emulada** de dicho elemento como si toda la pila de colas se interpretara como una **matriz** (colección de colecciones).

* La posición se representará mediante dos índices:

  * `i`: índice de la pila (qué cola dentro de la pila lo contiene).
  * `j`: índice dentro de la cola (la posición dentro de esa cola).

Si el elemento no se encuentra en ninguna de las colas, el algoritmo deberá devolver `-1`.

---

```
   [7, 9] #esto sería un objeto de tipo cola lleno con 7,9
] #esto es una pila con las tres colas mencionadas antes
e = 4
```

**Salida:**

```
(1, 1)
```
