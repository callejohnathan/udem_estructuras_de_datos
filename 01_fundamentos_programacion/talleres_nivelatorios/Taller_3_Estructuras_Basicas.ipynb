## Ejercicio 1: Mecánica de listas: índices y slicing

### Enunciado

Dada la lista `nombres = ["Ana", "Luis", "Marta", "Pedro", "Sofía"]`, obtén (sin ciclos): (a) el primer nombre, (b) el último usando índice negativo, (c) los tres primeros, (d) los nombres desde la posición 1 hasta el final, y (e) la lista en orden invertido.

### Modelo de pensamiento

1. Un índice indica "posición desde el inicio", empezando en 0 — el primer elemento es índice 0, no 1.
2. Un índice negativo cuenta "desde el final hacia atrás": -1 es el último, -2 el penúltimo, etc. Es útil cuando no sabes (o no te importa) el largo exacto de la lista.
3. El slicing `lista[inicio:fin]` toma desde "inicio" **hasta** "fin" **sin incluir** "fin" — ese límite exclusivo es la fuente más común de errores de conteo.
4. Dejar "inicio" o "fin" vacíos en el slicing significa "desde el principio" o "hasta el final", respectivamente.
5. `lista[::-1]` es un caso especial de slicing con "paso" -1: recorre la lista completa, pero en reversa.

```python
nombres: list[str] = ["Ana", "Luis", "Marta", "Pedro", "Sofía"]

primero: str = nombres[0]
ultimo: str = nombres[-1]
primeros_tres: list[str] = nombres[0:3]
desde_uno: list[str] = nombres[1:]
invertida: list[str] = nombres[::-1]

print(primero, ultimo, primeros_tres, desde_uno, invertida)
```

## Ejercicio 2: Transformar una lista en otra

### Enunciado

Dada una lista de precios sin IVA, construye una **nueva** lista con cada precio incluyendo un 19% de IVA.

### Modelo de pensamiento

1. La idea clave: **no** modificas la lista original — construyes una lista nueva, vacía al inicio, y la vas llenando con los valores transformados.
2. El patrón es: por cada elemento de la lista original, calcula el nuevo valor, y agrégalo a la lista nueva con `.append(...)`.
3. Es el mismo `for elemento in lista` que ya conoces del Taller 2, solo que en vez de acumular una suma o un contador, acumulas una **lista completa**.
4. (Para más adelante: esto es exactamente lo que hace una "comprensión de listas" en una sola línea — pero primero hay que dominarlo con el `for` explícito.)

```python
def agregar_iva(precios: list[float]) -> list[float]:
    precios_con_iva: list[float] = []
    for precio in precios:
        precios_con_iva.append(precio * 1.19)
    return precios_con_iva

print(agregar_iva([100.0, 200.0, 50.0]))
```

## Ejercicio 3: Matrices: lista de listas

### Enunciado

Representa una matriz de 3x3 como una lista de listas. Escribe una función que reciba la matriz y devuelva una lista con la suma de cada fila.

### Modelo de pensamiento

1. Una matriz como lista de listas es "una lista, donde cada elemento es a su vez otra lista" (una fila). `matriz[0]` es la primera fila completa; `matriz[0][2]` es el elemento de la fila 0, columna 2.
2. Para sumar cada fila por separado, recorre las filas (ciclo externo), y usa `sum(fila)` directamente sobre cada una — cada fila ya es una lista normal de números.
3. El resultado no es un solo número sino una **lista** (una suma por cada fila) — igual que en el ejercicio anterior, necesitas ir construyendo una lista de resultados con `.append(...)`.
4. Antes de programar, dibuja la matriz en papel y verifica a mano cuál sería la fila 0, y qué representa `matriz[0][1]` — confundir fila/columna es el error más común aquí.

```python
def sumar_filas(matriz: list[list[int]]) -> list[int]:
    sumas_por_fila: list[int] = []
    for fila in matriz:
        sumas_por_fila.append(sum(fila))
    return sumas_por_fila

matriz: list[list[int]] = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
print(sumar_filas(matriz))  # [6, 15, 24]
```

## Ejercicio 4: Diccionario de frecuencias

### Enunciado

Dada una lista de letras obtenidas (por ejemplo `["A","B","A","C","A","B"]`), construye un diccionario que cuente cuántas veces aparece cada letra.

### Modelo de pensamiento

1. Un diccionario asocia una "clave" (la letra) con un "valor" (cuántas veces ha aparecido) — piénsalo como una tabla de dos columnas que vas llenando sobre la marcha.
2. Recorres la lista igual que siempre (`for elemento in lista`), pero ahora, por cada elemento, hay dos casos: (a) es la primera vez que aparece esa letra (no está en el diccionario todavía), o (b) ya la habías visto antes.
3. Antes de sumar 1 al contador de una clave, asegúrate de que esa clave ya exista en el diccionario — si no, la creas con valor inicial 0.
4. Este es, en el fondo, el mismo patrón de "contador" del Taller 2, aplicado a **muchas claves distintas** en simultáneo en vez de una sola variable.

```python
def contar_frecuencias(letras: list[str]) -> dict[str, int]:
    frecuencias: dict[str, int] = {}
    for letra in letras:
        if letra not in frecuencias:
            frecuencias[letra] = 0
        frecuencias[letra] += 1
    return frecuencias

print(contar_frecuencias(["A", "B", "A", "C", "A", "B"]))
# {'A': 3, 'B': 2, 'C': 1}
```

## Ejercicio 5: Combinar lista y diccionario: agrupar por rango

### Enunciado

Dada una lista de tuplas `(nombre, nota)`, agrupa los nombres en un diccionario donde las claves sean las letras (A/B/C/D/F, reutilizando la lógica de `nota_a_letra` del Taller 1) y los valores sean listas de nombres.

### Modelo de pensamiento

1. Este ejercicio junta tres cosas que ya sabes por separado: clasificar con condicionales (Taller 1), recorrer con un ciclo (Taller 2), y ahora, en vez de contar, ir agregando nombres a listas dentro de un diccionario.
2. Para cada estudiante, primero decide su letra (reutilizando `nota_a_letra`) — esa letra es la **clave** del diccionario.
3. El **valor** asociado a cada clave no es un número esta vez, sino una lista de nombres — si la clave no existe todavía, créala con una lista vacía `[]`, y luego usa `.append(nombre)` en cualquier caso.
4. Es el mismo patrón "¿existe la clave? si no, créala" del ejercicio de frecuencias, solo que el valor inicial es `[]` en vez de `0`, y en vez de sumar 1, agregas un elemento a la lista.

```python
def nota_a_letra(nota: float) -> str:
    if nota >= 4.5:
        return "A"
    elif nota >= 4.0:
        return "B"
    elif nota >= 3.0:
        return "C"
    elif nota >= 2.0:
        return "D"
    else:
        return "F"


def agrupar_por_letra(estudiantes: list[tuple[str, float]]) -> dict[str, list[str]]:
    grupos: dict[str, list[str]] = {}
    for nombre, nota in estudiantes:
        letra = nota_a_letra(nota)
        if letra not in grupos:
            grupos[letra] = []
        grupos[letra].append(nombre)
    return grupos

estudiantes = [("Ana", 4.8), ("Luis", 3.2), ("Marta", 4.9), ("Pedro", 1.5)]
print(agrupar_por_letra(estudiantes))
# {'A': ['Ana', 'Marta'], 'C': ['Luis'], 'F': ['Pedro']}
```

## Ejercicio 6: Estructura anidada: diccionario de listas (inventario)

### Enunciado

Representa un inventario como un diccionario donde la clave es el nombre del producto y el valor es una lista `[cantidad, precio_unitario]`. Escribe una función que devuelva el valor total del inventario.

### Modelo de pensamiento

1. Aquí el diccionario ya viene armado (no lo construyes tú); tu trabajo es **recorrerlo**. Para usar la clave y el valor a la vez, se usa `inventario.items()`.
2. Cada valor no es un número simple sino una lista de dos posiciones — antes de operar, identifica qué representa cada posición: [0] es cantidad, [1] es precio. "Desempacarlo" en dos variables con nombres claros evita confundirte.
3. El total general es, otra vez, un acumulador: arranca en 0 y en cada vuelta le sumas el aporte de ese producto (`cantidad * precio`).
4. No necesitas la clave (el nombre del producto) para este cálculo en particular — pero sí para otros problemas parecidos, así que vale la pena practicar `.items()` aunque aquí solo uses el valor.

```python
def valor_total_inventario(inventario: dict[str, list[float]]) -> float:
    total: float = 0.0
    for producto, datos in inventario.items():
        cantidad, precio_unitario = datos
        total += cantidad * precio_unitario
    return total

inventario: dict[str, list[float]] = {
    "lapiz": [100, 500],
    "cuaderno": [30, 3500],
    "borrador": [50, 800],
}
print(valor_total_inventario(inventario))
```

## Ejercicio 7: Lista de diccionarios: el mejor estudiante (reto)

### Enunciado

Representa varios estudiantes como una lista de diccionarios, cada uno con las claves `"nombre"` y `"notas"`. Escribe una función que devuelva el nombre del estudiante con el mejor promedio.

### Modelo de pensamiento

1. Esta es la estructura más "real": una lista de registros, donde cada registro es un diccionario con varios campos — así es exactamente como luego vas a ver datos en el mundo real (una fila de tabla = un diccionario).
2. Para cada estudiante, primero necesitas calcular **su** promedio (`sum(sus_notas) / len(sus_notas)`) antes de poder compararlo con los demás.
3. Una vez que sabes calcular el promedio de uno, el problema completo es el patrón "el mejor hasta ahora" del Taller 2: recorres la lista de estudiantes, y te quedas con el nombre de quien tenga el promedio más alto visto hasta el momento.
4. Ten cuidado de guardar tanto el promedio máximo **como** el nombre asociado a ese promedio — si solo guardas el número, pierdes de quién era.

```python
from typing import Any


def mejor_estudiante(estudiantes: list[dict[str, Any]]) -> str:
    mejor_nombre: str = estudiantes[0]["nombre"]
    mejor_promedio: float = sum(estudiantes[0]["notas"]) / len(estudiantes[0]["notas"])

    for estudiante in estudiantes[1:]:
        notas = estudiante["notas"]
        promedio = sum(notas) / len(notas)
        if promedio > mejor_promedio:
            mejor_promedio = promedio
            mejor_nombre = estudiante["nombre"]

    return mejor_nombre

estudiantes = [
    {"nombre": "Ana", "notas": [4.5, 4.8, 4.0]},
    {"nombre": "Luis", "notas": [3.0, 2.5, 3.5]},
    {"nombre": "Marta", "notas": [5.0, 4.9, 4.7]},
]
print(mejor_estudiante(estudiantes))  # Marta
```
