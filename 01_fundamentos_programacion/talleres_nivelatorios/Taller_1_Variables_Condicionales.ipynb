El objetivo de este taller no es la sintaxis de `if`/`elif`/`else` (eso ya lo conoces), sino **traducir una regla de decisión, escrita en palabras, a una secuencia de condiciones en código** — identificando qué casos son excluyentes, en qué orden comprobarlos, y cómo combinar varias condiciones con `and`/`or` sin perderte.

## Ejercicio 1: Clasificar el signo de un número

### Enunciado

Dado un número entero, escribe una función que indique si es `"positivo"`, `"negativo"` o `"cero"`.

### Modelo de pensamiento

1. Identifica que los tres casos son mutuamente excluyentes: un número solo puede ser positivo, negativo o cero — nunca dos a la vez.
2. Decide el orden de comprobación: empieza por el caso más fácil de comparar (cero), luego positivo, y lo que quede será negativo.
3. Traduce cada caso a una comparación directa con 0.
4. La función debe **devolver** un texto (`str`), no imprimirlo — quien la use decide después qué hacer con el resultado.

```python
def clasificar_signo(n: int) -> str:
    if n == 0:
        return "cero"
    elif n > 0:
        return "positivo"
    else:
        return "negativo"

print(clasificar_signo(5))   # positivo
print(clasificar_signo(-3))  # negativo
print(clasificar_signo(0))   # cero
```

## Ejercicio 2: Nota numérica a letra

### Enunciado

Escribe una función que reciba una nota entre 0.0 y 5.0 y devuelva la letra correspondiente: `A` si nota >= 4.5, `B` si nota >= 4.0, `C` si nota >= 3.0, `D` si nota >= 2.0, y `F` en cualquier otro caso.

### Modelo de pensamiento

1. Antes de programar, escribe los rangos en una tabla: [4.5, 5.0] -> A, [4.0, 4.5) -> B, etc. Fijar bien los límites (¿el borde entra o no?) es la parte que más errores genera.
2. Nota que los rangos están ordenados de mayor a menor. Si comparas con `elif` en ese mismo orden, en el momento en que una condición es verdadera ya sabes que las anteriores fueron falsas.
3. Es decir: si preguntas primero "¿es >= 4.5?" y no lo es, ya sabes que es menor que 4.5 — al preguntar después "¿es >= 4.0?" no hace falta repetir también "< 4.5".
4. El último caso (F) no necesita condición: es lo que queda después de descartar todo lo demás.

```python
def nota_a_letra(nota: float) -> str:
    if nota >= 4.5:
        return "A"
    elif nota >= 4.0:
        return "B"
    elif nota >= 3.0:
        return "C"
    elif nota >= 2.0:
        return "D"
    else:
        return "F"

for n in [5.0, 4.5, 4.4, 3.0, 1.5]:
    print(n, "->", nota_a_letra(n))
```

## Ejercicio 3: Elegibilidad con condición compuesta

### Enunciado

Una beca aplica si el estudiante tiene promedio >= 4.0 **Y** (es de estrato 1 o 2 **O** tiene alguna condición de discapacidad registrada). Escribe una función que reciba `promedio`, `estrato` y `tiene_condicion`, y devuelva si aplica.

### Modelo de pensamiento

1. Antes de escribir código, traduce el enunciado a una expresión lógica en palabras: "promedio alto" **Y** ("estrato bajo" **O** "tiene condición"). El Y separa un requisito obligatorio del resto; el O agrupa alternativas.
2. Identifica qué operador corresponde a cada palabra clave: "y" -> `and`, "o" -> `or`.
3. Ojo con la prioridad: `and` se evalúa antes que `or` si no hay paréntesis. No confíes en la precedencia por defecto — agrupa con paréntesis explícitos lo que va junto, para que sea visible.
4. Escribe primero la condición del promedio, y luego la condición compuesta completa entre paréntesis.

```python
def aplica_beca(promedio: float, estrato: int, tiene_condicion: bool) -> bool:
    promedio_alto = promedio >= 4.0
    estrato_bajo = estrato in (1, 2)
    return promedio_alto and (estrato_bajo or tiene_condicion)

print(aplica_beca(4.2, 3, True))   # True
print(aplica_beca(4.2, 3, False))  # False
print(aplica_beca(3.9, 1, False))  # False
```

## Ejercicio 4: Año bisiesto

### Enunciado

Escribe una función que reciba un año y devuelva si es bisiesto. Regla: es bisiesto si es divisible por 4, excepto los divisibles por 100 que no sean también divisibles por 400.

### Modelo de pensamiento

1. La regla tiene una excepción dentro de una excepción — es tentador escribirla de un tirón y equivocarse. Sepárala en tres preguntas simples: (a) ¿es divisible por 4?, (b) ¿es divisible por 100?, (c) ¿es divisible por 400?
2. Traduce la regla completa en palabras usando esas tres preguntas: "es bisiesto si (a) es verdad, Y ( (b) es falso O (c) es verdad )".
3. La divisibilidad se comprueba con el resto de la división: `n % k == 0`.
4. Nombra cada sub-condición como una variable booleana antes de combinarlas — así el `and`/`or` final se lee casi como la frase en español, y un error se detecta mucho más fácil.

```python
def es_bisiesto(anio: int) -> bool:
    divisible_por_4 = anio % 4 == 0
    divisible_por_100 = anio % 100 == 0
    divisible_por_400 = anio % 400 == 0
    return divisible_por_4 and (not divisible_por_100 or divisible_por_400)

for anio in [2000, 1900, 2024, 2023]:
    print(anio, "->", es_bisiesto(anio))
```

## Ejercicio 5: FizzBuzz: el orden de los casos importa

### Enunciado

Escribe una función que reciba un número `n` y devuelva `"FizzBuzz"` si es divisible por 3 y por 5, `"Fizz"` si solo es divisible por 3, `"Buzz"` si solo es divisible por 5, o el número (como texto) en cualquier otro caso.

### Modelo de pensamiento

1. Antes de escribir el `if`, identifica que aquí hay casos que **no son excluyentes en apariencia**: "divisible por 3" y "divisible por 5" pueden ser ciertos ambos a la vez. Si programas primero el caso "divisible por 3" sin más cuidado, nunca vas a llegar a comprobar el caso combinado.
2. Regla general: cuando un caso es *más específico* que otro (ser divisible por 3 y 5 a la vez es más específico que ser divisible solo por 3), el caso específico debe comprobarse primero. Si pones primero el caso general, el específico nunca se alcanza.
3. Es más claro escribir la condición compuesta explícita (`n % 3 == 0 and n % 5 == 0`) que asumir que "divisible por 3 y 5" es lo mismo que "divisible por 15" — así se nota la relación directa con el enunciado.
4. Ordena los `if`: primero el caso más específico (FizzBuzz), luego los casos individuales, y al final el caso por defecto.

```python
def fizzbuzz_de(n: int) -> str:
    if n % 3 == 0 and n % 5 == 0:
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        return str(n)

# Este ciclo es solo para probar la función con varios valores seguidos;
# el ciclo como tal se estudia a fondo en el Taller 2.
for n in range(1, 16):
    print(fizzbuzz_de(n))
```

## Ejercicio 6: Carrito de compra con descuentos escalonados (reto)

### Enunciado

Escribe una función que reciba `precio_unitario`, `cantidad` y `es_socio`, y devuelva el total a pagar. Reglas: si `cantidad >= 10` hay 15% de descuento; si `cantidad >= 5` hay 10%; en otro caso, 0%. Si además `es_socio` es verdadero, se resta un 5% adicional **sobre el total ya descontado**.

### Modelo de pensamiento

1. Separa el problema en dos decisiones independientes: (1) ¿qué descuento por cantidad aplica?, (2) ¿aplica el descuento adicional de socio? Resolverlas por separado evita mezclar todo en un solo `if` gigante.
2. Para el descuento por cantidad, los rangos son excluyentes y ordenados (igual que en el ejercicio de la nota a letra): compara de mayor a menor cantidad.
3. Calcula primero el total bruto (`precio_unitario * cantidad`), luego aplica el descuento por cantidad sobre ese total, y por último, si aplica, el descuento de socio sobre el resultado **ya descontado** (no sobre el bruto) — el orden importa.
4. Aplicar un descuento del d% significa multiplicar por `(1 - d/100)`. Encadenar descuentos es multiplicar varias veces por estos factores, en el orden correcto.

```python
def calcular_total(precio_unitario: float, cantidad: int, es_socio: bool) -> float:
    total = precio_unitario * cantidad

    if cantidad >= 10:
        descuento_cantidad = 0.15
    elif cantidad >= 5:
        descuento_cantidad = 0.10
    else:
        descuento_cantidad = 0.0

    total = total * (1 - descuento_cantidad)

    if es_socio:
        total = total * (1 - 0.05)

    return round(total, 2)

print(calcular_total(1000, 12, True))   # con descuento por cantidad y socio
print(calcular_total(1000, 3, False))   # sin descuentos
```
