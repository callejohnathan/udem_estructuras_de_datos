## Ejercicio 1: Una función con varios parámetros obligatorios

### Enunciado

Escribe una función `costo_total(precio_unitario, cantidad, descuento, impuesto) -> float` que calcule el costo total de una compra: primero calcula el subtotal (`precio_unitario × cantidad`), le aplica el descuento (como fracción, ej. `0.1` = 10%), y finalmente le aplica el impuesto (como fracción) sobre ese valor ya descontado. Llama la función dos veces: una con los argumentos en orden posicional, y otra pasando los mismos argumentos por nombre en un orden distinto.

### Modelo de pensamiento

1. Cuando una función necesita varios datos de entrada relacionados (aquí: precio, cantidad, descuento, impuesto), cada uno se convierte en su propio parámetro. El **orden** en que los defines en la firma importa cuando llamas a la función de forma **posicional** — los argumentos se asignan de izquierda a derecha, en el mismo orden en que aparecen en la definición.
2. El enunciado describe una secuencia de pasos, cada uno aplicado sobre el resultado del paso anterior — no sobre el precio original. Antes de programar, escribe la fórmula en papel, paso a paso: `subtotal = precio_unitario * cantidad`; `con_descuento = subtotal * (1 - descuento)`; `total = con_descuento * (1 + impuesto)`. Traducir cada paso a su propia línea de código evita mezclar todo en una sola expresión difícil de verificar a simple vista.
3. Python permite llamar a una función pasando los argumentos **por nombre** (`costo_total(cantidad=3, precio_unitario=1000, descuento=0.1, impuesto=0.19)`), y en ese caso el orden ya no importa — es más largo de escribir, pero elimina cualquier duda sobre "cuál argumento es cuál", algo especialmente útil cuando una función tiene varios parámetros del mismo tipo (aquí, todos son números).

```python
def costo_total(precio_unitario: float, cantidad: int, descuento: float, impuesto: float) -> float:
    subtotal = precio_unitario * cantidad
    con_descuento = subtotal * (1 - descuento)
    total = con_descuento * (1 + impuesto)
    return total


# Llamada posicional: el orden debe coincidir con la firma de la función
print(costo_total(1000.0, 3, 0.1, 0.19))

# Llamada con nombre: el orden ya no importa
print(costo_total(cantidad=3, precio_unitario=1000.0, impuesto=0.19, descuento=0.1))
```

**Complejidad:** tres multiplicaciones y una resta, sin ciclos ni estructuras adicionales → **O(1) en tiempo y en espacio**, sin importar los valores de entrada.

## Ejercicio 2: Una función que llama a otra

### Enunciado

Escribe `area_rectangulo(base, altura) -> float` y `perimetro_rectangulo(base, altura) -> float`. Luego escribe una tercera función `resumen_rectangulo(base, altura) -> tuple[float, float]` que **llame** a las dos anteriores (sin repetir ninguna fórmula) y devuelva ambos resultados juntos.

### Modelo de pensamiento

1. Área y perímetro comparten los mismos dos parámetros de entrada, pero calculan cosas distintas — cada cálculo merece su propia función, en vez de mezclar ambas fórmulas en un solo bloque.
2. `resumen_rectangulo` no repite ninguna fórmula: recibe los mismos dos parámetros (`base`, `altura`) y simplemente los **reenvía**, sin modificarlos, a `area_rectangulo` y a `perimetro_rectangulo` — practicando el patrón "una función que llama a otra" cuando hay más de un parámetro que pasar de una función a la siguiente.
3. Los nombres de los parámetros no tienen que coincidir entre funciones — lo que importa es el **orden** en que se pasan (o usar nombres explícitos en la llamada), no que se llamen igual en ambas firmas.

```python
def area_rectangulo(base: float, altura: float) -> float:
    return base * altura


def perimetro_rectangulo(base: float, altura: float) -> float:
    return 2 * (base + altura)


def resumen_rectangulo(base: float, altura: float) -> tuple[float, float]:
    area = area_rectangulo(base, altura)
    perimetro = perimetro_rectangulo(base, altura)
    return area, perimetro


area, perimetro = resumen_rectangulo(4.0, 5.0)
print(f"Área: {area}, Perímetro: {perimetro}")
```

**Complejidad:** cada función hace una cantidad fija de operaciones aritméticas, sin ciclos → **O(1) en tiempo y en espacio**, tanto para las funciones individuales como para `resumen_rectangulo`.

## Ejercicio 3: Parámetro opcional, combinado con varios obligatorios

### Enunciado

Escribe `precio_final(precio_unitario, cantidad, descuento=0.0) -> float`, que calcule el costo de comprar `cantidad` unidades a `precio_unitario`, aplicando un `descuento` opcional (como fracción). Si no se indica descuento, debe calcular el precio sin ningún descuento. Debe poder llamarse tanto con dos argumentos como con tres.

### Modelo de pensamiento

1. Un parámetro con valor por defecto (`descuento: float = 0.0`) debe ir siempre **después** de los parámetros obligatorios en la firma de la función — Python no permite que un parámetro sin valor por defecto aparezca después de uno que sí lo tiene, porque no sabría a cuál asignar un argumento posicional que llegue después.
2. Con varios parámetros obligatorios más uno opcional, hay más de una forma válida de llamar la función: `precio_final(1000, 3)`, `precio_final(1000, 3, 0.1)`, o `precio_final(precio_unitario=1000, cantidad=3, descuento=0.1)` — las tres son correctas, y cuál usar depende de qué tan explícito quieras ser en cada llamada particular.
3. El cálculo reutiliza la misma idea de "aplicar un descuento como fracción" del Taller 4 (`precio * (1 - descuento)`), ahora combinada con el nuevo parámetro `cantidad` en el mismo cálculo.

```python
def precio_final(precio_unitario: float, cantidad: int, descuento: float = 0.0) -> float:
    return precio_unitario * cantidad * (1 - descuento)


print(precio_final(1000.0, 3))          # sin descuento
print(precio_final(1000.0, 3, 0.15))    # con 15% de descuento
```

**Complejidad:** dos multiplicaciones y una resta → **O(1) en tiempo y en espacio**, sin importar si se llama con dos o con tres argumentos.

## Ejercicio 4: Retorno múltiple con tuplas, a partir de dos listas paralelas

### Enunciado

Escribe `estadisticas_ponderadas(numeros, pesos) -> tuple[float, float]`, que reciba dos listas del mismo largo (números y sus pesos correspondientes) y devuelva la **suma ponderada** y el **promedio ponderado**, calculados en un solo recorrido.

### Modelo de pensamiento

1. Esta función recibe **dos** listas que deben recorrerse en paralelo (misma posición `i` en ambas), no una sola — practicando coordinar más de un parámetro de tipo lista dentro de la misma función.
2. La suma ponderada no es la suma directa de los números: cada número se multiplica por su peso correspondiente antes de acumularlo — `suma_ponderada += numeros[i] * pesos[i]`.
3. El promedio ponderado **no** se calcula dividiendo entre la cantidad de números — se divide la suma ponderada entre la **suma de los pesos** (que también hay que acumular en el mismo recorrido). Confundir estos dos denominadores es el error más común en este tipo de cálculo, así que antes de programar conviene escribir la fórmula completa en papel: `promedio_ponderado = suma_ponderada / suma_pesos`.
4. Como la función devuelve dos valores relacionados, se empaquetan en una tupla, y quien la llama los desempaqueta en dos variables separadas en una sola línea.

```python
def estadisticas_ponderadas(numeros: list[float], pesos: list[float]) -> tuple[float, float]:
    suma_ponderada = 0.0
    suma_pesos = 0.0
    for i in range(len(numeros)):
        suma_ponderada += numeros[i] * pesos[i]
        suma_pesos += pesos[i]
    promedio_ponderado = suma_ponderada / suma_pesos
    return suma_ponderada, promedio_ponderado


notas = [4.5, 3.0, 5.0]
creditos = [3, 4, 2]  # peso de cada nota según los créditos de la materia

suma, promedio = estadisticas_ponderadas(notas, creditos)
print(f"Suma ponderada: {suma}, Promedio ponderado: {promedio:.2f}")
```

**Complejidad:** un solo recorrido de ambas listas en paralelo, con operaciones O(1) por posición → **O(n) en tiempo**, con `n = len(numeros)` (se asume que ambas listas tienen el mismo largo). En **espacio** es **O(1)** adicional: solo se mantienen dos acumuladores, sin estructuras que crezcan con `n`.

## Ejercicio 5: Cadena de funciones en tres niveles

### Enunciado

Un empleado tiene un salario base, unas horas extra trabajadas (pagadas a 1.5 veces el valor de la hora normal) y un porcentaje de bonificación sobre el salario bruto. Escribe tres funciones encadenadas: `calcular_pago_extra(horas_extra, valor_hora) -> float` (el nivel más bajo); `calcular_salario_bruto(salario_base, horas_extra, valor_hora) -> float` (llama a la anterior y le suma el salario base); y `calcular_salario_con_bono(salario_base, horas_extra, valor_hora, porcentaje_bono) -> float` (llama a la anterior y le agrega el bono calculado sobre el salario bruto).

### Modelo de pensamiento

1. Cada nivel de esta cadena tiene una responsabilidad clara y usa solo los parámetros que necesita para **su propio** cálculo: `calcular_pago_extra` no sabe nada del salario base ni del bono, solo conoce horas extra y valor por hora. Diseñar así (cada función con la mínima información necesaria) hace que cada una sea más fácil de entender y de probar por separado, sin depender de las demás.
2. La función del medio (`calcular_salario_bruto`) recibe **más** parámetros que la de abajo, porque necesita tanto el suyo propio (`salario_base`) como los que le va a pasar hacia adelante a `calcular_pago_extra` (`horas_extra`, `valor_hora`). Antes de escribir el código, identifica qué parámetros "pasan de largo" hacia una función más interna, y cuáles se usan directamente en el nivel actual.
3. La función de más arriba (`calcular_salario_con_bono`) tiene el mayor número de parámetros de las tres, porque reúne todo lo que las dos funciones internas requieren, más su propio dato nuevo (`porcentaje_bono`). Dibuja en papel qué función llama a cuál, y qué parámetros viajan de una a otra, antes de programar — visualizar esa cadena evita confundir en qué nivel corresponde cada cálculo.

```python
def calcular_pago_extra(horas_extra: float, valor_hora: float) -> float:
    return horas_extra * valor_hora * 1.5


def calcular_salario_bruto(salario_base: float, horas_extra: float, valor_hora: float) -> float:
    pago_extra = calcular_pago_extra(horas_extra, valor_hora)
    return salario_base + pago_extra


def calcular_salario_con_bono(
    salario_base: float, horas_extra: float, valor_hora: float, porcentaje_bono: float
) -> float:
    salario_bruto = calcular_salario_bruto(salario_base, horas_extra, valor_hora)
    return salario_bruto * (1 + porcentaje_bono)


print(calcular_salario_con_bono(2000000.0, 10.0, 15000.0, 0.05))
```

**Complejidad:** los tres niveles hacen una cantidad fija de operaciones aritméticas cada uno, sin ciclos ni estructuras adicionales → **O(1) en tiempo y en espacio** en total, sin importar los valores de entrada (la profundidad de la cadena de llamadas es fija: siempre son exactamente tres niveles, no depende del tamaño de ningún dato).

## Ejercicio 6: Descomponer un problema en varias funciones pequeñas

### Enunciado

Tienes una lista de artículos de un pedido (diccionarios con `"precio"` y `"cantidad"`), un valor booleano que indica si el envío es gratis, y una tasa de impuesto. Divide el cálculo de la factura final en varias funciones pequeñas: `costo_item(item)` (el costo de un solo artículo); `costo_pedido(items)` (la suma de todos los artículos, usando la anterior); `costo_envio(es_gratis)` (0 si el envío es gratis, o un valor fijo si no); y `factura_final(items, es_envio_gratis, impuesto)` (junta todo: costo de los artículos más envío, y aplica el impuesto al final).

### Modelo de pensamiento

1. Identifica la "unidad de cálculo" más pequeña del problema: el costo de **un** artículo es la pieza más chica, y todo lo demás se construye sobre ella — por eso `costo_item` va primero y no depende de ninguna otra función que definas en este ejercicio.
2. `costo_pedido` recorre la lista de artículos y llama a `costo_item` en cada vuelta — el mismo patrón de acumulador de siempre, pero acumulando el resultado de una llamada a función en vez de un cálculo escrito directamente en el ciclo.
3. `costo_envio` es una función pequeña con una decisión simple: si `es_gratis` es `True`, el envío no suma nada; si no, se cobra un valor fijo. No necesita ningún otro dato del pedido — es intencionalmente independiente de `costo_pedido`, y eso la hace fácil de entender y de probar sola, sin depender del resto.
4. `factura_final` es la función que "orquesta" a las demás: no calcula nada aritmético por sí misma, solo llama a las funciones anteriores en el orden correcto, junta sus resultados, y aplica la fórmula final (subtotal + impuesto) — el mismo rol de función orquestadora que ya viste en el Taller 4, ahora repartiendo el trabajo entre tres funciones auxiliares en vez de una sola.

```python
def costo_item(item: dict[str, float]) -> float:
    return item["precio"] * item["cantidad"]


def costo_pedido(items: list[dict[str, float]]) -> float:
    total = 0.0
    for item in items:
        total += costo_item(item)
    return total


def costo_envio(es_gratis: bool) -> float:
    if es_gratis:
        return 0.0
    return 15000.0


def factura_final(items: list[dict[str, float]], es_envio_gratis: bool, impuesto: float) -> float:
    subtotal = costo_pedido(items) + costo_envio(es_envio_gratis)
    return subtotal * (1 + impuesto)


pedido = [
    {"precio": 20000.0, "cantidad": 2},
    {"precio": 5000.0, "cantidad": 3},
]

print(factura_final(pedido, es_envio_gratis=False, impuesto=0.19))
print(factura_final(pedido, es_envio_gratis=True, impuesto=0.19))
```

**Complejidad:** `costo_item` y `costo_envio` son O(1) cada una. `costo_pedido` recorre los `k` artículos del pedido, llamando a `costo_item` una vez por cada uno → **O(k) en tiempo**. `factura_final` llama a `costo_pedido` (O(k)) y a `costo_envio` (O(1)) una vez cada una → **O(k) en tiempo** en total, con `k` el número de artículos del pedido. En **espacio** es **O(1)** adicional: no se construyen estructuras nuevas proporcionales a `k`, solo se acumulan números.

## Ejercicio 7: Reporte de curso con funciones encadenadas (reto)

### Enunciado

Construye el reporte de un curso con funciones que se llaman entre sí, en cuatro niveles: `promedio_estudiante(notas) -> float`; `es_aprobado(promedio, nota_minima) -> bool`; `clasificar_estudiante(nombre, notas, nota_minima) -> dict[str, Any]` (llama a las dos anteriores y arma un diccionario con `"nombre"`, `"promedio"` y `"aprobado"`); y `resumen_curso(estudiantes, nota_minima) -> dict[str, Any]` (recorre la lista de estudiantes, llama a `clasificar_estudiante` para cada uno, y arma un resumen general con el promedio del curso y la cantidad de aprobados).

### Modelo de pensamiento

1. Este reto encadena cuatro funciones, cada una construida sobre la anterior — antes de programar, dibuja el árbol de llamadas: `resumen_curso` llama a `clasificar_estudiante`, que a su vez llama tanto a `promedio_estudiante` como a `es_aprobado`. Ver ese árbol en papel ayuda a decidir qué parámetros necesita cada nivel, y en qué orden se calculan las cosas.
2. `es_aprobado` recibe el **umbral** (`nota_minima`) como parámetro en vez de tenerlo fijo dentro de la función (a diferencia del Taller 4, donde `>= 3.0` estaba escrito directamente) — así, el mismo código sirve si algún curso usa una nota mínima distinta, sin tener que duplicar la función para cada caso.
3. `clasificar_estudiante` es el punto donde se juntan los dos cálculos por estudiante: primero obtiene el promedio llamando a `promedio_estudiante(notas)`, y **con ese resultado ya calculado**, se lo pasa como argumento a `es_aprobado(promedio, nota_minima)` — nota que el resultado de una llamada se convierte en el argumento de entrada de la siguiente, encadenando los datos entre funciones en vez de recalcular el promedio dos veces.
4. `resumen_curso` es el nivel más alto: recorre a todos los estudiantes, llamando a `clasificar_estudiante` con los tres parámetros que necesita cada uno (nombre, notas, y el mismo `nota_minima` para todos), y aplica sobre los resultados que le van devolviendo esas llamadas los mismos patrones de acumulador y contador que ya conoces del Taller 2 (suma de promedios, conteo de aprobados).

```python
from typing import Any


def promedio_estudiante(notas: list[float]) -> float:
    return sum(notas) / len(notas)


def es_aprobado(promedio: float, nota_minima: float) -> bool:
    return promedio >= nota_minima


def clasificar_estudiante(nombre: str, notas: list[float], nota_minima: float) -> dict[str, Any]:
    promedio = promedio_estudiante(notas)
    aprobado = es_aprobado(promedio, nota_minima)
    return {"nombre": nombre, "promedio": promedio, "aprobado": aprobado}


def resumen_curso(estudiantes: list[dict[str, Any]], nota_minima: float) -> dict[str, Any]:
    suma_promedios = 0.0
    aprobados = 0
    clasificaciones: list[dict[str, Any]] = []

    for estudiante in estudiantes:
        clasificacion = clasificar_estudiante(estudiante["nombre"], estudiante["notas"], nota_minima)
        clasificaciones.append(clasificacion)
        suma_promedios += clasificacion["promedio"]
        if clasificacion["aprobado"]:
            aprobados += 1

    return {
        "detalle": clasificaciones,
        "promedio_curso": suma_promedios / len(estudiantes),
        "cantidad_aprobados": aprobados,
    }


estudiantes = [
    {"nombre": "Ana", "notas": [4.5, 4.8, 4.0]},
    {"nombre": "Luis", "notas": [3.0, 2.5, 3.5]},
    {"nombre": "Marta", "notas": [5.0, 4.9, 4.7]},
]

print(resumen_curso(estudiantes, nota_minima=3.0))
```

**Complejidad:** sea `e` el número de estudiantes y `m` el número de notas por estudiante. `promedio_estudiante` es O(m); `es_aprobado` es O(1); `clasificar_estudiante` llama a ambas una vez → O(m). `resumen_curso` llama a `clasificar_estudiante` una vez por cada uno de los `e` estudiantes → **O(e × m) en tiempo** en total, equivalente a recorrer una vez cada nota de cada estudiante. En **espacio**, la lista `clasificaciones` guarda un diccionario pequeño (tres valores) por cada estudiante → **O(e)**.
