## Ejercicio 1: Suma acumulada

### Enunciado

Escribe una función que reciba un entero `n` y devuelva la suma de todos los enteros de 1 a n.

### Modelo de pensamiento

1. Identifica el patrón "acumulador": necesitas una variable que guarde el resultado parcial y se actualice en cada vuelta del ciclo. Antes de escribir el `for`, pregúntate: ¿qué valor debe tener esa variable **antes** de empezar a sumar? (para una suma, el "elemento neutro" es 0).
2. Decide el rango de recorrido: si quieres incluir el número `n`, recuerda que `range(1, n)` **no** incluye a `n` — necesitas `range(1, n + 1)`.
3. En cada vuelta, la regla es siempre la misma: `acumulador = acumulador + elemento_actual`.
4. Al terminar el ciclo, el acumulador ya tiene el resultado final — no hay que hacer nada especial "después", solo devolverlo.

```python
def suma_hasta(n: int) -> int:
    acumulador = 0
    for numero in range(1, n + 1):
        acumulador += numero
    return acumulador

print(suma_hasta(5))    # 1+2+3+4+5 = 15
print(suma_hasta(100))  # 5050
```

## Ejercicio 2: Contar elementos que cumplen una condición

### Enunciado

Dada una lista de notas, cuenta cuántas son aprobatorias (`>= 3.0`).

### Modelo de pensamiento

1. Este problema combina dos patrones que ya conoces por separado: recorrer una lista (ciclo) y decidir con una condición (`if`). La clave es notar que el `if` vive **dentro** del `for`: en cada vuelta primero preguntas, luego decides si acumular.
2. El acumulador aquí es un **contador**, no una suma: empieza en 0, y solo sube de a uno cuando la condición es verdadera (no en cada vuelta).
3. Recorre la lista con `for nota in notas` — no necesitas el índice, solo el valor, así que no uses `range(len(...))` a menos que lo necesites explícitamente.

```python
def contar_aprobados(notas: list[float]) -> int:
    contador = 0
    for nota in notas:
        if nota >= 3.0:
            contador += 1
    return contador

print(contar_aprobados([4.5, 2.0, 3.0, 1.5, 5.0]))  # 3
```

## Ejercicio 3: Máximo sin usar max()

### Enunciado

Dada una lista de enteros no vacía, encuentra el valor máximo sin usar la función `max()`.

### Modelo de pensamiento

1. El patrón mental es "el mejor hasta ahora": necesitas una variable que guarde el mayor valor visto hasta el momento en el recorrido.
2. ¿Con qué la inicializas? La opción más segura es con el **primer** elemento de la lista (no con 0 — si todos los números fueran negativos, inicializar en 0 daría una respuesta incorrecta).
3. En cada vuelta, compara el elemento actual contra el mejor guardado hasta ahora: si el actual es mayor, reemplaza al guardado.
4. Este patrón es idéntico al de "contar aprobados": la única diferencia está en qué haces dentro del `if` (reemplazar vs. contar).

```python
def encontrar_maximo(numeros: list[int]) -> int:
    mejor_hasta_ahora = numeros[0]
    for numero in numeros[1:]:
        if numero > mejor_hasta_ahora:
            mejor_hasta_ahora = numero
    return mejor_hasta_ahora

print(encontrar_maximo([3, 7, 2, 9, 4]))  # 9
```

## Ejercicio 4: Meses para alcanzar una meta de ahorro

### Enunciado

Escribe una función que reciba una lista de depósitos mensuales y una meta, y devuelva cuántos meses se necesitan para alcanzar o superar la meta, sumando depósito a depósito en orden. Si la lista se acaba antes de alcanzar la meta, devuelve `-1`.

### Modelo de pensamiento

1. Aquí no sabemos de antemano cuántas vueltas vamos a dar — depende de cuándo se cumpla la condición "acumulado >= meta". Esa es la señal de que se necesita `while` y no `for`.
2. Necesitas dos cosas que cambien juntas en cada vuelta: el acumulado (dinero guardado) y el índice del mes que estás revisando.
3. La condición del `while` debe cubrir **dos** motivos para detenerse: (a) ya alcanzaste la meta, o (b) ya no quedan más depósitos por revisar. Si te detienes por (b) sin haber alcanzado la meta, ese es el caso especial que devuelve -1.
4. Plantea primero "mientras siga sin cumplirse la condición de éxito Y todavía haya depósitos", y dentro decide al final si terminaste por éxito o por agotamiento.

```python
def meses_para_meta(depositos: list[float], meta: float) -> int:
    acumulado = 0.0
    indice = 0
    while acumulado < meta and indice < len(depositos):
        acumulado += depositos[indice]
        indice += 1

    if acumulado >= meta:
        return indice
    else:
        return -1

print(meses_para_meta([100, 150, 200, 300], 400))  # 3
print(meses_para_meta([100, 100], 1000))            # -1
```

## Ejercicio 5: Ciclo anidado: tabla de multiplicar

### Enunciado

Imprime la tabla de multiplicar del 1 al `hasta`, en forma de tabla: para cada número de fila, imprime la fila completa de productos separados por un tabulador.

### Modelo de pensamiento

1. Cuando hay dos "for" en el enunciado (uno para las filas, otro para las columnas de cada fila), necesitas un ciclo dentro de otro. El de afuera controla en qué fila estás; el de adentro recorre completa esa fila antes de pasar a la siguiente.
2. Es clave no confundir las dos variables: dales nombres distintos (`fila` y `columna`) para no perderte cuál controla qué.
3. Piensa en el ciclo interno como "todo lo que pasa dentro de una sola fila" — se ejecuta completo, de principio a fin, antes de que el ciclo externo avance una vuelta.
4. Para imprimir varios valores en la misma línea antes de saltar de fila, acumula un texto dentro del ciclo interno y solo imprime al terminar esa fila (fuera del ciclo interno, dentro del externo).

```python
def imprimir_tabla(hasta: int) -> None:
    for fila in range(1, hasta + 1):
        texto_fila = ""
        for columna in range(1, hasta + 1):
            texto_fila += f"{fila * columna}\t"
        print(texto_fila)

imprimir_tabla(5)
```

## Ejercicio 6: Búsqueda lineal con corte anticipado

### Enunciado

Escribe una función que reciba una lista de enteros y un valor objetivo, y devuelva la posición (índice) donde aparece por primera vez, o `-1` si no está.

### Modelo de pensamiento

1. Este problema tiene una particularidad: apenas encuentras lo que buscas, no tiene sentido seguir recorriendo el resto de la lista — puedes "cortar" el ciclo ahí mismo con `break`.
2. Necesitas el índice, no solo el valor, así que aquí sí conviene recorrer con `for indice in range(len(lista))`.
3. Plantea el caso por defecto **antes** del ciclo: asume que no lo vas a encontrar (`resultado = -1`), y solo cambias esa suposición si efectivamente lo encuentras.
4. Este es el mismo patrón de "bandera": una variable que arranca en un valor por defecto y se actualiza solo si algo especial ocurre durante el recorrido.

```python
def buscar_posicion(lista: list[int], objetivo: int) -> int:
    posicion_encontrada = -1
    for indice in range(len(lista)):
        if lista[indice] == objetivo:
            posicion_encontrada = indice
            break
    return posicion_encontrada

print(buscar_posicion([4, 8, 15, 16, 23], 16))  # 3
print(buscar_posicion([4, 8, 15, 16, 23], 99))  # -1
```

## Ejercicio 7: Resumen de notas: aprobados, reprobados y promedio (reto)

### Enunciado

Dada una lista de notas, calcula en una sola función cuántas son aprobatorias, cuántas reprobatorias, y el promedio general. Devuelve los tres valores.

### Modelo de pensamiento

1. Identifica que necesitas **tres acumuladores en paralelo**, actualizados dentro del mismo recorrido: contador de aprobados, contador de reprobados, y suma total (para el promedio).
2. Antes del ciclo, inicializa los tres en su "valor neutro" (0).
3. Dentro del ciclo, cada nota participa en la suma **siempre** (para el promedio), pero solo suma a uno de los dos contadores, según el resultado del `if`.
4. El promedio se calcula **una sola vez**, al final, dividiendo la suma acumulada entre la cantidad total de notas (no dentro del ciclo).
5. Aquí anticipamos algo que resolveremos mejor con funciones (Taller 4): por ahora, devuelve los tres valores juntos como una tupla.

```python
def resumen_notas(notas: list[float]) -> tuple[int, int, float]:
    aprobados = 0
    reprobados = 0
    suma = 0.0

    for nota in notas:
        suma += nota
        if nota >= 3.0:
            aprobados += 1
        else:
            reprobados += 1

    promedio = suma / len(notas)
    return aprobados, reprobados, promedio

aprobados, reprobados, promedio = resumen_notas([4.5, 2.0, 3.0, 1.5, 5.0])
print(f"Aprobados: {aprobados}, Reprobados: {reprobados}, Promedio: {promedio:.2f}")
```
