## Ejercicio 1: Una clase simple: Estudiante

### Enunciado

Define una clase `Estudiante` con atributos `nombre: str` y `notas: list[float]`, inicializados en el constructor. Crea una instancia con un nombre y una lista de notas, e imprime sus atributos.

### Modelo de pensamiento

1. Una clase es un "molde" para crear objetos que agrupan datos relacionados (aquí, nombre y notas) bajo un mismo nombre, en vez de manejarlos como variables sueltas y separadas.
2. El método `__init__` se ejecuta automáticamente al crear un objeto nuevo (`Estudiante(...)`) y es responsable de dejar listos sus atributos iniciales — es como una "receta de armado" del objeto.
3. `self` es el nombre (por convención) que usamos dentro de la clase para referirnos "al objeto que se está construyendo o usando en este momento". `self.nombre = nombre` significa "guarda el valor del parámetro `nombre`, en el atributo `nombre` de ESTE objeto en particular".
4. Cada objeto que crees a partir de la clase tiene su propia copia de los atributos — dos estudiantes distintos no comparten la misma lista de notas.

```python
class Estudiante:
    def __init__(self, nombre: str, notas: list[float]) -> None:
        self.nombre = nombre
        self.notas = notas

ana = Estudiante("Ana", [4.5, 4.8, 4.0])
print(ana.nombre)
print(ana.notas)
```

## Ejercicio 2: Métodos que usan los atributos

### Enunciado

Agrega a la clase `Estudiante` un método `promedio() -> float` que calcule el promedio de `self.notas`, y un método `aprobado() -> bool` que use `promedio()` para decidir si el promedio es >= 3.0.

### Modelo de pensamiento

1. Un método es una función que vive dentro de la clase y que, a través de `self`, tiene acceso directo a los datos del objeto (`self.notas`) sin que se los tengas que pasar como parámetro explícito — el objeto "ya sabe" cuáles son sus propias notas.
2. `promedio()` no necesita ningún parámetro adicional (aparte de `self`) porque toda la información que necesita ya está guardada en el objeto mismo.
3. Un método puede llamar a otro método del mismo objeto usando `self.nombre_del_metodo()` — así, `aprobado()` no repite la lógica de sumar y dividir; reutiliza `self.promedio()`, igual que en el Taller 4 una función usaba a otra.

```python
class Estudiante:
    def __init__(self, nombre: str, notas: list[float]) -> None:
        self.nombre = nombre
        self.notas = notas

    def promedio(self) -> float:
        return sum(self.notas) / len(self.notas)

    def aprobado(self) -> bool:
        return self.promedio() >= 3.0

ana = Estudiante("Ana", [4.5, 4.8, 4.0])
print(ana.promedio())   # 4.43...
print(ana.aprobado())   # True
```

## Ejercicio 3: Un método que modifica el estado interno

### Enunciado

Agrega un método `agregar_nota(nueva_nota: float) -> None` que agregue una nota a `self.notas`. Crea un estudiante con una lista inicial de notas, agrégale una nota nueva, y muestra cómo cambia su promedio antes y después.

### Modelo de pensamiento

1. Hasta ahora, los métodos solo **leían** datos del objeto (promedio, aprobado); este método, en cambio, **modifica** el estado interno del objeto — después de llamarlo, el objeto "recuerda" el cambio en llamadas futuras.
2. Esta es una diferencia clave frente a las funciones del taller anterior: una función normal no "recuerda" nada entre una llamada y otra (cada llamada empieza de cero), pero un objeto sí, porque el cambio queda guardado en sus atributos.
3. Para modificar una lista que ya es un atributo, usa `self.notas.append(nueva_nota)` — estás modificando la lista que vive dentro del objeto, no creando una lista nueva.
4. El método no necesita devolver nada (`-> None`) porque su propósito no es calcular un valor, sino cambiar el estado del objeto.

```python
class Estudiante:
    def __init__(self, nombre: str, notas: list[float]) -> None:
        self.nombre = nombre
        self.notas = notas

    def promedio(self) -> float:
        return sum(self.notas) / len(self.notas)

    def aprobado(self) -> bool:
        return self.promedio() >= 3.0

    def agregar_nota(self, nueva_nota: float) -> None:
        self.notas.append(nueva_nota)

ana = Estudiante("Ana", [4.5, 4.8, 4.0])
print("Antes:", ana.promedio())
ana.agregar_nota(2.0)
print("Después:", ana.promedio())
```

## Ejercicio 4: De función suelta a método

### Enunciado

En el Taller 4, `contar_aprobados(notas)` recibía una lista de notas sueltas como parámetro. Compara esa versión con el método `aprobado()` que ya tienes en `Estudiante`, y explica en un comentario por qué este último ya no necesita recibir las notas como parámetro.

### Modelo de pensamiento

1. Compara las dos versiones: la función suelta necesitaba que le pasaran la lista de notas cada vez que se llamaba (`contar_aprobados(lista_de_ana)`); el método, en cambio, ya vive "adentro" del objeto, así que accede a `self.notas` directamente.
2. Esta comparación es la esencia de "por qué objetos": cuando una función siempre se usa junto con los mismos datos (una lista de notas asociada a un estudiante en particular), tiene más sentido que esos datos y esa función vivan juntos, dentro de una misma unidad (la clase), en vez de pasarlos como argumento cada vez.
3. En términos prácticos: lo que antes era una función global aplicable a cualquier lista, ahora es una capacidad específica del objeto `Estudiante` — solo tiene sentido preguntarle "¿estás aprobado?" a un estudiante concreto.

```python
# Versión función suelta (Taller 4):
# def contar_aprobados(notas: list[float]) -> int: ...
# Requería que le pasaras la lista de notas explícitamente cada vez.

class Estudiante:
    def __init__(self, nombre: str, notas: list[float]) -> None:
        self.nombre = nombre
        self.notas = notas

    def promedio(self) -> float:
        return sum(self.notas) / len(self.notas)

    def aprobado(self) -> bool:
        # Ya no recibe "notas" como parámetro: el objeto ya sabe
        # cuáles son sus propias notas a través de self.notas.
        return self.promedio() >= 3.0

ana = Estudiante("Ana", [4.5, 4.8, 4.0])
print(ana.aprobado())
```

## Ejercicio 5: Lista de objetos: estadísticas grupales

### Enunciado

Crea una lista de varios objetos `Estudiante`. Escribe una función (fuera de la clase) `mejor_estudiante(estudiantes) -> Estudiante` que devuelva el objeto (no solo el nombre) con el mejor promedio.

### Modelo de pensamiento

1. Una lista de objetos es igual, en mecánica, a cualquier otra lista que ya conoces (de números o de strings) — la diferencia es que cada elemento ahora es "más rico": trae consigo sus propios datos y sus propios métodos.
2. Al recorrer la lista con `for estudiante in estudiantes`, cada `estudiante` ya es un objeto completo — puedes llamar `estudiante.promedio()` directamente dentro del ciclo, sin ir a buscar sus notas por separado.
3. El patrón sigue siendo "el mejor hasta ahora" del Taller 2, pero ahora lo que guardas como "el mejor" es el objeto completo (no solo su nombre o su promedio) — así después puedes seguir usando cualquiera de sus atributos o métodos.
4. Esta función queda fuera de la clase porque no pertenece a un solo estudiante — actúa sobre una colección de ellos.

```python
def mejor_estudiante(estudiantes: list[Estudiante]) -> Estudiante:
    mejor = estudiantes[0]
    for estudiante in estudiantes[1:]:
        if estudiante.promedio() > mejor.promedio():
            mejor = estudiante
    return mejor

lista_estudiantes = [
    Estudiante("Ana", [4.5, 4.8, 4.0]),
    Estudiante("Luis", [3.0, 2.5, 3.5]),
    Estudiante("Marta", [5.0, 4.9, 4.7]),
]

top = mejor_estudiante(lista_estudiantes)
print(top.nombre, top.promedio())
```

## Ejercicio 6: Composición: un Curso que contiene Estudiantes

### Enunciado

Define una clase `Curso` con un atributo `nombre: str` y una lista interna de estudiantes (`list[Estudiante]`), inicialmente vacía. Agrega un método `agregar_estudiante(estudiante) -> None` y un método `promedio_curso() -> float` que calcule el promedio general del curso.

### Modelo de pensamiento

1. Un objeto puede tener, entre sus atributos, a otros objetos (aquí, `Curso` "tiene" una lista de `Estudiante`) — a esto se le llama **composición**: un objeto está hecho, en parte, de otros objetos.
2. `self.estudiantes` empieza como una lista vacía en `__init__` (no la recibes como parámetro obligatorio) porque un curso nuevo arranca sin estudiantes inscritos todavía — se van agregando después con `agregar_estudiante`.
3. `agregar_estudiante` es exactamente el mismo patrón de `agregar_nota` del ejercicio 3 (modificar una lista interna con `.append`), solo que ahora la lista contiene objetos completos en vez de números.
4. Para `promedio_curso()`, reutiliza lo que ya sabes: recorre `self.estudiantes`, y para cada uno usa `estudiante.promedio()` (que el objeto ya sabe calcular) en vez de volver a sumar y dividir sus notas desde cero.

```python
class Curso:
    def __init__(self, nombre: str) -> None:
        self.nombre = nombre
        self.estudiantes: list[Estudiante] = []

    def agregar_estudiante(self, estudiante: Estudiante) -> None:
        self.estudiantes.append(estudiante)

    def promedio_curso(self) -> float:
        suma_promedios = 0.0
        for estudiante in self.estudiantes:
            suma_promedios += estudiante.promedio()
        return suma_promedios / len(self.estudiantes)

ada = Curso("Análisis y Diseño de Algoritmos")
ada.agregar_estudiante(Estudiante("Ana", [4.5, 4.8, 4.0]))
ada.agregar_estudiante(Estudiante("Luis", [3.0, 2.5, 3.5]))

print(ada.promedio_curso())
```

## Ejercicio 7: Reporte completo del curso (reto)

### Enunciado

Agrega a `Curso` un método `reporte() -> dict[str, Any]` que devuelva un diccionario con `"nombre_curso"`, `"cantidad_estudiantes"`, `"promedio_curso"` y `"mejor_estudiante"` (el nombre). Pruébalo creando un curso con al menos tres estudiantes.

### Modelo de pensamiento

1. Este reto es, en esencia, el mismo `resumen_curso` del Taller 4 — pero ahora vive dentro de la clase `Curso`, como un método más, porque toda la información que necesita (`self.estudiantes`) ya está disponible ahí.
2. Antes de escribir el método completo, identifica qué piezas ya tienes resueltas y solo necesitas reutilizar: la cantidad de estudiantes es `len(self.estudiantes)`; el promedio del curso ya lo calcula `self.promedio_curso()`; el mejor estudiante lo puedes obtener con la función `mejor_estudiante(self.estudiantes)` del ejercicio 5.
3. Este es el mismo principio de "descomponer y reutilizar" del Taller 4, ahora aplicado dentro de una clase: un método no tiene que volver a calcular desde cero algo que otro método (o función) ya sabe hacer.
4. El resultado se organiza en un diccionario con nombres explícitos, igual que en el reto del Taller 4, para que sea fácil de leer e interpretar por quien lo consuma después.

```python
from typing import Any


class Curso:
    def __init__(self, nombre: str) -> None:
        self.nombre = nombre
        self.estudiantes: list[Estudiante] = []

    def agregar_estudiante(self, estudiante: Estudiante) -> None:
        self.estudiantes.append(estudiante)

    def promedio_curso(self) -> float:
        suma_promedios = 0.0
        for estudiante in self.estudiantes:
            suma_promedios += estudiante.promedio()
        return suma_promedios / len(self.estudiantes)

    def reporte(self) -> dict[str, Any]:
        return {
            "nombre_curso": self.nombre,
            "cantidad_estudiantes": len(self.estudiantes),
            "promedio_curso": self.promedio_curso(),
            "mejor_estudiante": mejor_estudiante(self.estudiantes).nombre,
        }

ada = Curso("Análisis y Diseño de Algoritmos")
ada.agregar_estudiante(Estudiante("Ana", [4.5, 4.8, 4.0]))
ada.agregar_estudiante(Estudiante("Luis", [3.0, 2.5, 3.5]))
ada.agregar_estudiante(Estudiante("Marta", [5.0, 4.9, 4.7]))

print(ada.reporte())
```
