# 📌 Variables en Python

## 1️⃣ ¿Qué es una variable?
Una variable es un **nombre asignado a un valor** que se almacena en memoria y puede cambiar a lo largo del programa. No pienses en una variable como una "casilla fija" — piensa en ella como una **etiqueta** que en un momento dado apunta a un valor, y que puedes volver a pegar sobre otro valor distinto en cualquier línea posterior.

### 📌 Reglas para nombrar variables
✅ Puede contener letras, números y guiones bajos (`_`).
✅ Debe comenzar con una **letra** o un guion bajo (`_`).
✅ **No puede** usar palabras reservadas de Python (`def`, `class`, `if`, etc.).
✅ Python es **sensible a mayúsculas y minúsculas** (`edad` ≠ `Edad`).

## 2️⃣ Asignación de variables
```python
# Asignación de variables
nombre = "Juan"      # String (Cadena de texto)
edad = 25            # Entero (int)
altura = 1.75        # Flotante (float)
es_estudiante = True # Booleano (bool)
```

🔹 **Python infiere automáticamente** el tipo de datos.
🔹 Para conocer el tipo de una variable, usamos `type()`.
```python
print(type(nombre))  # <class 'str'>
print(type(edad))    # <class 'int'>
```

---

## 3️⃣ Múltiples asignaciones en una sola línea
```python
x, y, z = 10, 20, 30
a = b = c = "Python"
```

📌 **Intercambio de valores** sin variable auxiliar:
```python
x, y = y, x  # Intercambia los valores de x e y
```

⚠️ Esta línea es un ejemplo perfecto de código que **parece obvio pero engaña** si no lo trazas: mucha gente asume que se ejecuta "de a una asignación", como si `x` cambiara primero y luego se usara ese nuevo valor para `y`. No es así — Python primero evalúa **todo el lado derecho** (`y, x`) con los valores que tenían *antes* de la línea, y solo después asigna. Vas a practicar exactamente este tipo de trampa en la siguiente sección.

## 🧠 Modelo mental: sigue el valor, no la sintaxis

Cuando lees código con varias asignaciones seguidas, **no lo leas de corrido esperando "entender a ojo"**. El hábito que te va a servir en todo el curso (y que vas a necesitar sí o sí para condicionales, ciclos y recursión) es construir una **tabla de traza**: una tabla con una columna por cada variable, y una fila por cada línea que se ejecuta, donde anotas el valor de cada variable después de esa línea.

**Antes de correr el código**, intenta llenar tú la tabla con lápiz y papel (o mentalmente, línea por línea). Solo después corre la celda para comprobar si acertaste.

```python
a = 5
b = 10
a = b
b = a + 1
a, b = b, a
```

| Línea ejecutada | `a` | `b` | Comentario |
|---|---|---|---|
| `a = 5` | `5` | `?` | `b` todavía no existe |
| `b = 10` | `5` | `10` | |
| `a = b` | `10` | `10` | `a` toma el valor **actual** de `b` |
| `b = a + 1` | `10` | `11` | usa el `a` ya actualizado (`10`), no el original (`5`) |
| `a, b = b, a` | `11` | `10` | el lado derecho se evalúa completo (`b, a` → `11, 10`) **antes** de asignar |

📌 **¿Por qué importa este hábito?** Porque cuando el código falla, la mayoría de los estudiantes intenta "leerlo mentalmente muy rápido" y se equivoca por las mismas razones que se equivocaría llenando esta tabla sin cuidado — asumir que una variable "ya cambió" cuando todavía no, o que el orden de evaluación es distinto al real. Trazar a mano fuerza a ser explícito sobre el orden exacto en que Python ejecuta cada instrucción.

```python
x = 3
y = x + 2
x = y - 1
y = x * y
```

````python
---

## 6️⃣ Variables mutables e inmutables
En Python, los **tipos de datos se dividen en mutables e inmutables**:

🔹 **Mutables** (pueden cambiar su valor sin cambiar su identidad en memoria):
✅ `list`, `dict`, `set`

🔹 **Inmutables** (no pueden modificarse después de ser creados):
✅ `int`, `float`, `str`, `tuple`, `bool`

Ejemplo:
```python
# Variable mutable (Lista)
lista = [1, 2, 3]
lista.append(4)  # Modificamos la lista
print(lista)  # [1, 2, 3, 4]

# Variable inmutable (Tupla)
tupla = (1, 2, 3)
# tupla.append(4)  # ❌ Esto daría error porque las tuplas son inmutables
```

⚠️ Esto también afecta cómo debes trazar el estado a mano: si una variable apunta a una lista y **otra variable apunta a la misma lista**, modificar una de las dos también cambia lo que ve la otra — porque ambas etiquetas apuntan al mismo objeto en memoria, no a copias independientes. Trázalo:
```python
a = [1, 2, 3]
b = a          # b apunta a la MISMA lista que a, no a una copia
b.append(4)
print(a)       # ¿qué imprime? ¿[1, 2, 3] o [1, 2, 3, 4]?
```
📌 Si tu tabla de traza solo tiene una columna "valor" por variable, este ejemplo te va a hacer fallar la predicción — para listas, diccionarios y sets necesitas pensar en términos de "¿a qué objeto apunta cada nombre?", no solo "¿qué valor tiene?".

---

## 7️⃣ Scope (alcance) de variables
El **ámbito o alcance** de una variable define dónde puede ser accedida dentro del código.

### 📌 Tipos de scope
1. **Local:** Definida dentro de una función y solo accesible dentro de ella.
2. **Global:** Definida fuera de una función y accesible en todo el código.
3. **Nonlocal:** Se usa dentro de funciones anidadas para modificar variables de funciones superiores.

Ejemplo:
```python
x = "Global"

def funcion():
    x = "Local"
    print("Dentro de la función:", x)

funcion()
print("Fuera de la función:", x)
```
📌 **Si queremos modificar la variable global dentro de la función:**
```python
x = "Global"

def funcion():
    global x  # Indica que queremos modificar la variable global
    x = "Modificada dentro de la función"

funcion()
print(x)  # Ahora 'x' es "Modificada dentro de la función"
```

📌 **Uso de `nonlocal` en funciones anidadas:**
```python
def externa():
    x = "Estoy en externa"

    def interna():
        nonlocal x  # Modifica la variable en la función contenedora
        x = "Modificada en interna"

    interna()
    print(x)

externa()
```

---

## 8️⃣ Constantes (buena práctica)
Python **no** tiene constantes como otros lenguajes, pero por convención usamos nombres en mayúsculas:
```python
PI = 3.1416
GRAVEDAD = 9.81
```
Aunque técnicamente se pueden modificar, **por convención** no debemos hacerlo.

---

## 9️⃣ Variables con `None`
`None` representa la ausencia de valor (similar a `null` en otros lenguajes):
```python
x = None
if x is None:
    print("x no tiene valor aún")
```

---

## 🔟 Buenas prácticas de nombres
✅ **Usar nombres descriptivos:**
❌ `x = 100`
✅ `precio_producto = 100`

✅ **Seguir el estilo `snake_case` para variables:**
```python
total_factura = 500
nombre_cliente = "Pedro"
```

---

## 🎯 Ejemplo completo
```python
# Definir variables
nombre_cliente = "Ana"
saldo_inicial = 1000.50
tasa_interes = 0.02
es_vip = True

# Actualización de saldo
nuevo_saldo = saldo_inicial * (1 + tasa_interes)

# Mostrar información
print(f"Cliente: {nombre_cliente}")
print(f"Saldo inicial: ${saldo_inicial}")
print(f"Nuevo saldo con interés: ${nuevo_saldo:.2f}")
print(f"Cliente VIP: {es_vip}")
```

📌 Antes de correrlo, trázalo: ¿cuál es el valor final de `nuevo_saldo`? Escríbelo en papel, luego corre la celda y verifica.

---
````
