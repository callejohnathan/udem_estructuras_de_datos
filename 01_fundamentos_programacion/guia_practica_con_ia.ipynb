# 🧠 Cómo Estudiar Programación con IA (ChatGPT y similares)

Las herramientas de IA pueden ser poderosas para **aprender programación**, pero para aprovecharlas hay que formular buenos **prompts** y enfocarse en **el proceso**, no en obtener la respuesta más rápida posible.

📌 **El riesgo real de usar IA mal en este curso:** si le pides código directamente, obtienes una solución que memorizas la *forma* de replicar, sin entender cuándo aplica ni por qué funciona — que es exactamente el problema que más le cuesta superar a los estudiantes. Usada bien, la IA puede hacer lo contrario: forzarte a planear, cuestionarte, y darte ejercicios a tu medida.

---

## 1️⃣ Enfócate en el concepto, no en la solución

### ❌ Mala práctica (pedir la solución directamente)
> "Dame el código en Python para ordenar una lista con el algoritmo de burbuja."

### ✅ Buena práctica (enfocarse en el concepto)
> "Explícame paso a paso cómo funciona el algoritmo de burbuja y qué conceptos de programación se aplican en su implementación."

📌 Aprendes el **proceso lógico** detrás del código y qué conceptos clave está usando, en vez de memorizar 15 líneas sin entenderlas.

---

## 2️⃣ Pide ayuda para *planear* antes de programar

Esta es la aplicación más directa de lo que vas a practicar en toda la parte de fundamentos del curso: **antes de escribir código, planea la solución en papel o mentalmente** (trázala, descríbela en una frase, identifica casos borde). La IA puede ayudarte en *esa* etapa, no solo en la de escribir código.

### ✅ Prompts útiles para planear (no para que te escriban código)
> "Tengo este problema: [descríbelo]. Ayúdame a identificar qué casos borde debería considerar antes de programarlo — no me des el código todavía."

> "Voy a resolver esto con un ciclo `while`. Ayúdame a construir una tabla de traza para 2-3 iteraciones con esta entrada de ejemplo: [...] — otra vez, sin código, solo la tabla."

> "Este es mi plan en palabras para resolver el problema: [tu plan]. ¿Ves algún hueco o caso que no estoy considerando?"

📌 **Por qué importa:** si le pides directamente el código a la IA, terminas en el mismo lugar que memorizar una solución del profesor — una forma que replicas sin entender. Si en cambio usas la IA para **presionar tu propio plan** (¿pensaste en este caso borde? ¿tu tabla de traza está completa?), practicas exactamente el hábito de planear antes de codear.

---

## 3️⃣ Pide ejercicios progresivos y ajusta el ritmo a tu nivel

Aprender programación es un proceso escalonado. Pedir ejercicios con dificultad progresiva —y ajustarla cuando no calza con tu nivel— te ayuda a construir una base sólida antes de enfrentar problemas más complejos.

### ✅ Cómo pedir progresión
> "Dame un ejercicio de nivel básico sobre listas en Python. Luego, dame un ejercicio intermedio basado en la misma temática, pero con más dificultad."

### ✅ Cómo ajustar cuando un ejercicio no calza
> "Este ejercicio me parece muy fácil, ¿puedes hacer que sea más desafiante sin salirme del tema?"
> "Este problema me parece demasiado difícil, ¿puedes dividirlo en pasos más pequeños?"

### ✅ Cómo pedir que te expliquen algo por partes, no todo de golpe
> "Explícame cómo funciona la recursión con un ejemplo simple. Cuando confirme que lo entendí, dame un problema más difícil para practicar."

📌 **Beneficio:** evitas dos extremos igual de improductivos — frustrarte con algo muy avanzado, o aburrirte con algo que ya dominas.

---

## 4️⃣ Pide revisión de código sin obtener la solución

Para mejorar tu código, en lugar de pedir la solución correcta, pídele a la IA que lo **revise y sugiera mejoras** sin mostrarte la respuesta directa.

### ✅ Cómo pedirlo
> "Este es mi código, ¿puedes señalar errores o sugerir mejoras sin decirme la solución correcta?"

**Ejemplo de interacción:**
🔹 Tu código:
```python
def suma(a, b):
    print(a + b)  # Creo que falta algo aquí...
```
🔹 Revisión esperada:
> "Parece que estás usando `print()` en lugar de `return`. Esto puede ser un problema si necesitas reutilizar el resultado de la función en otro lugar."

📌 Esto refuerza la depuración activa: te acostumbras a encontrar y corregir errores por ti mismo, con pistas, no con la respuesta servida.

---

## 5️⃣ Pregunta "por qué" y no solo "cómo"

Cuando pides una explicación, es clave preguntar **por qué** se usa cierto enfoque, no solo **cómo** implementarlo.

### ✅ Ejemplo
> "¿Por qué es más eficiente usar un diccionario en lugar de una lista en este caso?"

📌 Aprendes el **razonamiento** detrás de las decisiones — cuándo y por qué usar cada estructura de datos — que es justo lo que necesitas para *reconocer* cuándo aplica un concepto en un problema nuevo, en vez de reconocer solo "la forma" de una solución que ya viste.

---

## 6️⃣ Pide que te cuestione: evaluación activa y simulacros

Una de las formas más efectivas de aprender es **ser desafiado con preguntas**, en vez de solo recibir información.

### ✅ Evaluación de conocimiento
> "Pregúntame sobre listas en Python y evalúa si mi respuesta es correcta."

### ✅ Simulacro de entrevista técnica
> "Hazme una pregunta de entrevista técnica sobre estructuras de datos. Luego revisa mi respuesta y dime en qué puedo mejorar."

📌 Te obliga a razonar y explicar en tus propias palabras antes de confirmar si estás en lo correcto — y practicas explicar código, que es una habilidad distinta (y necesaria) a solo escribirlo.

---

## 7️⃣ Usa ejemplos de la vida real

Relacionar la programación con situaciones cotidianas ayuda a entender mejor los conceptos abstractos.

### ✅ Cómo pedirlo
> "Explícame cómo funcionan las colas en programación con un ejemplo de la vida real."

---

## 🔥 Conclusión: cómo aprender de forma activa con IA

✔ **Enfócate en el proceso, no en la solución.**
✔ **Usa la IA para presionar tu plan (traza, contrato, casos borde) antes de escribir código, no para que te dé el código.**
✔ **Pide ejercicios progresivos y ajusta la dificultad a tu ritmo.**
✔ **Solicita revisión de código sin recibir la respuesta directa.**
✔ **Pregunta "por qué", no solo "cómo".**
✔ **Deja que te cuestione y evalúe — practicar explicar es tan importante como practicar programar.**
