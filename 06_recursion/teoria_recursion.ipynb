# 🔄 Introducción a la Recursión

## ¿Qué es la recursión?
La **recursión** es una técnica para resolver un problema haciendo que una función se llame a sí misma sobre una versión **más pequeña** del mismo problema, hasta llegar a un caso tan simple que se puede resolver directamente.

### Analogía: las muñecas matrioskas
Imagina una serie de muñecas rusas, cada una con una copia más pequeña adentro. Para llegar a la más pequeña no necesitas saber de antemano cuántas muñecas hay ni memorizar el proceso completo: solo necesitas una regla simple, repetida — *"abre la muñeca actual; si hay otra adentro, repite el proceso con ella; si no hay otra adentro, ya llegaste al fondo"*. Cerrarlas de nuevo es el camino de regreso.

## 🧠 El cambio de mentalidad más importante: el "salto de fe"
La razón por la que la recursión se siente confusa al principio es que la mayoría de estudiantes intenta **desenrollar mentalmente TODO el árbol de llamadas** para poder "confiar" en que la función funciona. Eso alcanza a hacerse con 2 o 3 llamadas, pero se vuelve imposible (y agotador) en problemas reales — y es exactamente lo que lleva a bloquearse: demasiado código, demasiadas llamadas, cero claridad.

La forma correcta de pensar la recursión es dar un **salto de fe**:

> Si asumo que `funcion(problema_más_pequeño)` YA resuelve correctamente ese subproblema — sin importarme cómo lo hace por dentro —, ¿cómo uso ese resultado para construir la solución de `funcion(problema_actual)`?

Esto no es "fe ciega": es el mismo principio de la **inducción matemática**. No verificas manualmente cada caso; confías en la definición y solo verificas dos cosas:
1. El caso base es correcto por sí solo.
2. El caso recursivo es correcto **asumiendo que la llamada recursiva ya funciona**.

Si esas dos cosas son ciertas, la función funciona para **cualquier tamaño de entrada**, sin que tengas que trazarla mentalmente completa cada vez.

⚠️ Esto no significa que nunca vayas a trazar una recursión a mano — al contrario, en este notebook vas a practicar trazar varios ejemplos completos a mano. La diferencia está en *cuándo* y *para qué* lo haces: lo haces **una o dos veces por tipo de problema**, para verificar tu diseño y construir intuición — no como sustituto de confiar en la definición cada vez que escribes una función recursiva nueva.

### Partes de toda función recursiva
1. **Caso base** (obligatorio ✅): la condición que detiene la recursión y evita que se convierta en un bucle infinito.
2. **Caso recursivo**: donde la función se llama a sí misma con un problema más pequeño.
3. **Convergencia**: cada llamada debe acercarse al caso base, asegurando que la recursión finalice en algún momento.

## 🧭 El marco de trabajo: los 6 pasos

### 🔍 Aplicando el marco: `imprimir_hasta_n`

Antes de ver el código, apliquemos los pasos 2 a 5 al problema "imprimir los números de 1 a `n`":

**Paso 2 — en términos de sí mismo:** imprimir de `i` a `n` es "imprimir `i`, y luego imprimir de `i+1` a `n`" (el subproblema es sobre lo que falta por imprimir).

**Paso 3 — caso base:** cuando `i > n` ya no queda nada por imprimir — no hago nada y retorno.

**Paso 4 — caso recursivo:** imprimo `i`, y delego el resto (`i+1` hasta `n`) a la misma función.

**Paso 5 — traza a mano para `imprimir_hasta_n(3)`** (empezando en `i=1`):

```
imprimir_hasta_n(n=3, i=1)
│  print(1)
└─ imprimir_hasta_n(n=3, i=2)
   │  print(2)
   └─ imprimir_hasta_n(n=3, i=3)
      │  print(3)
      └─ imprimir_hasta_n(n=3, i=4)
         │  i > n  →  caso base, no imprime nada, retorna None
```

Cada línea con `│` es un **frame activo** en la pila; entre más abajo en el dibujo, más "adentro" de la pila está. Como esta función no combina ningún valor de retorno (no hay `return valor` que use el resultado de la llamada recursiva), la traza es puramente "hacia adentro" — no queda trabajo pendiente al volver hacia afuera. Eso va a cambiar en el siguiente ejemplo.

**Paso 6 — código:**

````python
def imprimir_hasta_n(n: int, i: int = 1) -> None:
  #caso base -> restricción que me hace frenar
  if(i > n):
    return None

  #caso recursivo -> lo que repito --> solución a mi subproblema
  print(i)
  imprimir_hasta_n(n, i+1)

imprimir_hasta_n(3)
````

### 💭 ¿Ya diste el salto de fe?
Fíjate que para escribir `imprimir_hasta_n` no necesitaste pensar en la pila completa: solo necesitabas confiar en que `imprimir_hasta_n(n, i+1)` iba a imprimir correctamente el resto de los números. La traza de arriba fue para *verificar* esa confianza una vez, no para diseñarla.

Ahora vamos a un segundo ejemplo lineal, pero esta vez con algo nuevo: trabajo pendiente **después** de la llamada recursiva.

---
## 📐 Ejemplo completo: Factorial, aplicando los 6 pasos

**Problema:** calcular `n! = n × (n-1) × (n-2) × ... × 1`, con `0! = 1`.

**Paso 1 — salto de fe:** voy a asumir que `factorial(n-1)` YA calcula correctamente `(n-1)!`. No me pregunto cómo lo hace por dentro.

**Paso 2 — en términos de sí mismo:** *"Si ya supiera calcular `(n-1)!`, entonces `n!` es simplemente `n` multiplicado por ese resultado."*

**Paso 3 — caso base:** `factorial(0) = 1` (por definición matemática; es la entrada más pequeña que puedo responder sin recursión).

**Paso 4 — caso recursivo:** `factorial(n) = n * factorial(n - 1)`.

**Paso 5 — traza a mano para `factorial(4)`:**

A diferencia de `imprimir_hasta_n`, aquí SÍ hay trabajo pendiente después de la llamada recursiva (la multiplicación por `n`), así que hay que trazar tanto la bajada como la subida:

```
factorial(4)
│  necesita: 4 * factorial(3)
└─ factorial(3)
   │  necesita: 3 * factorial(2)
   └─ factorial(2)
      │  necesita: 2 * factorial(1)
      └─ factorial(1)
         │  necesita: 1 * factorial(0)
         └─ factorial(0)
            │  caso base → retorna 1
         ← factorial(1) retorna 1 * 1 = 1
      ← factorial(2) retorna 2 * 1 = 2
   ← factorial(3) retorna 3 * 2 = 6
← factorial(4) retorna 4 * 6 = 24
```

Cada `└─` es un frame que se **apila** (bajada, con parámetros nuevos); cada `←` es un frame que **retorna un valor y se libera** (subida). La multiplicación por `n` solo puede hacerse DESPUÉS de que la llamada recursiva ya retornó — por eso el orden de las multiplicaciones va de adentro hacia afuera.

**Paso 6 — código:**

````python
# Función recursiva para calcular el factorial
def factorial(n: int) -> int:
    if n == 0:  # Caso base
        return 1
    return n * factorial(n - 1)  # Caso recursivo

print(factorial(4))  # 24
print(factorial(5))  # 120
````

# 🔄 Uso del `return` en Recursión y el Flujo de Valores

## 📝 Importancia del `return` en Recursión
En la programación recursiva, el uso de `return` es fundamental para que los valores se propaguen correctamente desde las llamadas internas hasta la llamada original. Cada llamada recursiva crea un nuevo **frame** en la pila (**call stack**), y `return` es lo que permite que un valor calculado "adentro" viaje hacia "afuera".

Esto es exactamente lo que trazaste a mano en el ejemplo de `factorial`: la bajada construye los frames, y `return` es lo que dispara la subida.

---

## 🎓 Ejemplo: Suma de los primeros `n` números naturales

Apliquemos el marco otra vez, rápido:
- **Caso base:** `suma_n(0) = 0`.
- **Caso recursivo:** `suma_n(n) = n + suma_n(n - 1)`.

**Traza a mano para `suma_n(3)`:**
```
suma_n(3)
│  necesita: 3 + suma_n(2)
└─ suma_n(2)
   │  necesita: 2 + suma_n(1)
   └─ suma_n(1)
      │  necesita: 1 + suma_n(0)
      └─ suma_n(0)
         │  caso base → retorna 0
      ← suma_n(1) retorna 1 + 0 = 1
   ← suma_n(2) retorna 2 + 1 = 3
← suma_n(3) retorna 3 + 3 = 6
```

```python
def suma_n(n):
    if n == 0:  # Caso base
        return 0
    return n + suma_n(n - 1)  # Llamada recursiva con propagación de valores

print(suma_n(5))  # 15
```

## 🌐 Ejemplo: Inversión de una cadena

- **Caso base:** cuando el índice llega al último carácter, ese carácter es el "primero" del reverso.
- **Caso recursivo:** el reverso de `cadena` desde `indice` es "el reverso del resto" + "el carácter en `indice`".

**Traza a mano para `invertir_cadena("sol")`:**
```
invertir_cadena("sol", 0)
│  necesita: invertir_cadena("sol", 1) + "s"
└─ invertir_cadena("sol", 1)
   │  necesita: invertir_cadena("sol", 2) + "o"
   └─ invertir_cadena("sol", 2)
      │  índice == len(cadena)-1 → caso base, retorna "l"
   ← invertir_cadena("sol", 1) retorna "l" + "o" = "lo"
← invertir_cadena("sol", 0) retorna "lo" + "s" = "los"
```

```python
def invertir_cadena(cadena, indice=0):
    if indice == len(cadena) - 1:
        return cadena[indice]
    return invertir_cadena(cadena, indice + 1) + cadena[indice]

print(invertir_cadena("Hola"))  # "aloH"
```

---

## 🌟 Conclusiones
- `return` es lo que permite que la información fluya hacia afuera de la pila.
- Los valores **bajan** como parámetros hasta el caso base, y luego **suben** combinándose en cada `return`.
- Trazar esto a mano (paso 5 del marco) es la mejor herramienta que tienes para depurar "en la cabeza" antes de correr código — justo lo que te va a ahorrar tiempo en el parcial.

## 🌿 Recursión múltiple: cuando el problema se ramifica

Hasta ahora cada llamada recursiva generaba **una sola** llamada recursiva más (recursión **lineal**). Muchos problemas interesantes requieren que una llamada genere **dos o más** llamadas recursivas — la pila deja de ser una fila y se convierte en un **árbol**.

El marco de 6 pasos es el mismo; lo único que cambia es que en el paso 5 (traza a mano) ahora dibujas un árbol en vez de una fila.

### 🔍 Ejemplo: Fibonacci

**Problema:** `fib(n)` = el n-ésimo número de Fibonacci, con `fib(0) = 0`, `fib(1) = 1`, y para `n ≥ 2`: `fib(n) = fib(n-1) + fib(n-2)`.

**Paso 1 — salto de fe:** asumo que `fib(n-1)` y `fib(n-2)` YA calculan correctamente esos dos números, sin importarme cómo.

**Paso 2 — en términos de sí mismo:** *"si ya supiera los dos Fibonacci anteriores, el actual es la suma de ambos."*

**Paso 3 — casos base:** `fib(0) = 0` y `fib(1) = 1` (dos casos base, porque el caso recursivo necesita DOS términos anteriores).

**Paso 4 — caso recursivo:** `fib(n) = fib(n-1) + fib(n-2)`.

**Paso 5 — traza a mano para `fib(4)`:**

```
                    fib(4)
                   /      \
              fib(3)        fib(2)
             /     \        /    \
        fib(2)    fib(1) fib(1) fib(0)
       /    \        |      |     |
   fib(1) fib(0)     1      1     0
     |      |
     1      0
```

Resolviendo de abajo hacia arriba:
```
fib(2) = fib(1) + fib(0) = 1 + 0 = 1     (aparece DOS veces en el árbol)
fib(3) = fib(2) + fib(1) = 1 + 1 = 2
fib(4) = fib(3) + fib(2) = 2 + 1 = 3
```

👀 **Observación clave:** `fib(2)` se calculó dos veces de manera completamente independiente (una vez como parte de `fib(3)`, otra como hijo directo de `fib(4)`). Nadie "recuerda" el resultado entre ramas. Esto es normal en recursión múltiple simple, y es la razón por la que este `fib` es exponencial en tiempo (`O(2ⁿ)` aprox.) — algo que ya puedes reconocer con lo que sabes de análisis de complejidad. (La técnica para evitar este recálculo, *memoización*, la puedes explorar por tu cuenta, pero no es examinable en este curso).

**Paso 6 — código:**

````python
def fib(n: int) -> int:
    if n == 0:  # Caso base 1
        return 0
    if n == 1:  # Caso base 2
        return 1
    return fib(n - 1) + fib(n - 2)  # Caso recursivo: dos llamadas

print(fib(4))  # 3
print([fib(i) for i in range(8)])  # [0, 1, 1, 2, 3, 5, 8, 13]
````

💭 Nota cómo el árbol de `fib` es exactamente lo que necesitas dibujar en el paso 5 cada vez que una función se llama a sí misma **más de una vez**: no basta con una fila de frames, necesitas ramas. En el examen, si te bloqueas con un problema de recursión múltiple, dibujar aunque sea las primeras 2-3 ramas a mano casi siempre revela el patrón que necesitas para el caso recursivo.

## 🔙 Backtracking: explorar y retroceder

El **backtracking** es un caso particular de recursión múltiple donde en cada paso **decides entre varias opciones**, exploras una completamente (llamando recursivamente), y al terminar esa exploración **retrocedes** para probar la siguiente opción. Lo vas a necesitar explícitamente esta semana para varios ejercicios (buscar palabras en una matriz, contar islas, encontrar caminos en un laberinto).

### 🔍 Ejemplo: generar todos los subconjuntos de una lista

**Problema:** dada una lista, generar todos sus subconjuntos (el "conjunto potencia").

**Paso 1 — salto de fe:** asumo que `subconjuntos(resto)` ya me da TODOS los subconjuntos de "todo menos el primer elemento".

**Paso 2 — en términos de sí mismo:** *"si ya tengo todos los subconjuntos del resto de la lista, los subconjuntos de la lista completa son: cada uno de esos subconjuntos SIN el primer elemento, más cada uno de esos subconjuntos CON el primer elemento agregado."* — esta es la decisión de backtracking: **incluir o no incluir** el elemento actual.

**Paso 3 — caso base:** lista vacía → el único subconjunto es la lista vacía: `[[]]`.

**Paso 4 — caso recursivo:** combino "sin el primero" y "con el primero" (dos ramas de decisión).

**Paso 5 — traza a mano para `subconjuntos([1, 2])`:**

```
subconjuntos([1, 2])
├─ NO incluir 1 → subconjuntos([2])
│                 ├─ NO incluir 2 → subconjuntos([]) = [[]]
│                 └─ SÍ incluir 2 → resultado: [[2]]
│                 (subconjuntos([2]) retorna [[], [2]])
└─ SÍ incluir 1 → tomar cada subconjunto de subconjuntos([2]) y agregarle el 1
                   → [[1], [1, 2]]

subconjuntos([1, 2]) retorna [[], [2]] + [[1], [1, 2]] = [[], [2], [1], [1, 2]]
```

Cada rama del árbol es una **decisión** (incluir / no incluir); "retroceder" significa que, después de terminar de explorar completamente la rama "incluir 1" (con toda su subrecursión), la ejecución regresa al nivel anterior para explorar "no incluir 1" (o viceversa, según el orden que programes).

**Paso 6 — código:**

````python
def subconjuntos(lista: list) -> list:
    if len(lista) == 0:  # Caso base
        return [[]]

    primero = lista[0]
    resto = lista[1:]
    sin_incluir = subconjuntos(resto)
    con_incluir = [[primero] + sub for sub in sin_incluir]
    return sin_incluir + con_incluir

print(subconjuntos([1, 2]))     # [[], [2], [1], [1, 2]]
print(subconjuntos([1, 2, 3]))  # 8 subconjuntos
````

Ya viste los tres "tipos" de forma en que una recursión puede estar estructurada:
- **Lineal** (`imprimir_hasta_n`, `factorial`, `suma_n`, `invertir_cadena`): cada llamada genera **una** llamada recursiva más.
- **Múltiple / ramificada** (`fib`): cada llamada genera **varias** llamadas recursivas independientes.
- **Backtracking** (`subconjuntos`): recursión múltiple donde las ramas representan **decisiones** explícitas (incluir/no incluir, ir por aquí/ir por allá).

En los tres casos el marco de 6 pasos es idéntico. Lo único que cambia es qué tan complicado se ve el dibujo del paso 5 — y por eso vale la pena practicarlo siempre, sin saltárselo.

# 📝 **Recursión de Cola vs Recursión No de Cola**

> 🔧 Esta sección es un **refinamiento** de cómo construyes el paso 4 (caso recursivo) del marco de 6 pasos — no un tema nuevo ni un reemplazo del marco. Ya viste recursión lineal, múltiple y backtracking; ahora afinamos *cómo* se estructura la llamada recursiva dentro de la función.

Existen dos maneras de estructurar dónde ocurre la llamada recursiva dentro de una función:

1️⃣ **Recursión de Cola (Tail Recursion)**
2️⃣ **Recursión No de Cola (Non-Tail Recursion)**

---

## 🌀 **1. Recursión de Cola (Tail Recursion)**
### ✅ **Definición**  
Una función tiene **recursión de cola** si la **llamada recursiva es la última operación** que realiza la función antes de retornar el resultado. No hay ninguna operación pendiente después de la recursión.  

### 🚀 **Características Clave**
- La llamada recursiva **es la última instrucción** en la función.
- No requiere almacenar contextos intermedios en la pila de llamadas.
- Puede optimizarse mediante **Tail Call Optimization (TCO)** en algunos lenguajes, reduciendo el consumo de memoria.
- Puede convertirse fácilmente en una solución **iterativa**.

### 🔎 **Ejemplo en Python**
```python
def factorial_tail(n, acc=1):
    if n == 0:
        return acc  # Caso base: devuelve el acumulador
    return factorial_tail(n - 1, acc * n)  # Llamada recursiva como última operación

print(factorial_tail(5))  # 120
```
🔹 Aquí, `acc` (acumulador) almacena el resultado parcial y se pasa a la siguiente llamada, evitando acumulaciones en la pila.

---

## 🌀 **2. Recursión No de Cola (Non-Tail Recursion)**
### ✅ **Definición**  
Una función tiene **recursión no de cola** cuando **la llamada recursiva no es la última operación** de la función. Después de la recursión, hay más cálculos o combinaciones de resultados antes de retornar.

### 🚀 **Características Clave**
- La llamada recursiva **no es la última instrucción**.
- La pila de llamadas crece porque cada llamada debe recordar operaciones pendientes.
- **Menos eficiente en términos de memoria**, especialmente en problemas con profundidad alta de recursión.
- No se puede optimizar con **Tail Call Optimization (TCO)**.

### 🔎 **Ejemplo en Python**
```python
def factorial_non_tail(n):
    if n == 0:
        return 1  # Caso base
    return n * factorial_non_tail(n - 1)  # Multiplicación ocurre después de la recursión

print(factorial_non_tail(5))  # 120
```
🔹 Aquí, la llamada `factorial_non_tail(n - 1)` se resuelve antes de realizar la multiplicación con `n`, lo que impide optimización de la pila. Es exactamente lo que trazaste a mano en el ejemplo de `factorial` al inicio del notebook.

---

## 📊 **Comparación Recursión de Cola vs No de Cola**
| Característica           | Recursión de Cola  | Recursión No de Cola |
|--------------------------|-------------------|----------------------|
| **Posición de la llamada recursiva** | Última operación  | Antes de retornar |
| **Uso de pila**         | No crece (optimizable) | Crece (llamadas pendientes) |
| **Eficiencia**          | Más eficiente      | Menos eficiente |
| **Conversión a iterativo** | Fácil            | Difícil |
| **Optimización (TCO)** | Posible en algunos lenguajes | No aplicable |

---

## 🎯 **Conclusión**
- ✅ **Usa recursión de cola** cuando sea posible para mejorar la eficiencia y evitar desbordamientos de pila.  
- ❌ **Evita la recursión no de cola** en problemas con profundidad grande, ya que puede consumir demasiada memoria.  
- 🔁 **Si el lenguaje no optimiza la recursión de cola**, considera **usar iteración** en su lugar.
- ⚠️ Python **no** aplica Tail Call Optimization de forma nativa — el beneficio real en Python es evitar cálculos pendientes y hacer más clara la relación con una versión iterativa, no ahorrar memoria de pila.

## **Ejemplo 1: Suma de los primeros N números naturales**  
📌 **Problema:** Dado un número entero `N`, calcular la suma de los primeros `N` números naturales.  

### 🔎 **Solución con Recursión No de Cola**
```python
def suma_non_tail(n):
    if n == 0:
        return 0  # Caso base
    return n + suma_non_tail(n - 1)  # Llamada recursiva NO es la última operación

print(suma_non_tail(5))  # 15
```
🔹 **Análisis:**  
- La llamada `suma_non_tail(n - 1)` **se evalúa primero**, y luego `n` se suma al resultado.
- Esto hace que **las llamadas queden pendientes en la pila**, esperando que se resuelvan antes de retornar.
- La profundidad de la pila será `O(N)`, lo que puede generar desbordamientos en valores grandes de `N`.

---

### 🔎 **Solución con Recursión de Cola**
```python
def suma_tail(n, acc=0):
    if n == 0:
        return acc  # Caso base: devolvemos el acumulador
    return suma_tail(n - 1, acc + n)  # Llamada recursiva en la última posición

print(suma_tail(5))  # 15
```
🔹 **Análisis:**  
- Usamos un **acumulador (`acc`)** para mantener la suma parcial.
- La llamada recursiva ocurre **al final**, sin cálculos pendientes.
- **Ventaja:** No crece la pila, lo que permite resolver `N` grande sin riesgo de desbordamiento.
- Si el lenguaje admite **Tail Call Optimization (TCO)**, la función se ejecutará en **O(1) espacio**, eliminando la sobrecarga de la pila.

## **Ejemplo 2: Invertir una cadena de texto**  
📌 **Problema:** Dada una cadena `s`, invertirla usando recursión.

### 🔎 **Solución con Recursión No de Cola**
```python
def invertir_non_tail(s):
    if len(s) == 0:
        return s  # Caso base: cadena vacía
    return invertir_non_tail(s[1:]) + s[0]  # Llamada recursiva NO es la última operación

print(invertir_non_tail("hola"))  # "aloh"
```
🔹 **Análisis:**  
- La llamada `invertir_non_tail(s[1:])` se resuelve antes de concatenar `s[0]`.
- Cada llamada almacena la **parte inicial de la cadena en la pila**, esperando la ejecución de la recursión.
- **Desventaja:** Cada llamada reserva memoria adicional para almacenar la concatenación (`O(N²)`) en lugar de `O(N)`.

---

### 🔎 **Solución con Recursión de Cola**
```python
def invertir_tail(s, acc=""):
    if len(s) == 0:
        return acc  # Caso base: devolvemos el acumulador
    return invertir_tail(s[1:], s[0] + acc)  # Llamada recursiva en la última posición

print(invertir_tail("hola"))  # "aloh"
```
🔹 **Análisis:**  
- Usamos un **acumulador (`acc`)** para almacenar la cadena invertida sin esperar retornos.
- En cada paso, el primer carácter (`s[0]`) se **agrega al inicio** del acumulador.
- **Ventaja:** No hay necesidad de operaciones pendientes después de la llamada recursiva, mejorando la eficiencia en memoria.

---

## **🎯 Conclusión**
### 🆚 **Comparación de Pensamiento para Resolver con Tail vs Non-Tail**
| Paso | Recursión No de Cola | Recursión de Cola |
|------|---------------------|------------------|
| **Caso base** | Se define un valor de retorno simple (ej. `0` para suma o `""` para invertir cadena). | Igual. |
| **Llamada recursiva** | Se resuelve primero, luego se usa el resultado en un cálculo. | Se pasa un **acumulador** que almacena progresivamente el resultado final. |
| **Uso de pila** | Crece con cada llamada, ocupando más memoria. | Puede optimizarse en algunos lenguajes para no usar pila (TCO). |
| **Conversión a iterativo** | Más difícil, ya que hay operaciones pendientes. | Más fácil de convertir en un bucle `while`. |

🔹 **Conclusión:**  
- Usa **recursión no de cola** cuando la naturaleza del problema no permita un **acumulador** o cuando se necesiten cálculos después de la llamada recursiva.  
- Usa **recursión de cola** siempre que puedas para mejorar eficiencia y evitar problemas de memoria.
