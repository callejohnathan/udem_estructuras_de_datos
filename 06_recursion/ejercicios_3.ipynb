### 1️⃣ **Contar el número de formas de alcanzar una suma con enteros positivos**
Dado un número `n`, encuentra cuántas formas distintas existen de sumarlo usando enteros positivos menores o iguales a `n`, sin importar el orden.

**Ejemplo de entrada y salida:**
```python
contar_formas(4)  
# Output: 5  
# Explicación:
# 1+1+1+1, 1+1+2, 1+3, 2+2, 4
```




**🧭 Planeación obligatoria (backtracking)**
1. **Caso base:** ¿cuándo `n` llega a 0 (encontraste una combinación válida)? ¿Cuándo el número máximo permitido es mayor que lo que queda por sumar (esa rama no sirve)?
2. **Caso recursivo (en palabras):** en cada paso, ¿qué números puedes "usar" para sumar (menores o iguales al máximo permitido y a lo que falta)? ¿Cómo evitas contar la misma combinación en distinto orden? (pista: no vuelvas a usar números mayores al último que usaste)
3. **Traza a mano:** para `contar_formas(3)`, dibuja el árbol de decisiones (qué número se elige en cada rama) y cuenta las hojas válidas.

### 2️⃣ **Resolver el problema de la mochila (Knapsack Problem)**
Dado un conjunto de objetos con pesos y valores, y una capacidad máxima de carga, encuentra el valor máximo que se puede obtener sin exceder la capacidad.

**Ejemplo de entrada y salida:**
```python
pesos = [2, 3, 4, 5]
valores = [3, 4, 5, 6]
capacidad = 5

mochila(pesos, valores, capacidad)  
# Output: 7  (Seleccionando los elementos de peso 2 y 3)
```



**🧭 Planeación obligatoria (backtracking)**
1. **Caso base:** ¿cuándo ya no quedan objetos por considerar, o la capacidad restante es 0?
2. **Caso recursivo (en palabras):** para el objeto actual, ¿qué dos decisiones puedes tomar (incluirlo o no incluirlo)? ¿Cómo decides cuál de las dos ramas da mejor valor total?
3. **Traza a mano:** para `pesos=[2, 3]`, `valores=[3, 4]`, `capacidad=3`, dibuja las dos ramas (incluir / no incluir el primer objeto) y qué valor da cada una.

````python
#Tail - Recursion
from typing import List
x = [2,3,4]
def sumatoria_lista(lista: List[int], i: int, sumatoria: int) -> int:
  if(i == len(lista)):
    return sumatoria

  return sumatoria_lista(lista, i+1, sumatoria + lista[i])

sumatoria_lista(x, 0, 0)
````

````python
#Non-Tail Recursion
from typing import List
x = [2,3,4]
def sumatoria_lista(lista: List[int], i: int) -> int:
  if(i == len(lista)):
    return 0

  return lista[i] + sumatoria_lista(lista, i + 1)

sumatoria_lista(x, 0)
````

````python
def recorrer_matriz(matriz, i=0, j=0) -> None:
  if(i == len(matriz)):
    return None

  if(j == len(matriz[i])): #j ya tiene el tamaño de la fila en la que estamos?
    recorrer_matriz(matriz, i+1, 0)
  else:
    print(matriz[i][j])
    recorrer_matriz(matriz, i, j+1)

matriz = [[1,2,3],[5,4,5],[6,7,8]]
recorrer_matriz(matriz)
````

````python
def sumatoria_matriz(matriz, i=0, j=0, sumatoria = 0) -> int:
  if(i == len(matriz)):
    return sumatoria

  if(j == len(matriz[i])): #j ya tiene el tamaño de la fila en la que estamos?
    return sumatoria_matriz(matriz, i+1, 0, sumatoria)
  else: #estoy en posición válida
    sumatoria += matriz[i][j]
    return sumatoria_matriz(matriz, i, j+1, sumatoria)

matriz = [[1,2,3],[5,4,5],[6,7,8]]
sumatoria_matriz(matriz)
````

````python
#sin cola
def sumatoria_matriz(matriz, i=0, j=0) -> int:
  if(i == len(matriz)):
    return 0

  if(j == len(matriz[i])): #j ya tiene el tamaño de la fila en la que estamos?
    return 0 + sumatoria_matriz(matriz, i+1, 0)
  else: #estoy en posición válida
    return matriz[i][j] + sumatoria_matriz(matriz, i, j+1)

matriz = [[1,2,3],[5,4,5],[6,7,8]]
sumatoria_matriz(matriz)
````
