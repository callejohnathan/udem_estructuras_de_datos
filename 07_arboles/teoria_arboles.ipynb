# Conceptualización

## 🌳 Conceptualización de Árboles en Estructuras de Datos

## 1. Definición de árbol

Un árbol es una estructura de datos jerárquica que consiste en nodos conectados mediante aristas, donde cada nodo (excepto la raíz) tiene exactamente un padre, y puede tener cero o más hijos. Es una estructura no lineal usada para representar relaciones jerárquicas como directorios, familias o categorías.

## 2. Terminología clave

- **Nodo**: unidad fundamental que contiene un valor y puede tener enlaces a otros nodos.
- **Raíz**: nodo principal del árbol, sin padre.
- **Hijo**: nodo que desciende de otro nodo (su padre).
- **Padre**: nodo que tiene al menos un hijo.
- **Hoja**: nodo sin hijos.
- **Hermano**: nodos que comparten el mismo padre.
- **Grado de un nodo**: número de hijos del nodo.
- **Grado del árbol**: máximo grado entre todos los nodos del árbol.
- **Altura de un nodo**: longitud del camino más largo desde el nodo hasta una hoja (una hoja tiene altura 0).
- **Altura del árbol**: altura de la raíz.
- **Profundidad de un nodo**: longitud del camino desde la raíz hasta ese nodo (la raíz tiene profundidad 0).
- **Nivel**: la profundidad de un nodo más uno (iniciando en 1 para la raíz).
- **Subárbol**: cualquier nodo con sus descendientes forma un subárbol.
- **Camino**: secuencia de nodos conectados por aristas.
- **Longitud de camino**: número de aristas en un camino.
- **Anchura del árbol**: máximo número de nodos en cualquier nivel.

## 3. Propiedades de los árboles

- Un árbol con \(n\) nodos siempre tiene \(n - 1\) aristas.
- En un árbol binario, el número máximo de nodos en el nivel \(l\) es \(2^{l-1}\).
- El número máximo de nodos en un árbol binario de altura \(h\) es \(2^h - 1\).
- La altura mínima de un árbol con \(n\) nodos es \(\lceil \log_2(n+1) \rceil - 1\).
- Un árbol es conexo y acíclico por definición.

## 4. Representación visual y formal de un árbol

### Forma jerárquica (diagrama)

```
        A
       / \
      B   C
     / \   \
    D   E   F
```

### Forma tabular (parentesco)

| Nodo | Padre |
| ---- | ----- |
| A    | -     |
| B    | A     |
| C    | A     |
| D    | B     |
| E    | B     |
| F    | C     |

Esta representación ayuda a ilustrar los conceptos de raíz, padre, hijo, hojas y subárboles de manera clara.

## 🔁 La idea central de este notebook: pensar en árboles es pensar recursivamente, un nodo a la vez

Ya viste en el notebook de **recursión** (carpeta `06_recursion`) que un problema recursivo se resuelve confiando en que una versión más pequeña del mismo problema ya está resuelta, y combinando ese resultado para resolver el caso actual. Un árbol es, literalmente, esa idea hecha estructura de datos: **cada nodo es la raíz de un árbol más pequeño** (su subárbol). Por eso casi todo algoritmo sobre árboles es una función recursiva donde:

- El **caso base** es el nodo vacío (`None`) o una hoja.
- El **caso recursivo** confía en que ya sabemos resolver el problema para los subárboles izquierdo/derecho (o para cada hijo, si es un árbol general), y combina esos resultados.

> ⚠️ Este es exactamente el error más común: memorizar la "forma" del código de un recorrido o de un BST y copiarla sin pensar, en vez de preguntarse qué necesita este problema en particular. La forma cambia según el problema; lo que **no** cambia es la disciplina de hacerte estas tres preguntas **antes de escribir una sola línea de código**:

### 🧠 Las tres preguntas que debes responder antes de programar

1. **¿Qué necesito *del subárbol izquierdo* y *del subárbol derecho* (o de cada hijo) para resolver el nodo actual?**
   Esto es un patrón **post-order**: primero resuelves los hijos, y con esas respuestas resuelves el padre. Ejemplos: altura de un árbol, si está balanceado, la suma de todos los valores, contar nodos.

2. **¿Qué necesito *pasar hacia abajo* desde el padre a los hijos?**
   Esto es un patrón **pre-order**: hay información que solo el padre (o un ancestro) conoce, y que el hijo necesita recibir *antes* de resolver su propia parte. Ejemplos: la profundidad acumulada de un nodo, los límites válidos (mínimo/máximo) que debe cumplir un valor para que el árbol siga siendo un BST válido.

3. **¿En qué orden necesito *visitar* los nodos para este problema en particular?**
   Preorden, inorden, postorden o por niveles (BFS) no son "cuatro formas intercambiables de recorrer" — cada una sirve para algo distinto. Por ejemplo, el inorden en un BST te da los valores en orden ascendente (más adelante vemos por qué); el recorrido por niveles sirve cuando el problema depende de la profundidad o "capa" del nodo.

### ✍️ El hábito que debes instalar

**Antes de escribir código**, dibuja el árbol (o una parte pequeña, representativa) en papel y traza la recursión a mano: para cada nodo relevante, escribe qué llamada recursiva se hace y qué devuelve. Si puedes completar ese trazado a mano para un ejemplo pequeño, el código es casi una traducción directa. Si no puedes, programar no te va a salvar — vas a terminar adivinando y frustrándote. En todo este notebook vamos a hacer ese trazado a mano **antes** de mostrar el código, para que practiques ese orden de trabajo.

## 🌳 Tipos de Árboles: Conceptos Básicos

En esta sección exploraremos algunos de los tipos más comunes de árboles utilizados en estructuras de datos. Cada tipo tiene propiedades y restricciones particulares que lo hacen adecuado para distintos problemas computacionales. Los dos tipos con más contenido propio (**Árbol Binario de Búsqueda** y **Árbol AVL**) tienen su propia sección más adelante con su modelo de pensamiento específico — aquí solo los ubicamos dentro del panorama general.

---

## 1. Árbol General 🌳

Un árbol general es una estructura jerárquica donde cada nodo puede tener **cero o más hijos** sin restricción en su cantidad. Es el tipo más flexible y menos estructurado.

### Ejemplo visual:

```
        A
      / | \
     B  C  D
       / \
      E   F
```

---

## 2. Árbol Binario 🌳

Un árbol binario es un tipo especial de árbol donde cada nodo puede tener **como máximo dos hijos**, conocidos como **hijo izquierdo** y **hijo derecho**. La posición (izquierda/derecha) importa y es la base de estructuras más específicas como el BST o el AVL.

### Ejemplo visual:

```
        A
       / \
      B   C
     /     \
    D       E
```

---

## 3. Árbol Binario Completo 🌳

Un árbol binario completo es un árbol binario donde **todos los niveles están completamente llenos excepto posiblemente el último nivel**, que debe estar lleno de izquierda a derecha.

```
        A
       / \
      B   C
     / \  /
    D  E F
```

---

## 4. Árbol Binario Lleno (Full Binary Tree) 🌳

Un árbol binario lleno es aquel en el que **cada nodo tiene 0 o 2 hijos**. No existen nodos con un solo hijo.

```
        A
       / \
      B   C
     / \ / \
    D  E F  G
```

---

## 5. Árbol Binario Balanceado ⚖️

Un árbol binario está **balanceado** si la diferencia de altura entre los subárboles izquierdo y derecho de cualquier nodo es, como máximo, 1. Evita el caso extremo donde el árbol degenera en una lista enlazada (altura \(O(n)\) en vez de \(O(\log n)\)).

```
        5
       / \
      3   8
     / \   \
    2   4   10
```

---

Estos tipos sirven como base para estructuras más avanzadas. El **Árbol Binario de Búsqueda (BST)** añade una regla de orden sobre un árbol binario, y el **Árbol AVL** es un BST que además se auto-balancea — ambos se desarrollan en detalle más abajo.

## 🧠 Representación y Almacenamiento de Árboles

Comprender cómo se representan los árboles en memoria es clave para poder razonar sobre su implementación y eficiencia. Los dos enfoques más comunes son:

1. **Nodos enlazados**: cada nodo contiene referencias a sus hijos.
2. **Arrays o listas**: los nodos se almacenan de forma contigua según una posición lógica derivada de su ubicación en el árbol.

---

## 💐 Representación con nodos enlazados

Ideal para estructuras jerárquicas **dinámicas** (el caso que más vamos a usar en este curso).

- Cada **nodo** contiene un valor y referencias a sus **hijos**.
- En un árbol binario, esas referencias se limitan a **dos**: izquierda y derecha.
- El árbol se construye **nodo por nodo**, agregando enlaces según la posición deseada.

**Ventajas:** fácil de expandir/modificar, natural para recorridos recursivos.
**Desventajas:** mayor uso de memoria (punteros), más difícil de representar en arrays.

```
Nodo: valor = 10
      izquierda → Nodo(5)
      derecha   → Nodo(15)
```

---

## 📋 Representación con arrays (vectores)

Se numera cada nodo según su posición en un recorrido **por niveles (BFS)** y se guarda en una posición fija del array. Para un nodo en la posición `i`:

- Hijo izquierdo → posición `2i + 1`
- Hijo derecho → posición `2i + 2`
- Padre → posición `floor((i - 1) / 2)`

```
       A
      / \
     B   C
    / \
   D   E

Array: [A, B, C, D, E]
  pos:  0  1  2  3  4
```

**Ventajas:** ocupa menos memoria si el árbol es completo o casi completo, acceso muy rápido por índice.
**Desventajas:** ineficiente para árboles dispersos, costoso insertar/eliminar dinámicamente.

> 💡 Este es exactamente el truco que usa un **montículo (heap)** — lo viste (o lo verás) en la carpeta de pilas y colas de prioridad: un árbol binario casi-completo representado como array, sin necesidad de punteros.

### ✨ Conclusión
- Si el árbol crece y se modifica con frecuencia → **nodos enlazados** (lo que usaremos casi siempre aquí).
- Si el árbol es estático y completo → **arrays**, por eficiencia.

# Árbol General

## 📚 Diseñando un árbol general

Antes de pensar en algoritmos sobre el árbol, hay que tener claro qué representa un nodo:

- Un **valor** (el contenido).
- Una **lista de hijos** (puede tener 0, 1 o muchos — sin límite, a diferencia del árbol binario).

> 🧠 En `base.py` esto está implementado como `GeneralNode` (con `.value` y `.children`) y `GeneralTree` (con `.root` y `.insert(padre, valor)`). Ábrelo y revísalo — es la referencia que vamos a usar de aquí en adelante en vez de reinventar la clase en cada sección.

Como cualquier estructura jerárquica, casi toda operación sobre un árbol general sigue el mismo esqueleto recursivo:

```python
def operacion(nodo):
    if nodo is None:
        return caso_base
    resultado = ... algo con nodo.value ...
    for hijo in nodo.children:
        resultado = combinar(resultado, operacion(hijo))
    return resultado
```

Lo que cambia entre un ejercicio y otro es **qué es `caso_base`** y **cómo se combina** — no la estructura del `for`. Vamos a practicar identificar eso antes de programar.

````python
org = GeneralTree()
org.insert(None, "CEO")
org.insert("CEO", "Finanzas")
org.insert("CEO", "Operaciones")
org.insert("CEO", "Marketing")
org.insert("Finanzas", "Analista 1")
org.insert("Finanzas", "Analista 2")
print(org)
````

## 🔎 Aplicando las 3 preguntas: "¿cuántos nodos tiene el árbol?"

Antes de programar, respondamos la primera pregunta del modelo mental: **¿qué necesito de cada hijo para resolver el nodo actual?**

Para contar nodos: si ya sé cuántos nodos tiene cada subárbol hijo, el total del nodo actual es `1 (por mí mismo) + suma de lo que devolvió cada hijo`. Eso es un patrón **post-order**: resuelvo todos los hijos primero, combino al final.

### ✍️ Trazado a mano sobre el árbol del organigrama

```
CEO
├── Finanzas
│   ├── Analista 1
│   └── Analista 2
├── Operaciones
└── Marketing
```

Como es recursivo, la primera llamada que *termina* es la de una hoja. El orden en que cada llamada devuelve su resultado (de adentro hacia afuera) es:

| Llamada | ¿Qué devuelve? | ¿Por qué? |
|---|---|---|
| `contar_nodos(Analista 1)` | `1` | es hoja, no tiene hijos |
| `contar_nodos(Analista 2)` | `1` | es hoja |
| `contar_nodos(Finanzas)` | `1 + 1 + 1 = 3` | ella misma + lo que devolvieron sus 2 hijos |
| `contar_nodos(Operaciones)` | `1` | es hoja |
| `contar_nodos(Marketing)` | `1` | es hoja |
| `contar_nodos(CEO)` | `1 + 3 + 1 + 1 = 6` | ella misma + lo que devolvió cada uno de sus 3 hijos |

Solo después de poder llenar una tabla como esta a mano, escribimos el código — que termina siendo casi una traducción directa de la tabla:

````python
def contar_nodos(nodo):
    if nodo is None:
        return 0
    total = 1  # me cuento a mí mismo
    for hijo in nodo.children:
        total += contar_nodos(hijo)
    return total

contar_nodos(org.root)
````

# Árbol Binario

## 🌲 Del árbol general al árbol binario

Un árbol binario es un caso particular de árbol general: cada nodo tiene **como máximo dos hijos**, pero ahora **la posición importa** — no es lo mismo ser el hijo izquierdo que el derecho. En `base.py`, `BinaryNode` tiene `.value`, `.left` y `.right`, y `BinaryTree` reutiliza el árbol de ejemplo que construimos arriba (`ejemplo`).

El esqueleto recursivo también cambia ligeramente respecto al árbol general (ya no hay un `for` sobre una lista de hijos, sino dos llamadas explícitas: una a `nodo.left` y otra a `nodo.right`):

```python
def operacion(nodo):
    if nodo is None:
        return caso_base
    izquierda = operacion(nodo.left)
    derecha = operacion(nodo.right)
    return combinar(nodo.value, izquierda, derecha)
```

Vamos a usar este esqueleto para practicar, uno por uno, los dos patrones del modelo mental: **post-order** (qué necesito de mis hijos) y **pre-order** (qué le paso a mis hijos).

## 📏 Patrón post-order: "¿cuál es la altura del árbol?"

**Pregunta 1 del modelo mental:** ¿qué necesito de mi subárbol izquierdo y de mi subárbol derecho para saber mi propia altura?

Respuesta: la altura de cada uno. Mi altura es `1 + max(altura(izquierda), altura(derecha))`. El caso base es el árbol vacío: por convención le damos altura `-1`, para que una hoja (que tiene dos hijos vacíos) dé `1 + max(-1, -1) = 0`, que es justo la definición de altura de una hoja que vimos en la sección de terminología.

### ✍️ Trazado a mano sobre `ejemplo`

```
        8
       / \
      3   10
     / \    \
    1   6    14
       / \   /
      4   7 13
```

Igual que antes, seguimos el orden en que cada llamada recursiva *termina* (post-order: izquierda, derecha, yo):

| Nodo | ¿Qué devuelve `altura(nodo)`? |
|---|---|
| `None` (cualquier hijo vacío) | `-1` |
| `1`, `4`, `7`, `13` (hojas) | `0` |
| `6` | `1 + max(altura(4), altura(7)) = 1 + max(0, 0) = 1` |
| `3` | `1 + max(altura(1), altura(6)) = 1 + max(0, 1) = 2` |
| `14` | `1 + max(altura(13), altura(None)) = 1 + max(0, -1) = 1` |
| `10` | `1 + max(altura(None), altura(14)) = 1 + max(-1, 1) = 2` |
| `8` (raíz) | `1 + max(altura(3), altura(10)) = 1 + max(2, 2) = 3` |

La altura del árbol completo es `3`. Nota que **no importa el problema** (altura, suma de valores, contar hojas, verificar si es balanceado...) — todos siguen exactamente esta misma tabla, solo cambia qué se calcula en cada fila. Ahora el código:

````python
def altura(nodo):
    if nodo is None:
        return -1
    return 1 + max(altura(nodo.left), altura(nodo.right))

altura(ejemplo.root)
````

## 🎯 Patrón pre-order: "¿qué profundidad tiene cada nodo?"

**Pregunta 2 del modelo mental:** ¿qué necesito pasar hacia abajo, del padre al hijo?

Para la altura, la información viajaba de **abajo hacia arriba** (de los hijos al padre). Para la profundidad es al revés: viaja de **arriba hacia abajo**. La raíz sabe que su profundidad es `0`; cada hijo tiene la profundidad de su padre `+ 1`. Ningún nodo puede calcular su propia profundidad sin que se la pase su padre — por eso este parámetro se pasa como argumento en la llamada recursiva, en vez de calcularse a partir de lo que devuelven los hijos.

### ✍️ Trazado a mano sobre `ejemplo`

| Nodo | Profundidad | ¿Cómo se obtuvo? |
|---|---|---|
| `8` | `0` | es la raíz, caso base |
| `3`, `10` | `1` | `profundidad(8) + 1` |
| `1`, `6`, `14` | `2` | `profundidad(su padre) + 1` |
| `4`, `7`, `13` | `3` | `profundidad(su padre) + 1` |

### 🧠 Comparación clave

- **Post-order (altura):** la llamada recursiva no recibe nada extra, pero **devuelve** información que el padre usa.
- **Pre-order (profundidad):** la llamada recursiva **recibe** un parámetro adicional que el padre calculó, y lo hijos reciben antes de resolver nada.

Este mismo patrón de "pasar información hacia abajo" es el que vas a necesitar para el ejercicio de validar si un árbol es un BST correcto: cada nodo recibe de su padre un rango válido `(mínimo, máximo)` en el que su valor debe caer.

````python
def imprimir_profundidades(nodo, profundidad=0):
    if nodo is None:
        return
    print("  " * profundidad + f"{nodo.value} (profundidad {profundidad})")
    imprimir_profundidades(nodo.left, profundidad + 1)
    imprimir_profundidades(nodo.right, profundidad + 1)

imprimir_profundidades(ejemplo.root)
````

# Árbol Binario de Búsqueda

## 🔎 El invariante del BST

Un Árbol Binario de Búsqueda (BST) es un árbol binario con una regla extra en **cada nodo**:

> Todos los valores del subárbol izquierdo son **menores**, y todos los del subárbol derecho son **mayores**.

Esa regla se cumple de forma recursiva: no solo el hijo directo, sino *todo* el subárbol izquierdo debe ser menor. Por eso `insertar` y `buscar` en un BST siguen el mismo patrón descendente: en cada nodo, comparo el valor buscado contra `nodo.value` y decido si sigo por la izquierda o por la derecha — nunca reviso ambos lados.

### 🧠 El modelo mental de BST: es búsqueda binaria sobre un árbol

Ya viste en `02_analisis_asintotico` que la búsqueda binaria sobre un arreglo ordenado logra \(O(\log n)\) porque **descarta la mitad de las opciones restantes en cada paso**. Un BST hace exactamente lo mismo, pero en vez de calcular un punto medio sobre un arreglo, usa la estructura del árbol: al comparar el valor buscado contra la raíz, todo un subárbol completo (la mitad "lógica" de lo que queda por revisar) queda descartado de una sola vez.

> ⚠️ Esa complejidad \(O(\log n)\) **depende de que el árbol esté razonablemente balanceado**. Si insertas valores ya ordenados (`1,2,3,4,5...`) en un BST normal, degenera en una lista enlazada y buscar pasa a costar \(O(n)\) — exactamente el problema que un árbol **AVL** (más abajo) resuelve balanceándose solo.

### ✍️ Trazado a mano: insertar 65 en el árbol `[50, 30, 70, 20, 40, 60, 80]`

```
           50
         /    \
       30      70
      /  \    /  \
    20   40  60   80
```

| Paso | Nodo actual | Comparación | Decisión |
|---|---|---|---|
| 1 | `50` | `65 > 50` | ir a la derecha |
| 2 | `70` | `65 < 70` | ir a la izquierda |
| 3 | `60` | `65 > 60` | ir a la derecha → hueco vacío |
| 4 | — | — | insertar `65` como hijo derecho de `60` |

En 3 comparaciones descartamos, en cada paso, todo un subárbol — igual que la búsqueda binaria descarta la mitad del arreglo. Ahora sí, el código (usando `BinarySearchTree` de `base.py`, que ya trae `insert`):

````python
bst = BinarySearchTree()
for valor in [50, 30, 70, 20, 40, 60, 80]:
    bst.insert(valor)
bst.print()
````

## 🧭 Pregunta 3 del modelo mental: ¿en qué orden visito los nodos?

Ya sabemos *qué* calcular en cada nodo (preguntas 1 y 2). Ahora falta decidir *el orden* en que visitamos los nodos — y ese orden depende completamente de qué necesita el problema:

| Recorrido | Orden de visita | ¿Para qué sirve? |
|---|---|---|
| **Preorden** | raíz → izquierda → derecha | Cuando necesitas procesar al padre *antes* que a sus hijos (p. ej. copiar/serializar un árbol para reconstruirlo después). |
| **Inorden** | izquierda → raíz → derecha | En un BST, entrega los valores en **orden ascendente** (ver por qué abajo). |
| **Postorder** | izquierda → derecha → raíz | Cuando necesitas resolver a los hijos *antes* que al padre (p. ej. eliminar un árbol liberando primero los hijos, o cualquiera de los cálculos post-order que ya practicamos arriba). |
| **Por niveles (BFS)** | nivel 0, luego nivel 1, luego nivel 2... | Cuando el problema depende de la profundidad/nivel del nodo (p. ej. "imprime el árbol nivel por nivel", o encontrar el nodo más cercano a la raíz que cumple algo). |

### 🤔 ¿Por qué el inorden da orden ascendente en un BST?

Por el invariante mismo del BST: en cualquier nodo, *todo* el subárbol izquierdo es menor y *todo* el subárbol derecho es mayor. Si primero visito completo el subárbol izquierdo (todos los menores), luego al nodo, y luego completo el subárbol derecho (todos los mayores), el resultado tiene que salir ordenado — es el invariante aplicado recursivamente a cada subárbol.

### ✍️ Trazado a mano: inorden sobre el BST de arriba

```
           50
         /    \
       30      70
      /  \    /  \
    20   40  60   80
```

Orden de visita: `20, 30, 40, 50, 60, 70, 80` — ascendente, como predice el argumento de arriba.

````python
def preorden(nodo):
    if nodo is None:
        return
    print(nodo.value)
    preorden(nodo.left)
    preorden(nodo.right)

preorden(bst.root)
````

````python
def inorden(nodo):
    if nodo is None:
        return
    inorden(nodo.left)
    print(nodo.value)
    inorden(nodo.right)

inorden(bst.root)
````

````python
def postorden(nodo):
    if nodo is None:
        return
    postorden(nodo.left)
    postorden(nodo.right)
    print(nodo.value)

postorden(bst.root)
````

````python
from collections import deque

def por_niveles(raiz):
    if raiz is None:
        return
    cola = deque([raiz])
    while cola:
        nodo = cola.popleft()
        print(nodo.value)
        if nodo.left:
            cola.append(nodo.left)
        if nodo.right:
            cola.append(nodo.right)

por_niveles(bst.root)
````

# Árbol AVL

## ⚖️ Modelo de Pensamiento: Construcción de un Árbol AVL

El **Árbol AVL** es un tipo de Árbol Binario de Búsqueda (BST) **auto-balanceado**. Esto significa que, tras cada inserción o eliminación, el árbol **se ajusta automáticamente** para mantener una diferencia de alturas aceptable entre los subárboles. Su nombre proviene de sus creadores, Adelson-Velsky y Landis.

> 🔗 **Conexión con lo que ya practicamos:** insertar en un AVL empieza *exactamente* igual que insertar en un BST — el mismo descenso comparando contra cada nodo (pregunta 1 del modelo mental, patrón pre-order: decides hacia dónde bajar). La diferencia aparece al **subir** de la recursión: en cada nodo, por el camino de regreso, necesitas la altura de cada hijo para calcular tu propio balance — eso es exactamente el patrón **post-order** que practicamos con `altura(nodo)` arriba. Un AVL no es un concepto aislado: es "BST + post-order de altura/balance en cada nivel de la recursión, aplicando una rotación si hace falta".

---

## 🔍 1. ¿Qué es el balance?

En un árbol AVL, para **cada nodo**, la diferencia de altura entre sus subárboles izquierdo y derecho debe ser como máximo **1**.

> 📏 Balance = altura(izquierdo) - altura(derecho) ∈ {-1, 0, 1}

Si este criterio no se cumple, el árbol realiza **rotaciones** para recuperar el balance.

---

## 🧱 2. ¿Qué estructura necesitas?

Al igual que en un BST, necesitas dos clases:
- `NodoAVL`: como un nodo de BST, pero también debe guardar la **altura** del nodo y permitir actualizaciones.
- `ArbolAVL`: clase envoltorio que gestiona la raíz y maneja las inserciones balanceadas.

Cada nodo debe tener:
- `valor`
- `izquierdo` / `derecho`
- `altura`
- Métodos para insertar, calcular balance, y aplicar rotaciones

---

## ➕ 3. Inserción paso a paso

1. Inserta como en un BST (siguiendo el orden).
2. Luego, **actualiza la altura** del nodo actual (post-order, de abajo hacia arriba):
   ```
   altura = 1 + max(altura(hijo_izquierdo), altura(hijo_derecho))
   ```
3. **Calcula el balance** del nodo:
   ```
   balance = altura(izquierdo) - altura(derecho)
   ```
4. Si el balance está fuera del rango válido (−1, 0, 1), se debe **rebalancear**.

---

## 🔁 4. Proceso de rebalanceo y rotaciones

Cuando un nodo queda desbalanceado (balance = ±2), debemos detectar **el patrón de inserción** que causó el problema y aplicar la rotación adecuada:

### 💥 Casos de desbalance

#### 1. Izquierda-Izquierda (LL)
- El nodo fue insertado en el **subárbol izquierdo del hijo izquierdo**.
- ✅ Solución: **Rotación simple a la derecha**.

#### 2. Derecha-Derecha (RR)
- El nodo fue insertado en el **subárbol derecho del hijo derecho**.
- ✅ Solución: **Rotación simple a la izquierda**.

#### 3. Izquierda-Derecha (LR)
- El nodo fue insertado en el **subárbol derecho del hijo izquierdo**.
- ✅ Solución: **Rotación doble izquierda-derecha** (primero rotación izquierda, luego derecha).

#### 4. Derecha-Izquierda (RL)
- El nodo fue insertado en el **subárbol izquierdo del hijo derecho**.
- ✅ Solución: **Rotación doble derecha-izquierda** (primero rotación derecha, luego izquierda).

> 🎯 El rebalanceo ocurre en el **ancestro más cercano** donde se rompe el balance. Solo una rotación (o doble) es necesaria por inserción.

### 🔄 Ejemplo del flujo lógico
```python
if balance > 1:
    if valor < nodo.izquierdo.valor:
        return rotacion_derecha(nodo)        # LL
    else:
        nodo.izquierdo = rotacion_izquierda(nodo.izquierdo)
        return rotacion_derecha(nodo)        # LR

elif balance < -1:
    if valor > nodo.derecho.valor:
        return rotacion_izquierda(nodo)      # RR
    else:
        nodo.derecho = rotacion_derecha(nodo.derecho)
        return rotacion_izquierda(nodo)      # RL
```

> 🧠 Pensamiento clave: no importa dónde insertes el valor, al subir por la recursividad debes **verificar el balance** y aplicar la rotación si es necesario.

## ✅ Conclusión

Un árbol AVL requiere:
- Entender la **estructura BST** como base
- Mantener actualizadas las **alturas** de los nodos (patrón post-order)
- Calcular el **balance** de cada nodo luego de insertar o eliminar
- Aplicar **rotaciones correctas** según el patrón que causó el desbalance

Este modelo de pensamiento te prepara para construir árboles AVL robustos, eficientes y balanceados 🔧⚖️🌳

## ✅ Recapitulando: el mismo modelo mental para todo

Sin importar si el problema es sobre un árbol general, un árbol binario, un BST o un AVL, la forma de abordarlo antes de programar es siempre la misma:

1. ¿Qué necesito **de mis hijos** para resolver este nodo? (post-order)
2. ¿Qué necesito **pasarle a mis hijos**? (pre-order)
3. ¿En qué **orden** necesito visitar los nodos? (preorden / inorden / postorden / por niveles)
4. **Dibuja un ejemplo pequeño y traza la recursión a mano** antes de escribir código — completa una tabla como las que usamos en este notebook.
