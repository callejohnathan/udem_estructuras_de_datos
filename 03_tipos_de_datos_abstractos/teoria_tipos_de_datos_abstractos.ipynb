# Tipos de Datos Abstractos (ADTs) 🚀

## 1. Introducción 🧠

Un **Tipo de Dato Abstracto (ADT, por sus siglas en inglés)** es un modelo definido por su **comportamiento desde el punto de vista de quien lo usa**: qué operaciones soporta y qué resultados garantiza cada una. Un ADT **no dice nada sobre cómo se implementa** — eso es, a propósito, un detalle que se deja para después.

Esto es la aplicación, a nivel de estructura de datos completa, de la misma idea de **abstracción** que ya viste con funciones: ocultar el "cómo" y exponer solo el "qué". La sección siguiente retoma esa idea y la hace explícita, porque es el hábito de pensamiento central de esta unidad.

---

## 2. Importancia de los ADTs 🔍

Los ADTs son fundamentales en el diseño de software porque separan la implementación de los datos de su uso. Esto trae varias ventajas:

- **🧩 Modularidad**: se pueden desarrollar y mejorar componentes de forma independiente.
- **🔄 Reutilización**: un ADT puede usarse en contextos distintos sin conocer su implementación interna.
- **🛠️ Mantenimiento**: al definir claramente qué operaciones se permiten, es más fácil cambiar el código sin romper lo que ya funciona.
- **📌 Claridad**: enfocarse en el *qué* y no en el *cómo* permite expresar soluciones de forma más clara.

---

## 3. Ejemplos comunes de ADTs 📂

En este curso vas a trabajar, en este orden, con estos ADTs:

1. **📋 Lista**: almacena elementos en un orden específico y permite acceder a ellos por posición.
2. **📦 Pila (Stack)**: orden `LIFO` (Last In, First Out) — el último que entra es el primero que sale.
3. **📬 Cola (Queue) y Cola de prioridad**: la cola sigue `FIFO` (First In, First Out); la cola de prioridad extrae siempre el elemento de mayor prioridad, sin importar el orden de llegada.
4. **🌳 Árbol**: organiza elementos jerárquicamente (padres e hijos).
5. **🗺️ Grafo**: representa conexiones (aristas) entre elementos (nodos), sin la restricción jerárquica de un árbol.

También vas a usar con frecuencia dos ADTs que Python ya trae incorporados: **🟢 Conjunto (Set)** — colección de elementos únicos, sin orden — y **🗂️ Diccionario/Mapa** — pares clave-valor con búsqueda rápida por clave.

---

## 4. Ejemplos de ADTs en la vida real 🌍

- **📝 Lista de tareas**: cada tarea tiene una posición; se puede agregar, eliminar o modificar.
- **🍽️ Pila de platos**: el último plato en colocarse es el primero en tomarse (LIFO).
- **🏦 Fila en un banco**: los clientes son atendidos en el orden en que llegaron (FIFO).
- **🚑 Sala de urgencias**: es una cola de prioridad — no se atiende por orden de llegada, sino por gravedad.
- **📖 Agenda telefónica**: es un diccionario — los nombres son claves y los números son los valores.
- **🚉 Red de transporte**: es un grafo — las estaciones son nodos y las rutas son aristas.

Estos ejemplos muestran que el ADT correcto depende del **patrón de acceso** que necesitas — vas a formalizar esa idea en la siguiente sección.

## 🧠 Modelo mental: separa la interfaz de la implementación

Este es el hábito central de toda la unidad, y el que vas a necesitar constantemente en las próximas semanas cuando implementes listas enlazadas, pilas, colas y árboles: **antes de pensar en cómo vas a construir una estructura, define su interfaz — el conjunto de operaciones que necesitas, con su contrato (qué entra, qué sale, qué garantiza).** Solo después decides cómo implementarla.

* **Interfaz (el *qué*):** el conjunto de operaciones que el ADT ofrece hacia afuera, y qué promete cada una. Por ejemplo, una **Pila** promete: `apilar(x)` agrega `x` arriba; `desapilar()` quita y devuelve el elemento de arriba; `ver_tope()` muestra el de arriba sin quitarlo; `esta_vacia()` indica si no hay elementos.
* **Implementación (el *cómo*):** la estructura interna concreta que hace que esas operaciones funcionen — una lista de Python, un arreglo de tamaño fijo, una lista enlazada... Esto es exactamente lo que vas a construir con tus propias manos en las próximas semanas.

📌 **La idea clave:** dos implementaciones completamente distintas pueden cumplir la **misma interfaz**, y el código que las usa no tiene por qué notar la diferencia. Antes de correr la celda de código de abajo, léela y predice: ¿va a imprimir lo mismo `usar_pila(PilaConLista())` que `usar_pila(PilaConDiccionario())`, aunque por dentro guarden los datos de formas completamente distintas (una lista vs. un diccionario con contador)? Corre la celda y comprueba tu predicción.

⚠️ **Por qué importa de verdad:** si `usar_pila` hubiera asumido detalles internos (por ejemplo, hubiera hecho `pila._datos[-1]` en vez de llamar `pila.ver_tope()`), se habría **roto** al recibir `PilaConDiccionario` — un diccionario no soporta `[-1]` de la misma forma. Programar contra la interfaz, no contra la implementación, es lo que te permite **cambiar cómo construyes algo por dentro sin romper todo el código que ya lo usa**. Esto es exactamente lo que vas a hacer en el curso: la interfaz de "Lista" no cambia entre la semana 2 (lista enlazada simple) y la semana 4 (lista doblemente enlazada) — cambia la implementación, no el contrato.


```python
class PilaConLista:
    """Interfaz: apilar(x), desapilar(), ver_tope(), esta_vacia().
    Implementación: usa una lista de Python y opera sobre su extremo final."""
    def __init__(self):
        self._datos = []

    def apilar(self, x):
        self._datos.append(x)

    def desapilar(self):
        return self._datos.pop()

    def ver_tope(self):
        return self._datos[-1]

    def esta_vacia(self):
        return len(self._datos) == 0


class PilaConDiccionario:
    """Misma interfaz exactamente: apilar(x), desapilar(), ver_tope(), esta_vacia().
    Implementación completamente distinta: un diccionario {posición: valor} y un contador."""
    def __init__(self):
        self._datos = {}
        self._tope = -1

    def apilar(self, x):
        self._tope += 1
        self._datos[self._tope] = x

    def desapilar(self):
        x = self._datos.pop(self._tope)
        self._tope -= 1
        return x

    def ver_tope(self):
        return self._datos[self._tope]

    def esta_vacia(self):
        return self._tope == -1


def usar_pila(pila):
    """Esta función NO SABE (ni le importa) cuál implementación recibió."""
    pila.apilar(1)
    pila.apilar(2)
    pila.apilar(3)
    print("Tope:", pila.ver_tope())
    print("Desapilado:", pila.desapilar())
    print("¿Vacía?", pila.esta_vacia())


print("--- Usando PilaConLista ---")
usar_pila(PilaConLista())

print("--- Usando PilaConDiccionario ---")
usar_pila(PilaConDiccionario())
```

## 🎯 Cómo elegir qué ADT necesitas: mira el patrón de acceso

Dado un problema nuevo, la pregunta correcta **nunca** es "¿cómo lo implemento?" — es primero **"¿qué operaciones necesito, y en qué orden necesito acceder a los datos?"**. Esa pregunta, respondida con cuidado, casi siempre te dice qué ADT necesitas, mucho antes de pensar en código.

| Patrón de acceso que pide el problema | ADT que necesitas | Pista para reconocerlo |
|---|---|---|
| "Deshacer la última acción", "regresar al paso anterior" | **Pila** (LIFO) | Solo te interesa **lo último que agregaste** |
| "Atender en el orden en que llegaron", "procesar en orden de llegada" | **Cola** (FIFO) | Solo te interesa **lo primero que agregaste** |
| "Atender primero lo más urgente/importante", sin importar cuándo llegó | **Cola de prioridad** | El orden de salida depende de una **prioridad**, no de la llegada |
| "Guardar en orden y poder acceder por posición" | **Lista** | Te importa la **posición** de cada elemento |
| "Saber si ya vi este elemento antes", "eliminar duplicados" | **Conjunto** | Solo te importa **pertenencia**, no orden ni repeticiones |
| "Buscar rápido el valor asociado a un nombre/código/id" | **Diccionario / Mapa** | Accedes por una **clave**, no por posición |
| "Modelar relaciones o conexiones entre elementos" (rutas, amistades, dependencias) | **Grafo** | Lo que importa son las **conexiones**, no una jerarquía ni un orden lineal |
| "Organizar en niveles de padres e hijos" | **Árbol** | Hay una **jerarquía** clara (una raíz, y cada nodo tiene "hijos") |

📌 Nota que en ningún momento la tabla menciona *cómo* vas a construir cada estructura — eso es, otra vez, la separación entre interfaz e implementación. Elegir el ADT es un problema de **análisis del patrón de acceso**; elegir la implementación (arreglo, lista enlazada, etc.) viene después, y de hecho la vas a estudiar por separado en cada una de las próximas unidades del curso.
