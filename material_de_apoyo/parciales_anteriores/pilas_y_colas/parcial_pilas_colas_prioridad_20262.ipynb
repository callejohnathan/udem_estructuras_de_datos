### Enunciado: Sistema de despacho de TransAndes

La empresa de transporte de carga **TransAndes** administra el despacho de sus vehículos entre un centro de distribución (zona de cargue) y un patio de destino (zona de descargue). Deben construir un sistema controlado por un **menú interactivo por consola** que administre el ciclo de vida completo de un vehículo: su creación, su cargue con paquetes, su salida a viaje y su descargue en el patio de destino.

El menú debe ejecutarse en un ciclo que se repita hasta que el usuario elija salir, con al menos estas opciones:

1. Generar vehículo
2. Cargar vehículo
3. Poner en viaje
4. Descargar vehículo
5. Salir

#### 1. Generar vehículo

Crea un nuevo vehículo y lo deja disponible para ser cargado. El usuario indica su **capacidad máxima** (kg); el sistema le asigna un identificador único.

Hay una única bahía de cargue en el centro de distribución, así que los vehículos deben pasar por ella exactamente en el mismo orden en que fueron generados: no se puede cargar uno más reciente si todavía hay uno generado antes que sigue esperando su turno para ser cargado.

#### 2. Cargar vehículo

Carga, en una sola operación, al vehículo al que le corresponde el turno de cargue (según la regla de la opción anterior) con un lote de paquetes que el usuario ingresa en ese momento. De cada paquete solo se indica su **peso** (kg) y su **valor**.

Los paquetes deben introducirse al vehículo por la parte trasera, uno después de otro, en el mismo orden en que aparecen en el lote. Físicamente **solo se puede retirar el último paquete que fue puesto**, mientras no se hayan retirado antes todos los que se pusieron después que él: al descargar el vehículo (opción 4), los paquetes deben salir exactamente en el **orden inverso** al que fueron cargados.

Si la suma de los pesos del lote supera la capacidad máxima del vehículo, la operación completa debe rechazarse (no se carga ningún paquete del lote) y se debe informar al usuario. Una vez cargado, el vehículo queda listo para salir de viaje (opción 3) y deja libre la bahía de cargue para el siguiente vehículo en espera.

#### 3. Poner en viaje

Envía a un vehículo ya cargado hacia el patio de destino. El usuario indica el identificador del vehículo (debe existir y ya debe haber sido cargado). A partir de este momento, el vehículo queda a la espera de un cupo de descargue en el patio de destino.

#### 4. Descargar vehículo

En el patio de destino solo hay una bahía de descargue disponible a la vez. Entre todos los vehículos que ya llegaron y siguen esperando ser descargados, se debe atender siempre primero al que traiga la carga de **mayor valor total** (la suma de los valores de todos sus paquetes), sin importar cuál haya llegado primero al patio.

Al descargar el vehículo seleccionado, se deben retirar todos sus paquetes en el orden inverso al que fueron cargados, mostrando la lista completa en el orden en que van saliendo del vehículo. Si en ese momento no hay ningún vehículo esperando en el patio de destino, el sistema debe informarlo.

#### 5. Salir

Termina la ejecución del programa.

    -> V1 cargado

    Lote para V2, en este orden: P3(peso 40, valor 300), P4(peso 60, valor 150)
    -> V2 cargado

> Descargar vehículo
    -> Entre V1 (valor total 1000) y V2 (valor total 450) esperando en el patio, se elige V1.
    -> Paquetes de V1 en orden de salida (inverso al que fueron cargados): P2, P1

#### Consideraciones y restricciones de complejidad

Sea `v` el número de vehículos que en un momento dado están esperando ser descargados en el patio de destino, y `k` el número de paquetes cargados en un vehículo.

- Generar vehículo: O(1).
- Cargar vehículo: O(k) sobre los k paquetes del lote.
- Poner en viaje: O(log v).
- Descargar vehículo: encontrar el vehículo a descargar debe costar O(log v) (no se permite recorrer todos los vehículos en espera para hallar el de mayor valor); una vez encontrado, retirar sus k paquetes debe costar O(k) (no se permite reordenar ni ordenar los paquetes en el momento de descargar).
- El espacio adicional usado por el sistema debe ser proporcional a la cantidad de vehículos y paquetes activos en cada momento (O(n) en total), sin estructuras redundantes de tamaño mayor a eso.
