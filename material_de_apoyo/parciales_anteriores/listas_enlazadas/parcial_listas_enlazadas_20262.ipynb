## Ejercicio 1 — Retiro de productos del mercado (recall)

Una tienda en línea mantiene su catálogo de productos activos como un objeto de lista enlazada simple, donde cada nodo contiene el código de un producto. Cuando el fabricante detecta un defecto de seguridad, emite una lista de retiro (*recall*) con los códigos de los productos afectados, la cual puede tener repeticiones o incluir productos que ya no estén en el catálogo. La tienda necesita depurar su catálogo eliminando todo producto (incluyendo posibles repeticiones) que aparezca en la lista de retiro.

Su función recibe como parámetros el objeto de la lista enlazada simple que representa el catálogo y la lista de retiro, y debe obtener el catálogo resultante.

**Ejemplo:**

- Catálogo: `101 -> 205 -> 309 -> 412 -> 205 -> 88`
- Retiro: `[205, 412, 999]`
- Catálogo resultante: `101 -> 309 -> 88`

## Ejercicio 2 — Política de distanciamiento en un cine

Un sistema de reservas de un cine guarda los asientos ya reservados de una fila como un objeto de lista doblemente enlazada, ordenada ascendentemente por número de asiento. Por política de accesibilidad, no se permite que dos reservas queden en asientos físicamente contiguos: si dos nodos consecutivos de la lista tienen una diferencia absoluta de valor igual a 1, debe insertarse entre ellos un nodo "asiento bloqueado", con un valor arbitrario que usted puede definir.

Su función recibe como parámetro el objeto de la lista doblemente enlazada ya construida y debe aplicar esta política, obteniendo la lista resultante.

**Ejemplo** (usando `-1` como valor arbitrario de asiento bloqueado):

- Asientos: `10 <-> 11 <-> 15 <-> 16 <-> 20`
- Lista resultante: `10 <-> -1 <-> 11 <-> 15 <-> -1 <-> 16 <-> 20`

## Ejercicio 3 — Rotación de turnos de guardia

Una empresa de soporte técnico organiza los turnos de guardia (*on-call*) de su equipo como un objeto de lista circular, donde cada nodo representa a un ingeniero y el orden de la lista define la secuencia en que se activan las guardias, semana tras semana, indefinidamente. El equipo ha crecido demasiado y la gerencia decide dividirlo en dos equipos de guardia independientes, cada uno con su propia rotación circular, conservando el orden relativo original y dejando los tamaños lo más equilibrados posible.

Su función recibe como parámetro el objeto de la lista circular ya construida con el equipo completo y debe retornar los dos objetos de lista circular correspondientes a los nuevos equipos.

**Ejemplo:**

- Equipo completo: `A -> B -> C -> D -> E -> (vuelve a A)`
- Equipo 1: `A -> B -> C -> (vuelve a A)`
- Equipo 2: `D -> E -> (vuelve a D)`
