```python
# EJERCICIO 2: Análisis de Complejidad
# Determina la complejidad temporal O(...) de la función respecto al tamaño de 'lista'.

def encontrar_maximo(lista):
    if not lista:
        return None
    
    maximo = lista[0]
    for elemento in lista:
        if elemento > maximo:
            maximo = elemento
            
    return maximo

# ¿Cuál es la complejidad temporal O(...) de esta función?
# Respuesta del estudiante: ____________________
```

```python
# EJERCICIO 3: Análisis de Complejidad
# Determina la complejidad temporal O(...) de esta función respecto a 'n'.

def imprimir_pares_combinados(n):
    for i in range(n):
        for j in range(n):
            print(f"Par: ({i}, {j})")

# ¿Cuál es la complejidad temporal O(...) de esta función?
# Respuesta del estudiante: ____________________
```

```python
# EJERCICIO 4: Análisis de Complejidad
# Analiza cuántas veces se ejecuta el ciclo en función de 'n'.

def dividir_hasta_uno(n):
    pasos = 0
    mientras_numero = n
    
    while mientras_numero > 1:
        mientras_numero = mientras_numero // 2
        pasos += 1
        
    return pasos

# ¿Cuál es la complejidad temporal O(...) de esta función?
# Respuesta del estudiante: ____________________
```

```python
# EJERCICIO 5: Análisis de Complejidad
# Determina la complejidad total. Pista: Recuerda la regla del término dominante.

def procesar_datos(lista):
    n = len(lista)
    
    # Parte A
    print(f"El primer elemento es: {lista[0]}")
    
    # Parte B
    for elemento in lista:
        print(f"Procesando: {elemento}")
        
    # Parte C
    for i in range(n):
        print(f"Resumen {i}: {lista[i]}")

# ¿Cuál es la complejidad de cada parte y la complejidad TOTAL O(...)?
# Parte A: ____ | Parte B: ____ | Parte C: ____ | TOTAL: ____
```
