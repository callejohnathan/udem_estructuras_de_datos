# **Lista Enlazada (Linked List)**  

#### **¿Qué es una Lista Enlazada?**  
Una **lista enlazada** (en inglés, *linked list*) es una estructura de datos lineal que consiste en una secuencia de elementos llamados **nodos**, donde cada nodo contiene dos partes principales:  
1. **Dato**: La información almacenada en el nodo.  
2. **Referencia (o enlace)**: Un puntero que indica la dirección del siguiente nodo en la lista.  

A diferencia de los **arreglos o listas tradicionales**, donde los elementos están almacenados en ubicaciones contiguas de memoria, en una lista enlazada cada nodo puede estar en cualquier lugar de la memoria y los nodos están conectados mediante referencias.

---

#### **¿Por qué usar una Lista Enlazada?**  
Las listas enlazadas son útiles en situaciones donde se requiere:  
- **Inserción y eliminación dinámica** de elementos sin necesidad de reorganizar la estructura.  
- **Optimización del uso de memoria**, ya que no requieren un bloque de memoria contigua como los arreglos.  
- **Crecimiento dinámico**, sin la necesidad de definir un tamaño fijo desde el inicio.  

Sin embargo, también tienen **desventajas**, como una mayor complejidad en el acceso a elementos en comparación con un arreglo, ya que se requiere recorrer la lista nodo por nodo para encontrar un elemento específico.

---

### **Tipos de Listas Enlazadas**
Existen varias formas de organizar una lista enlazada dependiendo de la cantidad de referencias que tenga cada nodo y cómo están conectados entre sí.  

#### **1. Lista Enlazada Simple (Singly Linked List)**  
En esta estructura, cada nodo tiene una referencia al **siguiente nodo** en la lista, pero no hay referencia al nodo anterior.  

📌 **Características:**  
✔ Permite recorrer la lista en **una sola dirección** (de la cabeza hacia el final).  
✔ **Más eficiente en memoria** que otros tipos de listas enlazadas, ya que solo almacena una referencia por nodo.  
❌ **No permite retroceder** en la lista fácilmente.  

#### **2. Lista Enlazada Doble (Doubly Linked List)**  
Cada nodo tiene dos referencias:  
- Una referencia al **siguiente nodo**.  
- Una referencia al **nodo anterior**.  

📌 **Características:**  
✔ Se puede recorrer en **ambas direcciones**.  
✔ **Mayor flexibilidad** en operaciones como inserción y eliminación.  
❌ Ocupa **más memoria** debido al almacenamiento de dos referencias por nodo.  

#### **3. Lista Enlazada Circular (Circular Linked List)**  
Es una variación en la que el último nodo de la lista apunta de vuelta al primer nodo, formando un ciclo. Puede ser:  
- **Circular simple**: El último nodo apunta al primer nodo, pero los nodos solo tienen una referencia al siguiente.  
- **Circular doble**: Tanto el primer como el último nodo están conectados en ambas direcciones.  

📌 **Características:**  
✔ **No tiene un punto final definido**, lo que la hace útil para ciertas aplicaciones como programación de tareas cíclicas.  
✔ Puede ser más eficiente en algunos casos donde se necesita un acceso continuo.  
❌ Puede ser **más difícil de gestionar** en términos de evitar ciclos infinitos.  

---

### **Comparación con los Arreglos**  
| **Característica** | **Lista Enlazada** | **Arreglo** |
|-------------------|-----------------|----------|
| **Ubicación en memoria** | Distribuida (nodos pueden estar en cualquier lugar) | Contigua (elementos en posiciones consecutivas) |
| **Acceso a elementos** | Secuencial (se debe recorrer desde el inicio hasta encontrar el elemento) | Índice directo (O(1)) |
| **Inserción/Eliminación** | O(1) en el mejor caso (cuando se tiene la referencia al nodo previo) | O(n) en el peor caso (requiere desplazamiento de elementos) |
| **Tamaño** | Dinámico (puede crecer sin restricciones) | Fijo o redimensionable (requiere reserva de memoria) |
| **Uso de memoria** | Más consumo (por referencias adicionales) | Más eficiente en almacenamiento puro de datos |

---

### **Aplicaciones de las Listas Enlazadas**
Las listas enlazadas tienen múltiples aplicaciones en computación, entre ellas:  
✔ **Gestión de memoria dinámica**: Utilizadas en la implementación de estructuras de datos como pilas (*stacks*) y colas (*queues*).  
✔ **Representación de grandes estructuras**: En editores de texto para manejar líneas de texto dinámicamente.  
✔ **Tablas hash**: En la implementación de técnicas de resolución de colisiones con encadenamiento.  
✔ **Sistema de navegación**: Como en listas de reproducción, historial de navegación en navegadores web, etc.  
✔ **Estructuras gráficas y árboles**: Son fundamentales en estructuras como árboles y grafos, donde los nodos están interconectados.

En la carpeta anterior vimos el concepto de **Tipo Abstracto de Datos (ADT)**: separar *qué* hace una estructura (su interfaz) de *cómo* lo hace por dentro (su implementación). Una lista (en el sentido de ADT — "colección ordenada de elementos") se puede implementar con un **arreglo** o con una **lista enlazada**, y ambas cumplen el mismo contrato. La pregunta que te debes entrenar a hacerte **antes** de escribir una sola línea de código es:

> Para *este* problema, ¿qué se hace más veces: **acceder por índice** o **insertar/eliminar en posiciones que no son el final**?

- Si necesitas acceso aleatorio rápido (`lista[i]`) y casi no insertas/eliminas en el medio → **arreglo**.
- Si insertas/eliminas mucho en posiciones arbitrarias (cabeza, medio) y puedes darte el lujo de recorrer para llegar hasta allá → **lista enlazada**.

Esta pregunta de reconocimiento es, en sí misma, parte del modelo mental que vas a construir en este notebook. No basta con saber *cómo* se implementa una lista enlazada: tienes que reconocer *cuándo* te conviene usarla.

## 🧩 El modelo mental que vas a usar TODO el tema: cajas y flechas

Antes de entrar en la implementación formal, instalemos el hábito que te va a servir para **cualquier** operación sobre listas enlazadas (insertar, borrar, revertir, detectar un ciclo, lo que sea). Nunca escribas código de punteros directamente: primero resuélvelo en papel con cajas (nodos) y flechas (referencias).

Para **cualquier operación**, respóndete estas 4 preguntas, en este orden, antes de tocar el teclado:

1. **¿Qué nodos están involucrados?** — nombra cada uno con una letra o variable (`a`, `b`, `nuevo`, `prev`, `curr`...). Si no puedes nombrarlos, no has entendido el problema todavía.
2. **¿Qué apunta a qué *antes*?** — dibuja el estado inicial. Cada flecha es una referencia (`.next`, `.prev`, `head`, `tail`).
3. **¿Qué debe apuntar a qué *después*?** — dibuja el estado final. Compáralo con el estado inicial: ¿qué flechas cambiaron?
4. **¿En qué orden reasigno las flechas para no perder ninguna referencia en el camino?** — esta es la pregunta que más cuesta, y es la que casi siempre produce el bug.

### 🪤 El error clásico: sobrescribir antes de guardar

Supón que quieres insertar un nodo nuevo `X` **al inicio** de la lista `A → B → C`. Si escribes esto sin haber dibujado nada primero:

```python
# ❌ MAL — perdiste la lista
head = nuevo_nodo          # 1) mueves head ANTES de conectar nuevo_nodo con el resto
nuevo_nodo.next = head     # 2) demasiado tarde: head YA es nuevo_nodo, así que esto lo apunta a sí mismo
```

Al ejecutar la línea 1, la única referencia que tenías al resto de la lista (`A → B → C`) se sobrescribe. Cuando en la línea 2 intentas usar `head` para conectar `nuevo_nodo`, `head` ya no apunta a `A`: apunta al propio `nuevo_nodo`. **Perdiste `A`, `B` y `C` para siempre** (y de paso creaste un nodo que se apunta a sí mismo).

```python
# ✅ BIEN — primero conectas, después mueves head
nuevo_nodo.next = head     # 1) primero, mientras head TODAVÍA apunta a A, lo copio a nuevo_nodo.next
head = nuevo_nodo          # 2) ahora sí puedo mover head, ya no necesito su valor anterior
```

**La regla general:** antes de sobrescribir una referencia, pregúntate "¿voy a necesitar el valor viejo de esta flecha más adelante?" Si la respuesta es sí, guárdalo primero (en una variable temporal, o úsalo para conectar otra cosa) **antes** de sobrescribirla.

Esto que acabas de ver con `head`/`next` es exactamente lo mismo que vas a tener que razonar en cabeza, cola, doblemente enlazadas, circulares... Practica dibujando el "antes" y el "después" en papel para cada operación de este notebook, incluso las que ya vienen resueltas: el objetivo no es que memorices el código, sino que puedas reconstruirlo desde el dibujo.

# **Modelo de Pensamiento para Implementar una Lista Enlazada Simple (Singly Linked List)**
---
Antes de escribir código, es crucial entender **cómo pensamos** la implementación de una lista enlazada simple (*singly linked list*). Este proceso nos ayuda a construir el **tipo abstracto de datos (ADT) LinkedList** de manera lógica y organizada.

---

## **1️⃣ Comprender la Estructura Fundamental**
Una **lista enlazada simple** es una secuencia de **nodos** conectados mediante referencias (punteros). Cada nodo almacena:
1. **Un dato** (el valor que queremos guardar).
2. **Una referencia al siguiente nodo** en la lista.

🔹 **¿Cómo se ve esto conceptualmente?**  
📌 Imagina un tren donde cada vagón contiene información y una conexión al siguiente vagón.

---
## **2️⃣ Diseñar la Representación del Nodo**
El núcleo de una lista enlazada es su **nodo** (*Node*). Este es el bloque fundamental de nuestra estructura de datos.

📌 **Pensamiento clave**: Un nodo individual debe poder:
✔ **Almacenar datos**.  
✔ **Enlazarse al siguiente nodo** (o ser el último nodo, en cuyo caso no apunta a nada).

🔍 **Visualización**:
```
[ Dato | → ] → [ Dato | → ] → [ Dato | None ]
```
Cada nodo se conecta al siguiente mediante una referencia (*puntero*), y el último nodo apunta a **nada (None)**, indicando el final de la lista.

---
## **3️⃣ Definir la Estructura de la Lista Enlazada**
La lista enlazada es una **colección de nodos** que mantiene un **puntero al primer nodo**, llamado **cabeza (head)**.

📌 **Pensamiento clave**:  
✔ La lista solo necesita **recordar el primer nodo**.  
✔ Si la lista está vacía, el puntero `head` será `None`.  
✔ Para recorrer la lista, seguimos los punteros de nodo en nodo.

🔍 **Visualización de una lista vacía vs. una lista con elementos**:
```
Lista vacía:
head → None

Lista con elementos:
head → [ A | → ] → [ B | → ] → [ C | None ]
```
---

## **4️⃣ Pensar en las Operaciones Básicas**
Ahora que tenemos una estructura, **¿qué operaciones queremos soportar?**  
Como estamos construyendo un **Tipo Abstracto de Datos (ADT)**, debemos definir **qué se puede hacer con la lista**.  

🔹 **Operaciones fundamentales:**
1. **Insertar un elemento**  
   - Al inicio (inserción en cabeza).  
   - Al final (requiere recorrer la lista).  
   - En una posición específica (requiere recorrer hasta esa posición).  
2. **Eliminar un elemento**  
   - De la cabeza (ajustar el puntero `head`).  
   - Del final (recorrer y actualizar referencias).  
   - En una posición específica.  
3. **Buscar un elemento**  
   - Recorrer la lista hasta encontrar el dato.  
4. **Mostrar la lista**  
   - Recorrer todos los nodos y devolver los valores.  

---
## **5️⃣ Modelar el Pensamiento para Cada Operación**
📌 **Para cada operación, seguimos un enfoque de "preguntas clave"**.  

### **✏ 1. Insertar un elemento al inicio**
🔍 **Preguntas clave:**  
✅ ¿Dónde está la cabeza actual?  
✅ ¿Cómo hago que el nuevo nodo sea el primero sin perder la referencia al resto?  
✅ ¿Qué pasa si la lista está vacía?  

📌 **Estrategia:**  
1. Crear un nuevo nodo.  
2. Hacer que su referencia apunte al nodo que actualmente es la cabeza.  
3. Actualizar `head` para que apunte al nuevo nodo.  

---

### **✏ 2. Insertar un elemento al final**
🔍 **Preguntas clave:**  
✅ ¿Cómo sé cuál es el último nodo?  
✅ ¿Cómo evito perder la referencia a los nodos previos?  

📌 **Estrategia:**  
1. Crear un nuevo nodo.  
2. Recorrer la lista hasta llegar al último nodo.  
3. Hacer que la referencia del último nodo apunte al nuevo nodo.  

---

### **✏ 3. Eliminar un elemento de la cabeza**
🔍 **Preguntas clave:**  
✅ ¿Cómo elimino el primer nodo sin perder el resto de la lista?  
✅ ¿Qué pasa si la lista está vacía o tiene solo un elemento?  

📌 **Estrategia:**  
1. Guardar la referencia del nodo actual en `head`.  
2. Mover `head` al siguiente nodo.  
3. (Opcional) Liberar el nodo eliminado.  

---

### **✏ 4. Eliminar un elemento del final**
🔍 **Preguntas clave:**  
✅ ¿Cómo encuentro el nodo anterior al último?  
✅ ¿Cómo actualizo la referencia sin perder el resto de la lista?  

📌 **Estrategia:**  
1. Recorrer la lista hasta encontrar el penúltimo nodo.  
2. Hacer que su referencia apunte a `None` (eliminando el último nodo).  

---

### **✏ 5. Buscar un elemento**
🔍 **Preguntas clave:**  
✅ ¿Cómo recorro la lista eficientemente?  
✅ ¿Qué hago si no encuentro el elemento?  

📌 **Estrategia:**  
1. Empezar desde `head` y recorrer nodo por nodo.  
2. Si el dato coincide, detenerse y devolver el nodo.  
3. Si se llega al final sin encontrarlo, devolver un mensaje de "No encontrado".  

---

### **✏ 6. Mostrar la lista**
🔍 **Preguntas clave:**  
✅ ¿Cómo recorro todos los nodos sin perder referencias?  

📌 **Estrategia:**  
1. Iniciar en `head`.  
2. Mientras el nodo actual no sea `None`, imprimir su dato.  
3. Moverse al siguiente nodo.  

---
## **6️⃣ Evaluar Casos Especiales**
Antes de implementar, es importante considerar **casos extremos** que podrían causar errores:
✔ **Lista vacía**: ¿Qué sucede si intentamos eliminar o buscar en una lista sin elementos?  
✔ **Un solo nodo**: ¿Cómo asegurarnos de que `head` se actualiza correctamente al eliminar?  
✔ **Eliminar el último nodo**: ¿Cómo evitar referencias perdidas?  
✔ **Posiciones inválidas**: ¿Cómo manejar intentos de acceso a posiciones inexistentes?  



## Implementación base de lista enlazada simple

````python
from typing import Any
class Node:
  def __init__(self, value: Any, next = None):
    self.value = value
    self.next = next

  def __repr__(self) -> str:
    return f"{self.value} -> {self.next}"

class LinkedList:
  def __init__(self, head = None):
    self.head = head
    self.size: int = 0

  def delete_at_pos(self, pos: int) -> None:
    if(self.head is None):
      return

    if(pos > 0 and pos >= self.size):
      return

    if(pos == 0):
      old_head_next = self.head.next
      self.head.next = None
      self.head = old_head_next
    else:
      current = self.head
      for i in range(pos - 1):
        current = current.next

      old_current_next_next = current.next.next
      current.next.next = None
      current.next = old_current_next_next

    self.size -= 1

  def append(self, value: Any) -> None:
    if(self.head is None):
      self.head = Node(value) #creando y asignando un nuevo nodo a la cabeza
    else:
      #tengo que llegar al último
      current_node = self.head
      while(current_node.next is not None):
        current_node = current_node.next

      current_node.next = Node(value)

    self.size += 1


  def traverse(self) -> None:

    current_node = self.head
    while(current_node is not None):
      print(current_node.value)
      current_node = current_node.next

  def __repr__(self) -> str:
    return f"{self.head}"



ll = LinkedList()
ll.append(1)
ll.append(2)
ll.append(3)
ll.append(5)
print(ll)
ll.delete_at_pos(4)
print(ll)
````

## ⚙️ Analicemos el diseño: ¿por qué esta implementación no tiene `tail`?

Mira de nuevo el método `append` que acabas de ejecutar arriba. Para agregar un elemento, **recorre toda la lista desde `head` hasta encontrar el último nodo** (`while current_node.next is not None`). Eso significa que `append` es **O(n)**, sin importar que solo estemos agregando *un* elemento al final.

Como parte del modelo de "por qué se diseña así", pregúntate:
- ¿Qué información me falta para no tener que recorrer toda la lista cada vez?
- Si guardara esa información, ¿qué tendría que actualizar cada vez que cambia el último nodo?

La respuesta es guardar una referencia directa al último nodo: `self.tail`. Con eso, `append` deja de recorrer la lista: simplemente conecta `tail.next` al nuevo nodo y mueve `tail`. Es el mismo principio de "cajas y flechas" que ya vimos: mantener una referencia extra te ahorra tener que "redescubrir" un nodo recorriendo desde `head`.

**El costo:** ahora tienes que ser disciplinado y actualizar `tail` en *cada* operación que pueda cambiar cuál es el último nodo (append, insertar al final, eliminar el último nodo, y el caso borde de que la lista quede vacía). Si olvidas actualizarlo en un solo lugar, `tail` queda "desincronizado": vas a tener bugs silenciosos (la lista sigue "funcionando" pero apunta a un nodo que ya no es el último).

Esto es exactamente lo que hace `SinglyLinkedList` en `base.py`, la implementación que vas a usar en los ejercicios: fíjate que mantiene `head`, `tail` y `size` sincronizados en cada operación (`append`, `delete_and_return_first`, `delete_and_return_last`).

Reconstruyamos la lista de arriba, pero llevando `tail`:

````python
from typing import Any


class NodeV2:
  def __init__(self, value: Any, next=None):
    self.value = value
    self.next = next

  def __repr__(self) -> str:
    return f"{self.value} -> {self.next}"


class LinkedListV2:
  """Misma lista de antes, pero ahora con `tail` para que append sea O(1)."""

  def __init__(self):
    self.head = None
    self.tail = None  # <-- la única diferencia estructural
    self.size = 0

  def append(self, value: Any) -> None:
    new_node = NodeV2(value)
    if self.head is None:
      # antes: self.head = Node(value)
      # ahora: head Y tail apuntan al mismo (único) nodo
      self.head = self.tail = new_node
    else:
      # ya NO recorremos la lista: tail YA es el último nodo
      self.tail.next = new_node
      self.tail = new_node
    self.size += 1

  def __repr__(self) -> str:
    return f"{self.head}"


ll2 = LinkedListV2()
ll2.append(1)
ll2.append(2)
ll2.append(3)
print(ll2, "| tail:", ll2.tail)
````

## 🛡️ Nodo dummy / centinela: eliminar los `if` especiales

Fíjate en `delete_at_pos` (la implementación original de arriba): tiene una rama separada para `pos == 0` (borrar la cabeza) y otra para el resto. Esa asimetría — "la cabeza es un caso especial" — aparece una y otra vez en listas enlazadas: insertar al inicio, borrar al inicio, listas que empiezan vacías... siempre hay un `if` extra por culpa de `head`.

**Idea del nodo dummy (o centinela):** en vez de que `head` apunte directamente al primer nodo *real*, la lista mantiene un nodo "falso" fijo al principio (`self.dummy`), cuyo único trabajo es apuntar (`.next`) al primer nodo real (o a `None` si la lista está vacía). Así, "insertar/borrar en la cabeza" deja de ser un caso especial: **siempre hay un nodo antes** del nodo que te interesa, incluso cuando ese nodo es el primero real.

🔍 **Antes (con `head` especial):**
```
Lista vacía:      head → None
Con elementos:     head → [A] → [B] → None
```
Borrar el primero requiere mover `head` — un caso aparte.

🔍 **Después (con dummy):**
```
Lista vacía:      dummy → None
Con elementos:     dummy → [A] → [B] → None
```
Borrar "el primero" ahora es *exactamente* la misma operación que borrar cualquier nodo intermedio: `prev.next = prev.next.next`, donde `prev` puede ser el `dummy` o un nodo real. Ya no necesitas preguntar "¿es la cabeza?".

**¿Cuándo vale la pena pagar el costo de un nodo extra?** Cuando vas a hacer muchas inserciones/eliminaciones cerca del inicio y quieres que tu código no esté lleno de casos borde para `pos == 0`. En estructuras pequeñas o de solo lectura no suele justificarse — es una decisión de diseño, no una regla obligatoria.

````python
from typing import Any


class LinkedListWithDummy:
  """delete_at_pos sin casos especiales para pos == 0, gracias al dummy."""

  def __init__(self):
    self.dummy = NodeV2(None)  # nodo centinela: nunca guarda un dato real
    self.size = 0

  def append(self, value: Any) -> None:
    current = self.dummy
    while current.next is not None:
      current = current.next
    current.next = NodeV2(value)
    self.size += 1

  def delete_at_pos(self, pos: int) -> None:
    if pos < 0 or pos >= self.size:
      return
    prev = self.dummy
    for _ in range(pos):
      prev = prev.next
    # esta línea funciona IGUAL para pos == 0 que para cualquier otra posición,
    # porque siempre hay un nodo "antes" (el dummy, si pos == 0)
    prev.next = prev.next.next
    self.size -= 1

  def __repr__(self) -> str:
    return f"{self.dummy.next}"


lwd = LinkedListWithDummy()
for v in [10, 20, 30, 40]:
  lwd.append(v)
print(lwd)
lwd.delete_at_pos(0)  # borrar la cabeza: ningún if especial
print(lwd)
````

## 🐢🐇 Dos punteros: lento y rápido

Otro modelo mental que se reutiliza muchísimo (detectar ciclos, encontrar el punto medio, y otras variantes que verás en los ejercicios) es el de **dos punteros que avanzan a distinta velocidad** sobre la misma lista: uno "lento" (`slow`) que avanza un nodo a la vez, y uno "rápido" (`fast`) que avanza dos nodos a la vez.

**La pregunta clave que debes hacerte:** si `fast` avanza el doble de rápido que `slow`, ¿qué implica eso?

- **Para encontrar el punto medio:** cuando `fast` llegue al final, `slow` habrá recorrido exactamente la mitad del camino — porque avanzó a la mitad de la velocidad. No necesitas contar cuántos nodos tiene la lista de antemano ni recorrerla dos veces.
- **Para detectar un ciclo:** si la lista tiene un ciclo, `fast` "da la vuelta" dentro del ciclo más rápido que `slow`. Tarde o temprano `fast` **alcanza** a `slow` desde atrás (como un corredor más veloz dando vueltas a una pista circular). Si `fast` llega a `None`, es que no había ciclo. Si `slow is fast` en algún momento (sin haber empezado juntos), había un ciclo. Esto se llama el **algoritmo de Floyd** (tortuga y liebre).

📌 **Antes de programar esto, respóndete:**
1. ¿Cuánto avanza cada puntero por iteración? (`slow` 1, `fast` 2)
2. ¿Cuál es la condición de parada normal? (`fast` o `fast.next` llegan a `None`)
3. ¿Cuál es la condición de "encontré lo que buscaba"? (para ciclo: `slow is fast`; para medio: `fast` ya no puede avanzar más)

````python
def find_middle(head):
  """Devuelve el nodo del medio (el segundo, si hay dos de en medio)."""
  slow = fast = head
  while fast is not None and fast.next is not None:
    slow = slow.next
    fast = fast.next.next
  return slow


def has_cycle(head) -> bool:
  """Algoritmo de Floyd: ¿la lista tiene un ciclo?"""
  slow = fast = head
  while fast is not None and fast.next is not None:
    slow = slow.next
    fast = fast.next.next
    if slow is fast:
      return True
  return False


# Prueba con la lista sin ciclo que ya construimos
demo = LinkedListV2()
for v in [1, 2, 3, 4, 5]:
  demo.append(v)
print("Medio:", find_middle(demo.head))
print("¿Tiene ciclo?", has_cycle(demo.head))

# Construimos un ciclo a mano para probar has_cycle: 1 -> 2 -> 3 -> (vuelve a 2)
a, b, c = NodeV2(1), NodeV2(2), NodeV2(3)
a.next, b.next, c.next = b, c, b  # c apunta de vuelta a b: hay ciclo
print("¿Tiene ciclo?", has_cycle(a))
````

## **Modelo de Pensamiento para la Implementación de una Lista Doblemente Enlazada (Doubly Linked List)**

---

Antes de escribir código, es importante construir un **modelo mental sólido** sobre cómo funciona y cómo se implementa una **lista doblemente enlazada (Doubly Linked List, DLL)**. Esta estructura es más flexible que la lista enlazada simple, pero también requiere una mayor gestión de referencias.

---

## **1️⃣ Comprender la Estructura Fundamental**

En una **lista doblemente enlazada**, cada nodo contiene **dos referencias** en lugar de una:
1. **Dato**: Almacena la información del nodo.
2. **Referencia al siguiente nodo** (*next*): Apunta al nodo que sigue en la lista.
3. **Referencia al nodo anterior** (*prev*): Apunta al nodo anterior en la lista.

🔹 **Diferencia clave con la lista enlazada simple:**  
✔ **Permite recorrer la lista en ambas direcciones.**  
✔ **Facilita la eliminación y la inserción en el medio de la lista.**  
✔ **Consume más memoria debido a la referencia adicional por nodo.**  

🔍 **Visualización conceptual de una lista doblemente enlazada:**  
```
(None ← [ A ] →) ⇄ ([ B ] →) ⇄ ([ C ] → None)
```
Cada nodo tiene un enlace hacia adelante (**→**) y otro hacia atrás (**←**), formando una estructura bidireccional.

---

## **2️⃣ Diseñar la Representación del Nodo**

El nodo en una **Doubly Linked List** es similar al de la lista simple, pero con la **adición de una referencia al nodo anterior**.

📌 **Pensamiento clave:** Un nodo individual debe poder:
✔ **Almacenar datos.**  
✔ **Enlazarse al siguiente nodo.**  
✔ **Enlazarse al nodo anterior.**

🔍 **Visualización del Nodo en memoria:**
```
[ Prev | Dato | Next ]
```
Ejemplo con tres nodos:
```
(None ← [ A ] →) ⇄ ([ B ] →) ⇄ ([ C ] → None)
```

---
## **3️⃣ Definir la Estructura de la Lista Doblemente Enlazada**
La estructura de la **Doubly Linked List** requiere:
✔ **Un puntero al primer nodo** (`head`).  
✔ **(Opcional) Un puntero al último nodo** (`tail`), lo que facilita la inserción y eliminación desde el final.  

🔍 **Lista vacía vs. Lista con elementos**
```
Lista vacía:
head → None

Lista con elementos:
head → (None ← [ A ] →) ⇄ ([ B ] →) ⇄ ([ C ] → None)
```

📌 **Diferencias con una lista enlazada simple:**  
- Se puede recorrer en **ambos sentidos**.  
- Se pueden eliminar nodos más fácilmente sin necesidad de recorrer toda la lista.  
- Se usa más memoria porque cada nodo tiene dos referencias en lugar de una.

---
## **4️⃣ Pensar en las Operaciones Básicas**
### **Operaciones fundamentales**
Al igual que en la lista enlazada simple, necesitamos definir las operaciones clave:

1. **Insertar un nodo**
   - Al inicio (**antes de la cabeza**).
   - Al final (**después de la cola**).
   - En una posición específica.
2. **Eliminar un nodo**
   - De la cabeza.
   - De la cola.
   - En una posición específica.
3. **Buscar un nodo**
   - Recorrer la lista hasta encontrar el elemento.
4. **Mostrar la lista**
   - Recorrer los nodos de **inicio a fin**.
   - Recorrer los nodos de **fin a inicio** (gracias a la referencia `prev`).

---
## **5️⃣ Modelar el Pensamiento para Cada Operación**
📌 **Para cada operación, seguimos un enfoque de "preguntas clave"**.

### **✏ 1. Insertar un nodo al inicio**
🔍 **Preguntas clave:**  
✅ ¿Cómo se actualiza la cabeza de la lista?  
✅ ¿Cómo aseguramos que el nuevo nodo apunte al anterior primer nodo?  
✅ ¿Qué pasa si la lista está vacía?  

📌 **Estrategia:**  
1. Crear un nuevo nodo.  
2. Hacer que su `next` apunte al nodo que actualmente es `head`.  
3. Si la lista no está vacía, hacer que `prev` del nodo original apunte al nuevo nodo.  
4. Actualizar `head` para que apunte al nuevo nodo.  

🔍 **Ejemplo gráfico antes y después de la inserción:**  
**Antes:**  
```
(None ← [ A ] →) ⇄ ([ B ] →)
```
**Después de insertar "X" al inicio:**  
```
(None ← [ X ] →) ⇄ ([ A ] →) ⇄ ([ B ] →)
```

---

### **✏ 2. Insertar un nodo al final**
🔍 **Preguntas clave:**  
✅ ¿Cómo sabemos cuál es el último nodo?  
✅ ¿Cómo enlazamos el nuevo nodo al anterior último nodo?  
✅ ¿Cómo se actualiza la referencia `tail` si existe?  

📌 **Estrategia:**  
1. Crear un nuevo nodo.  
2. Recorrer la lista hasta el último nodo (`tail`).  
3. Hacer que `prev` del nuevo nodo apunte al nodo anterior (`tail`).  
4. Hacer que `next` del último nodo apunte al nuevo nodo.  
5. Si existe un `tail`, actualizarlo para que apunte al nuevo nodo.  

---

### **✏ 3. Eliminar el primer nodo**
🔍 **Preguntas clave:**  
✅ ¿Cómo actualizamos la cabeza de la lista?  
✅ ¿Qué sucede si la lista queda vacía después de la eliminación?  

📌 **Estrategia:**  
1. Guardar referencia del nodo actual (`head`).  
2. Mover `head` al siguiente nodo.  
3. Si `head` no es `None`, actualizar su `prev` a `None`.  
4. Liberar el nodo eliminado.  

🔍 **Ejemplo gráfico antes y después de eliminar el primer nodo:**  
**Antes:**  
```
(None ← [ A ] →) ⇄ ([ B ] →) ⇄ ([ C ] →)
```
**Después de eliminar "A":**  
```
(None ← [ B ] →) ⇄ ([ C ] →)
```

---

### **✏ 4. Eliminar el último nodo**
🔍 **Preguntas clave:**  
✅ ¿Cómo identificamos el penúltimo nodo?  
✅ ¿Cómo eliminamos la referencia al último nodo?  

📌 **Estrategia:**  
1. Recorrer la lista hasta el último nodo.  
2. Hacer que el `next` del penúltimo nodo apunte a `None`.  
3. (Opcional) Si hay `tail`, actualizarlo al penúltimo nodo.  

🔍 **Ejemplo gráfico antes y después de eliminar el último nodo:**  
**Antes:**  
```
(None ← [ A ] →) ⇄ ([ B ] →) ⇄ ([ C ] → None)
```
**Después de eliminar "C":**  
```
(None ← [ A ] →) ⇄ ([ B ] → None)
```

---

### **✏ 5. Buscar un nodo**
🔍 **Preguntas clave:**  
✅ ¿Recorremos la lista desde la cabeza o la cola?  
✅ ¿Cómo manejamos el caso en que el elemento no está presente?  

📌 **Estrategia:**  
1. Comenzar desde `head`.  
2. Recorrer la lista nodo por nodo hasta encontrar el valor.  
3. Si se encuentra, devolver el nodo. Si no, devolver un mensaje de "No encontrado".  

---

### **✏ 6. Mostrar la lista**
🔍 **Preguntas clave:**  
✅ ¿Cómo recorremos la lista en ambos sentidos?  

📌 **Estrategia:**  
1. Comenzar en `head`.  
2. Recorrer hasta `None`, imprimiendo cada valor.  
3. (Opcional) Si queremos mostrar en orden inverso, recorrer desde `tail` usando `prev`.  

---

## **6️⃣ Evaluar Casos Especiales**
✔ **Lista vacía**: ¿Qué sucede si intentamos eliminar o buscar?  
✔ **Un solo nodo**: ¿Cómo asegurarnos de que `head` y `tail` se actualicen correctamente?  
✔ **Eliminar el último nodo**: ¿Cómo evitar referencias inválidas?  



## Implementación base de lista enlazada doble

````python
from typing import Any
class DoublyNode:
  def __init__(self, value: Any, next = None, prev = None):
    self.value = value
    self.next = next
    self.prev = prev

  def __repr__(self) -> str:
    return f"{self.value} <-> {self.next}"

class DoublyLinkedList:
  def __init__(self):
    self.head = None
    self.size: int = 0

  def append(self, value: Any) -> None:
    if(self.head is None): #Está vacía
      self.head = DoublyNode(value) #creo el nuevo nodo y lo asigno a la cabeza
    else:
      current_node = self.head
      while(current_node.next is not None):
        current_node = current_node.next #saltar al nodo siguiente

      #cuando se rompe el ciclo --> estoy en el último nodo
      new_node = DoublyNode(value) #creo nuevo nodo
      current_node.next = new_node #se lo asigno al next del último (quedó en current_node)
      new_node.prev = current_node #asigno al previo del nuevo, el último (current_node)

    self.size += 1

  def __repr__(self) -> str:
    return f"{self.head}"

  def __len__(self) -> int:
    return self.size

dll = DoublyLinkedList()
dll.append(4) #0
dll.append(1) #1

print(dll)
len(dll)
````

> 🔁 **Mismo razonamiento que ya vimos con la lista simple:** este `append` también recorre toda la lista para encontrar el último nodo, así que también es O(n). Aplica la misma pregunta de diseño: ¿qué referencia extra evitaría ese recorrido? La respuesta es la misma — `self.tail` — y el mismo cuidado aplica: hay que mantenerlo sincronizado en cada operación que toque el final de la lista. Por eso `DoublyLinkedList` en `base.py` sí mantiene `head` **y** `tail`, y por eso vas a verla usada así en `taller_3` para implementar una pila (`push`/`pop` en O(1) usando `tail`) y una cola (`enqueue` en O(1) con `tail`, `dequeue` en O(1) con `head`).

# 🔴 Lista Enlazada Circular: concepto y modelo mental

## 🤔 ¿Qué es?

Una **lista enlazada circular** es una variante de lista enlazada donde **el último nodo (`tail`) apunta de nuevo al primero (`head`)**. No hay `None` al final: el recorrido “da la vuelta” indefinidamente.

**Casos típicos de uso:** turnos (round-robin), juegos por rondas, playlists en loop, algoritmos tipo *Josephus*, buffers circulares simples, etc.

---

## 🧠 Modelo mental (paso a paso)

### 1) Representación mínima

* **Nodos:** tienen al menos `data` y `next`.
* **Punteros de la lista:** `head` y `tail`.
* **Tres estados clave:**

  1. **Vacía:** `head = tail = None`.
  2. **Un solo nodo:** `head is tail` y `head.next = head`.
  3. **Con varios nodos:** `tail.next = head` siempre.

> 🔒 **Invariante central:** si la lista no está vacía, **`tail.next == head`**.

---

### 2) Recorrer sin caer en bucles infinitos

Como no existe `None` para detener, usa un patrón tipo **do-while**:

```python
cur = head
if cur is not None:
    while True:
        # usa cur.data
        cur = cur.next
        if cur is head:   # ¡ojo a la condición de corte!
            break
```

> Tip: si necesitas contar `n`, da **una vuelta exacta** y detente al volver a `head`.

---

### 3) Insertar al final (append)

Idea: “engancha” el nuevo nodo **entre `tail` y `head`** y mueve `tail`.

1. **Vacía:** crea `new.next = new`; `head = tail = new`.
2. **No vacía:**

   * `new.next = head`
   * `tail.next = new`
   * `tail = new`

> 🎯 No toques `head` en `append` (salvo si estaba vacía).

---

### 4) Insertar al inicio (push\_front) — opcional pero ilustrativo

1. **Vacía:** igual que arriba.
2. **No vacía:**

   * `new.next = head`
   * `tail.next = new`
   * `head = new`

> Clave: **`tail.next` debe seguir apuntando a `head`** tras la operación.

---

### 5) Eliminar (primer valor `x`)

* **Vacía:** nada que hacer.
* **Un solo nodo:** si coincide, `head = tail = None`.
* **Varios nodos:** recorre manteniendo `prev`/`cur` **una vuelta máxima**:

  * Si borras `head`: mueve `head = head.next` **y** `tail.next = head`.
  * Si borras `tail`: mueve `tail = prev` **y** `tail.next = head`.
  * En caso general: `prev.next = cur.next`.

> ⚠️ Asegura cubrir explícitamente los tres casos: **vacía**, **uno**, **varios**.

---

## 🧪 Checklist de implementación

* [ ] Estructura: `head`, `tail`, nodos con `data`, `next`.
* [ ] **Invariante:** si no está vacía, `tail.next == head`.
* [ ] `append`:

  * [ ] Vacía: `new.next = new`, `head = tail = new`
  * [ ] No vacía: `new.next = head`, `tail.next = new`, `tail = new`
* [ ] Recorrido: patrón **do-while** con corte al volver a `head`.
* [ ] Borrado:

  * [ ] Caso único nodo
  * [ ] Caso borro `head` (ajustar `tail.next`)
  * [ ] Caso borro `tail` (mover `tail`)
* [ ] `__repr__`: recorre **una vuelta** y corta al regresar a `head`.

---

## 🪤 Errores comunes (y cómo evitarlos)

* **Bucle infinito** al imprimir/recorrer: olvidas cortar al volver a `head`.
* **Perder la circularidad**: no repones `tail.next = head` tras insertar/eliminar en cabeza.
* **No actualizar `tail`** si borras el último o el nodo `tail`.
* **Olvidar el caso de 1 nodo**: debe apuntarse a sí mismo.

---

## ⏱️ Complejidad

* Con `tail`, **`append` = O(1)**.
* Borrar/buscar por valor: **O(n)**.
* Espacio adicional: **O(1)** (punteros).

---

## 🧩 Mini-pruebas mentales (rápidas)

1. **Vacía → append(7)**
   Esperado: `head=tail` y `head.next=head`.
2. **\[A → B → C → (head=A)] → append(D)**
   Esperado: `tail=D` y `D.next = A`.
3. **Eliminar `A` (head) en \[A → B → C → (A)]**
   Esperado: `head=B`, `tail.next=B`.
4. **Eliminar `C` (tail)**
   Esperado: `tail=B`, `B.next=head`.
