# 🧩 Ejercicio 4: Detección de ciclo en grafo no dirigido

## Descripción
Dado un grafo no dirigido, determina si existe **algún ciclo**.

Un ciclo es una secuencia de aristas donde se puede empezar en un nodo y volver al mismo nodo siguiendo las conexiones, sin repetir aristas.

## Ejemplo
Entrada:
```python
grafo = {
  0: [1, 2],
  1: [0, 2],
  2: [0, 1, 3],
  3: [2]
}
```
Salida esperada:
```python
True
```
(Aquí 0-1-2-0 es un ciclo.)

**🧭 Antes de programar:** el grafo es **no dirigido** — al recorrerlo con DFS, el nodo del que acabas de venir siempre va a aparecer como vecino ya visitado, y eso **no** es un ciclo. ¿Cómo distingues "volver sobre el mismo padre inmediato" de "encontrar un vecino visitado que no es tu padre" (que sí es un ciclo)? Simula a mano el DFS del grafo de ejemplo desde el nodo 0, anotando en cada paso quién es el "padre" de cada nodo en el recorrido, y señala exactamente en qué paso se detecta el ciclo 0-1-2-0.

Salida esperada:
```python
1
```
(El grupo {0,1,2} forma una componente fuerte, y {3} y {4} son independientes.)

Resuélvelo solo después de dominar el resto de este notebook.

**🧭 Antes de programar (sabiendo que este es un reto opcional):** en un grafo **no dirigido**, "componente conexa" se resuelve con un solo DFS/BFS por componente. Aquí el grafo es **dirigido** — ¿por qué un DFS ingenuo desde cada nodo no basta para agrupar correctamente? (Pista: en el ejemplo, desde el nodo 3 se llega a 4, pero no puedes volver de 4 a 3 — ¿son la misma componente fuerte?).
