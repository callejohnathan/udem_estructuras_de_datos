### Ejercicio 1

Una empresa de inspección industrial utiliza drones autónomos para revisar puntos críticos dentro de una planta de producción. Cada punto de inspección está representado como un nodo dentro de un grafo, y las conexiones entre nodos representan rutas válidas que el dron puede recorrer directamente.

Debido a restricciones de batería y tiempo, la empresa necesita determinar si existe una ruta que permita al dron visitar todos los puntos exactamente una vez, sin repetir ubicaciones, iniciando desde cualquier punto de la planta.

Tu tarea consiste en implementar un algoritmo que, dado un grafo G, determine si existe algún camino hamiltoniano. En caso de existir, el sistema debe devolver el camino encontrado como una lista ordenada de nodos. Si no existe una solución válida, debe devolver una lista vacía.

**🧭 Antes de programar:** ¿qué pregunta te hace el enunciado — alcanzabilidad, camino más corto, u otra cosa? "Visitar todos los puntos exactamente una vez" es un camino Hamiltoniano (revisa `teoria_grafos.ipynb`): a diferencia de Euler, no hay un atajo por grados, así que el único camino real es explorar con DFS + backtracking, probando extender el camino con nodos no repetidos y retrocediendo cuando te quedes sin salida.

Dibuja un grafo pequeño (5-6 nodos) de ejemplo y simula a mano el backtracking: ¿en qué nodo te atoras primero? ¿A qué nodo retrocedes, y qué alternativa pruebas después?

**🧭 Antes de programar:** "grupo de estaciones desconectadas del resto" es exactamente una componente conexa (grafo no dirigido) — ya sabes identificarlas con BFS/DFS repetido desde cada nodo no visitado (revisa el ejercicio de clase de `teoria_grafos.ipynb`). ### Ejercicio 3

Tu tarea consiste en implementar un algoritmo que, dado un grafo ponderado G, un entero K y un nodo inicial N, devuelva todos los nodos alcanzables desde N con costo acumulado ≤K, sin considerar caminos que generen ciclos. Se puede asumir que todos los pesos del grafo son positivos.

**🧭 Antes de programar:** el grafo es ponderado (costos de envío) — ¿dirigido o no dirigido? La pregunta es "qué nodos son alcanzables con costo acumulado ≤ K" — no es un BFS/DFS estándar (esos ignoran el peso acumulado): necesitas que la "distancia" que llevas por cada rama sea la suma de pesos recorridos, y descartar la rama en cuanto se pase de K.

Dibuja un grafo ponderado pequeño de 5 nodos con un K concreto, y simula a mano qué ramas exploras y cuáles descartas por exceso de costo, antes de programar.
